```python

import json

file_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\2018\1.json"

with open(file_path, "r", encoding="utf-8") as file:
    data = json.load(file)

print(data)
```

```python
import pandas as pd

transaction_data = data["data"]["transactionData"]

df = pd.json_normalize(
    transaction_data,
    sep="_"
)

df.head()
```

```python
print("Shape:", df.shape)
print("Columns:", df.columns)
print("Data Types:")
print(df.dtypes)
```

```python
print(df.head())
```

```python

df["transaction_count"] = df["paymentInstruments"].apply(
    lambda x: x[0]["count"]
)

print(df[["name", "transaction_count"]])
```

```python
print(df["paymentInstruments"].iloc[0])
```

```python
clean_df = df[["name", "transaction_count"]].copy()

clean_df = clean_df.rename(columns={
    "name": "payment_category"
})

print(clean_df)
```

```python
total_transactions = clean_df["transaction_count"].sum()

print("Total Transactions:", total_transactions)
```

```python
highest_category = clean_df.loc[
    clean_df["transaction_count"].idxmax()
]

print("Highest Transaction Category:", highest_category["payment_category"])
print("Transaction Count:", highest_category["transaction_count"])

```

```python
clean_df["percentage"] = (
    clean_df["transaction_count"]
    / clean_df["transaction_count"].sum()
) * 100

clean_df["percentage"] = clean_df["percentage"].round(2)

print(clean_df)
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.bar(
    clean_df["payment_category"],
    clean_df["transaction_count"]
)

plt.title("PhonePe Transaction Count by Category - 2018 Q1")
plt.xlabel("Payment Category")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
clean_df = clean_df.sort_values(
    by="transaction_count",
    ascending=False
)

print(clean_df)
```

```python
print(clean_df["transaction_count"].describe())
```

```python
lowest = clean_df.loc[
    clean_df["transaction_count"].idxmin()
]

highest = clean_df.loc[
    clean_df["transaction_count"].idxmax()
]

print("Lowest Category:", lowest["payment_category"])
print("Lowest Count:", lowest["transaction_count"])

print("Highest Category:", highest["payment_category"])
print("Highest Count:", highest["transaction_count"])
```

```python
retail = clean_df.loc[
    clean_df["payment_category"] == "Retail",
    "transaction_count"
].iloc[0]

utility = clean_df.loc[
    clean_df["payment_category"] == "Utility",
    "transaction_count"
].iloc[0]

p2p = clean_df.loc[
    clean_df["payment_category"] == "P2P",
    "transaction_count"
].iloc[0]

print("Retail vs Utility:", retail - utility)
print("Retail vs P2P:", retail - p2p)
print("P2P vs Utility:", p2p - utility)
```

```python
total = clean_df["transaction_count"].sum()

top_row = clean_df.loc[
    clean_df["transaction_count"].idxmax()
]

low_row = clean_df.loc[
    clean_df["transaction_count"].idxmin()
]

print("PHONEPE TRANSACTION SUMMARY - 2018 Q1")
print("--------------------------------------")
print("Total Transactions:", total)
print("Top Category:", top_row["payment_category"])
print("Top Category Count:", top_row["transaction_count"])
print("Lowest Category:", low_row["payment_category"])
print("Lowest Category Count:", low_row["transaction_count"])
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(7, 7), facecolor="#0B1020")

plt.pie(
    clean_df["transaction_count"],
    labels=clean_df["payment_category"],
    colors=["#7040A0", "#719C35", "#2878B5"],
    autopct="%1.1f%%",
    startangle=90,
    textprops={"color": "white", "fontsize": 12},
    wedgeprops={"edgecolor": "#0B1020", "linewidth": 2}
)

plt.title(
    "PhonePe Payment Category Share - 2018 Q1",
    color="white",
    fontsize=14,
    pad=20
)

plt.legend(
    clean_df["payment_category"],
    loc="upper left",
    bbox_to_anchor=(0.85, 1),
    labelcolor="white",
    facecolor="#151D30",
    edgecolor="#35415A"
)

plt.axis("equal")
plt.show()
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5), facecolor="#0B1020")

ax = plt.gca()
ax.set_facecolor("#0B1020")

colors = {
    "Retail": "#7040A0",
    "Utility": "#719C35",
    "P2P": "#2878B5"
}

bar_colors = [
    colors[category]
    for category in clean_df["payment_category"]
]

bars = ax.barh(
    clean_df["payment_category"],
    clean_df["transaction_count"],
    color=bar_colors
)

ax.set_title(
    "PhonePe Transactions by Category - 2018 Q1",
    color="white",
    fontsize=14
)
ax.set_xlabel("Transaction Count", color="white")
ax.set_ylabel("Payment Category", color="white")

ax.tick_params(colors="white")
ax.spines["bottom"].set_color("#526078")
ax.spines["left"].set_color("#526078")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(axis="x", alpha=0.2)

for bar in bars:
    value = int(bar.get_width())
    ax.text(
        value,
        bar.get_y() + bar.get_height() / 2,
        f" {value:,}",
        color="white",
        va="center",
        fontsize=9
    )

plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt

quarters = ["Q1", "Q2", "Q3", "Q4"]

totals = [
    clean_df["transaction_count"].sum(),
    q2_clean_df["transaction_count"].sum(),
    q3_clean_df["transaction_count"].sum(),
    q4_clean_df["transaction_count"].sum()
]

plt.figure(figsize=(9, 5), facecolor="#0B1020")
ax = plt.gca()
ax.set_facecolor("#0B1020")

ax.plot(
    quarters,
    totals,
    marker="o",
    linewidth=3,
    markersize=8,
    color="#38BDF8"
)

ax.set_title(
    "PhonePe Quarterly Transaction Trend - 2018",
    color="white",
    fontsize=14
)
ax.set_xlabel("Quarter", color="white")
ax.set_ylabel("Total Transactions", color="white")
ax.tick_params(colors="white")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(alpha=0.2)

for i, total in enumerate(totals):
    ax.annotate(
        f"{total:,}",
        (quarters[i], total),
        xytext=(0, 10),
        textcoords="offset points",
        ha="center",
        color="white"
    )

plt.tight_layout()
plt.show()
```

```python

quarter_summary = pd.DataFrame({
    "Quarter": ["Q1", "Q2", "Q3", "Q4"],
    "Total_Transactions": totals
})

print(quarter_summary)
```

```python

quarter_summary["Growth_Percentage"] = (
    quarter_summary["Total_Transactions"].pct_change() * 100
).round(2)

print(quarter_summary)
```

```python

growth_df = quarter_summary.dropna(
    subset=["Growth_Percentage"]
)

plt.figure(figsize=(9, 5), facecolor="#0B1020")
ax = plt.gca()
ax.set_facecolor("#0B1020")

ax.bar(
    growth_df["Quarter"],
    growth_df["Growth_Percentage"],
    color="#7040A0"
)

ax.set_title(
    "PhonePe Quarterly Transaction Growth - 2018",
    color="white"
)
ax.set_xlabel("Quarter", color="white")
ax.set_ylabel("Growth Percentage (%)", color="white")
ax.tick_params(colors="white")

ax.axhline(0, color="white", linewidth=0.8)
ax.grid(axis="y", alpha=0.2)

for i, value in enumerate(growth_df["Growth_Percentage"]):
    ax.text(
        i,
        value,
        f"{value:.2f}%",
        ha="center",
        va="bottom" if value >= 0 else "top",
        color="white"
    )

plt.tight_layout()
plt.show()
```

```python

best_quarter = quarter_summary.loc[
    quarter_summary["Total_Transactions"].idxmax()
]

lowest_quarter = quarter_summary.loc[
    quarter_summary["Total_Transactions"].idxmin()
]

print("PHONEPE QUARTERLY ANALYSIS")
print("--------------------------")

print("Highest Quarter:", best_quarter["Quarter"])
print("Transactions:", f'{best_quarter["Total_Transactions"]:,}')

print("Lowest Quarter:", lowest_quarter["Quarter"])
print("Transactions:", f'{lowest_quarter["Total_Transactions"]:,}')
```

```python

q1_master = clean_df[
    ["payment_category", "transaction_count"]
].copy()
q1_master["quarter"] = "Q1"

q2_master = q2_clean_df[
    ["payment_category", "transaction_count"]
].copy()
q2_master["quarter"] = "Q2"

q3_master = q3_clean_df[
    ["payment_category", "transaction_count"]
].copy()
q3_master["quarter"] = "Q3"

q4_master = q4_clean_df[
    ["payment_category", "transaction_count"]
].copy()
q4_master["quarter"] = "Q4"

master_df = pd.concat(
    [q1_master, q2_master, q3_master, q4_master],
    ignore_index=True
)

master_df["year"] = 2018

print(master_df)
```

```python

master_df.to_csv(
    "phonepe_2018_master.csv",
    index=False
)

print("Master CSV saved successfully!")
print("Total Rows:", len(master_df))
print("Columns:", master_df.columns.tolist())
```

```python

quarter_comparison = master_df.pivot(
    index="quarter",
    columns="payment_category",
    values="transaction_count"
)

quarter_comparison = quarter_comparison.reindex(
    ["Q1", "Q2", "Q3", "Q4"]
)

print(quarter_comparison)
```

```python

plt.figure(figsize=(10, 6), facecolor="#0B1020")
ax = plt.gca()
ax.set_facecolor("#0B1020")

category_colors = {
    "Retail": "#A78BFA",
    "P2P": "#38BDF8",
    "Utility": "#A3E635"
}

for category in quarter_comparison.columns:
    ax.plot(
        quarter_comparison.index,
        quarter_comparison[category],
        marker="o",
        linewidth=2.5,
        markersize=7,
        color=category_colors[category],
        label=category
    )

ax.set_title(
    "PhonePe Category-wise Quarterly Trend - 2018",
    color="white",
    fontsize=14
)
ax.set_xlabel("Quarter", color="white")
ax.set_ylabel("Transaction Count", color="white")
ax.tick_params(colors="white")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(alpha=0.2)
ax.legend(facecolor="#151D30", labelcolor="white")

plt.tight_layout()
plt.show()
```

```python

import json
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load JSON file
file_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\2024\1.json"

with open(file_path, "r", encoding="utf-8") as file:
    data = json.load(file)

# 2. Extract transaction category and count
records = []

for item in data["data"]["transactionData"]:
    category = item["name"]
    count = item["paymentInstruments"][0]["count"]

    records.append({
        "payment_category": category,
        "transaction_count": count
    })

# 3. Create clean DataFrame
q1_2024_clean_df = pd.DataFrame(records)

q1_2024_total = q1_2024_clean_df["transaction_count"].sum()

q1_2024_clean_df["percentage"] = (
    q1_2024_clean_df["transaction_count"] / q1_2024_total * 100
).round(2)

# 4. Display results
print("PHONEPE TRANSACTION ANALYSIS — 2024 Q1")
print("-" * 45)
print("Total Transactions:", f"{q1_2024_total:,}")
display(q1_2024_clean_df)

# 5. Professional dark-theme chart
colors = {
    "Utility": "#8B5CF6",
    "Retail": "#60A5FA",
    "P2P": "#B7E4C7"
}

bar_colors = [
    colors.get(category, "#F59E0B")
    for category in q1_2024_clean_df["payment_category"]
]

plt.figure(figsize=(11, 6), facecolor="#0B0F19")
ax = plt.gca()
ax.set_facecolor("#0B0F19")

bars = ax.barh(
    q1_2024_clean_df["payment_category"],
    q1_2024_clean_df["transaction_count"],
    color=bar_colors,
    edgecolor="white",
    linewidth=0.5
)

ax.invert_yaxis()
ax.set_title(
    "PhonePe Transaction Categories — 2024 Q1",
    color="white",
    fontsize=16,
    fontweight="bold",
    pad=18
)
ax.set_xlabel("Transaction Count", color="white")
ax.set_ylabel("Payment Category", color="white")
ax.tick_params(colors="white")

for bar in bars:
    value = int(bar.get_width())
    ax.text(
        value,
        bar.get_y() + bar.get_height() / 2,
        f"  {value:,}",
        color="white",
        va="center",
        fontsize=10
    )

ax.spines["bottom"].set_color("#475569")
ax.spines["left"].set_color("#475569")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(axis="x", linestyle="--", alpha=0.2, color="white")
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()

# 6. Save clean CSV
q1_2024_clean_df.to_csv("phonepe_2024_Q1.csv", index=False)

print("CSV saved: phonepe_2024_Q1.csv")
```

```python

import json
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load 2025 Q1 JSON file
file_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\2025\1.json"

with open(file_path, "r", encoding="utf-8") as file:
    data = json.load(file)

# 2. Extract transaction categories and counts
records = []

for item in data["data"]["transactionData"]:
    category = item["name"]
    count = item["paymentInstruments"][0]["count"]

    records.append({
        "payment_category": category,
        "transaction_count": count
    })

# 3. Create clean DataFrame
q1_2025_clean_df = pd.DataFrame(records)

q1_2025_total = q1_2025_clean_df["transaction_count"].sum()

q1_2025_clean_df["percentage"] = (
    q1_2025_clean_df["transaction_count"] / q1_2025_total * 100
).round(2)

# 4. Display results
print("PHONEPE TRANSACTION ANALYSIS — 2025 Q1")
print("-" * 45)
print("Total Transactions:", f"{q1_2025_total:,}")
display(q1_2025_clean_df)

# 5. Professional dark-theme chart
colors = {
    "Utility": "#8B5CF6",
    "Retail": "#60A5FA",
    "P2P": "#B7E4C7"
}

bar_colors = [
    colors.get(category, "#F59E0B")
    for category in q1_2025_clean_df["payment_category"]
]

plt.figure(figsize=(11, 6), facecolor="#0B0F19")
ax = plt.gca()
ax.set_facecolor("#0B0F19")

bars = ax.barh(
    q1_2025_clean_df["payment_category"],
    q1_2025_clean_df["transaction_count"],
    color=bar_colors,
    edgecolor="white",
    linewidth=0.5
)

ax.invert_yaxis()
ax.set_title(
    "PhonePe Transaction Categories — 2025 Q1",
    color="white",
    fontsize=16,
    fontweight="bold",
    pad=18
)
ax.set_xlabel("Transaction Count", color="white")
ax.set_ylabel("Payment Category", color="white")
ax.tick_params(colors="white")

for bar in bars:
    value = int(bar.get_width())
    ax.text(
        value,
        bar.get_y() + bar.get_height() / 2,
        f"  {value:,}",
        color="white",
        va="center",
        fontsize=10
    )

ax.spines["bottom"].set_color("#475569")
ax.spines["left"].set_color("#475569")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(axis="x", linestyle="--", alpha=0.2, color="white")
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()

# 6. Save CSV
q1_2025_clean_df.to_csv("phonepe_2025_Q1.csv", index=False)

print("CSV saved: phonepe_2025_Q1.csv")
```

```python

import pandas as pd

# 1. Combine available quarters: Q1 and Q2
master_2026_df = pd.concat(
    [
        q1_2026_clean_df,
        q2_2026_clean_df
    ],
    ignore_index=True
)

# 2. Add year and quarter columns
master_2026_df["year"] = 2026

master_2026_df["quarter"] = (
    ["Q1"] * len(q1_2026_clean_df)
    + ["Q2"] * len(q2_2026_clean_df)
)

# 3. Arrange columns
master_2026_df = master_2026_df[
    [
        "year",
        "quarter",
        "payment_category",
        "transaction_count",
        "percentage"
    ]
]

# 4. Display master dataset
print("PHONEPE 2026 MASTER DATASET")
print("=" * 40)

display(master_2026_df)

print("Available quarters: Q1, Q2")
print("Total rows:", len(master_2026_df))
print(
    "Total transactions in available quarters:",
    f"{master_2026_df['transaction_count'].sum():,}"
)

# 5. Save master CSV
master_2026_df.to_csv("phonepe_2026_master.csv", index=False)

print("\nSaved: phonepe_2026_master.csv")
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# 1. Create category trend table
category_trend_2026 = master_2026_df.pivot_table(
    index="payment_category",
    columns="quarter",
    values="transaction_count",
    aggfunc="sum",
    fill_value=0
)

category_trend_2026 = category_trend_2026.reindex(
    columns=["Q1", "Q2"],
    fill_value=0
)

print("PHONEPE 2026 CATEGORY TREND")
display(category_trend_2026)

# 2. Professional dark-theme chart
colors = {
    "Utility": "#8B5CF6",
    "Retail": "#60A5FA",
    "P2P": "#B7E4C7"
}

plt.figure(figsize=(11, 6), facecolor="#0B0F19")
ax = plt.gca()
ax.set_facecolor("#0B0F19")

for category in category_trend_2026.index:
    ax.plot(
        category_trend_2026.columns,
        category_trend_2026.loc[category],
        marker="o",
        markersize=8,
        linewidth=2.5,
        label=category,
        color=colors.get(category, "#F59E0B")
    )

ax.set_title(
    "PhonePe Category Transaction Trends — 2026",
    color="white",
    fontsize=15,
    fontweight="bold",
    pad=18
)
ax.set_xlabel("Quarter", color="white")
ax.set_ylabel("Transaction Count", color="white")
ax.tick_params(colors="white")
ax.legend(facecolor="#111827", labelcolor="white")
ax.grid(True, linestyle="--", alpha=0.2, color="white")

for spine in ax.spines.values():
    spine.set_color("#475569")

plt.tight_layout()
plt.show()

# 3. Save CSV
category_trend_2026.to_csv(
    "phonepe_2026_category_trend.csv"
)

print("Saved: phonepe_2026_category_trend.csv")
```

```python

import os
import pandas as pd

# Available yearly master CSV files
years = range(2018, 2027)
all_years = []

for year in years:
    file_name = f"phonepe_{year}_master.csv"

    if os.path.exists(file_name):
        df = pd.read_csv(file_name)
        all_years.append(df)
        print(f"Loaded: {file_name} | Rows: {len(df)}")
    else:
        print(f"Not found: {file_name}")

# Combine available years
if all_years:
    india_master_df = pd.concat(all_years, ignore_index=True)

    # Sort for easier analysis
    india_master_df = india_master_df.sort_values(
        ["year", "quarter", "payment_category"]
    ).reset_index(drop=True)

    print("\nINDIA OVERALL TRANSACTION DATA")
    display(india_master_df.head(10))

    print("\nYears available:")
    print(sorted(india_master_df["year"].unique()))

    print("\nTotal rows:", len(india_master_df))

    india_master_df.to_csv(
        "phonepe_india_master_2018_2026.csv",
        index=False
    )

    print("\nSaved: phonepe_india_master_2018_2026.csv")
else:
    print("Master CSV files current folder-la kidaikkala.")
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# STEP 186: INDIA OVERALL KEY INSIGHTS

# 1. Total transactions
india_total_transactions = india_master_df["transaction_count"].sum()

# 2. Highest transaction year
yearly_totals = (
    india_master_df.groupby("year")["transaction_count"]
    .sum()
    .sort_values(ascending=False)
)

highest_year = yearly_totals.idxmax()
highest_year_transactions = yearly_totals.max()

# 3. Category totals
category_totals = (
    india_master_df.groupby("payment_category")["transaction_count"]
    .sum()
    .sort_values(ascending=False)
)

top_category = category_totals.index[0]
top_category_transactions = category_totals.iloc[0]

# 4. Category percentage
top_category_percentage = (
    top_category_transactions / india_total_transactions * 100
)

# 5. Display key insights
print("=" * 55)
print("       PHONEPE INDIA — KEY INSIGHTS")
print("=" * 55)

print("\nTotal transactions in available data:")
print(f"{india_total_transactions:,}")

print("\nHighest transaction year in available data:")
print(highest_year)
print(f"Transactions: {highest_year_transactions:,.0f}")

print("\nMost-used payment category:")
print(top_category)
print(f"Transactions: {top_category_transactions:,.0f}")
print(f"Percentage share: {top_category_percentage:.2f}%")

print("\nYear-wise totals:")
display(yearly_totals.rename("total_transactions").reset_index())

print("\nCategory-wise totals:")
display(category_totals.rename("total_transactions").reset_index())

# 6. Save key insights
insights_df = pd.DataFrame({
    "metric": [
        "Total Transactions",
        "Highest Transaction Year",
        "Highest Year Transactions",
        "Top Payment Category",
        "Top Category Transactions",
        "Top Category Percentage"
    ],
    "value": [
        int(india_total_transactions),
        int(highest_year),
        int(highest_year_transactions),
        top_category,
        int(top_category_transactions),
        round(top_category_percentage, 2)
    ]
})

insights_df.to_csv(
    "phonepe_india_key_insights_2018_2026.csv",
    index=False
)

print("\nSaved: phonepe_india_key_insights_2018_2026.csv")
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# 1. Calculate quarter-wise totals
quarter_totals = (
    andaman_2018_master_df.groupby("quarter")["transaction_count"]
    .sum()
    .reindex(["Q1", "Q2", "Q3", "Q4"])
    .reset_index()
)

# 2. Dark theme
plt.figure(figsize=(10, 6), facecolor="#0B0F19")
ax = plt.gca()
ax.set_facecolor("#0B0F19")

bars = ax.bar(
    quarter_totals["quarter"],
    quarter_totals["transaction_count"],
    color=["#8B5CF6", "#60A5FA", "#B7E4C7", "#F59E0B"],
    edgecolor="white",
    linewidth=0.6,
    width=0.55
)

# 3. Chart title and labels
ax.set_title(
    "Andaman & Nicobar Islands — Quarterly Transactions (2018)",
    color="white",
    fontsize=14,
    fontweight="bold",
    pad=18
)

ax.set_xlabel("Quarter", color="white")
ax.set_ylabel("Transaction Count", color="white")
ax.tick_params(colors="white")

# 4. Show transaction values
for bar in bars:
    value = int(bar.get_height())
    ax.annotate(
        f"{value:,}",
        (bar.get_x() + bar.get_width() / 2, bar.get_height()),
        xytext=(0, 6),
        textcoords="offset points",
        ha="center",
        color="white",
        fontsize=10
    )

ax.spines["bottom"].set_color("#475569")
ax.spines["left"].set_color("#475569")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(axis="y", linestyle="--", alpha=0.2, color="white")
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()

# 5. Save quarterly comparison
quarter_totals.to_csv(
    "andaman_nicobar_2018_quarterly_comparison.csv",
    index=False
)

print("Chart and quarterly comparison completed!")
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# 1. Category-wise transaction totals for 2018
category_totals_2018 = (
    andaman_2018_master_df.groupby("payment_category")["transaction_count"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

category_totals_2018.columns = [
    "payment_category",
    "total_transactions"
]

# 2. Calculate percentage share
grand_total = category_totals_2018["total_transactions"].sum()

category_totals_2018["percentage"] = (
    category_totals_2018["total_transactions"] / grand_total * 100
).round(2)

print("ANDAMAN & NICOBAR ISLANDS — 2018 CATEGORY ANALYSIS")
display(category_totals_2018)

# 3. Professional dark-theme chart
colors = {
    "Utility": "#8B5CF6",
    "Retail": "#60A5FA",
    "P2P": "#B7E4C7"
}

bar_colors = [
    colors.get(category, "#F59E0B")
    for category in category_totals_2018["payment_category"]
]

plt.figure(figsize=(10, 6), facecolor="#0B0F19")
ax = plt.gca()
ax.set_facecolor("#0B0F19")

bars = ax.barh(
    category_totals_2018["payment_category"],
    category_totals_2018["total_transactions"],
    color=bar_colors,
    edgecolor="white",
    linewidth=0.5
)

ax.invert_yaxis()
ax.set_title(
    "Andaman & Nicobar Islands — Category Analysis (2018)",
    color="white",
    fontsize=14,
    fontweight="bold",
    pad=18
)
ax.set_xlabel("Total Transactions", color="white")
ax.set_ylabel("Payment Category", color="white")
ax.tick_params(colors="white")

for bar in bars:
    ax.text(
        bar.get_width(),
        bar.get_y() + bar.get_height() / 2,
        f"  {int(bar.get_width()):,}",
        color="white",
        va="center",
        fontsize=10
    )

ax.spines["bottom"].set_color("#475569")
ax.spines["left"].set_color("#475569")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(axis="x", linestyle="--", alpha=0.2, color="white")
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()

# 4. Display highest category
top_category = category_totals_2018.iloc[0]

print("\nKEY INSIGHT")
print("Highest category:", top_category["payment_category"])
print("Transactions:", f"{int(top_category['total_transactions']):,}")
print("Percentage:", f"{top_category['percentage']}%")

# 5. Save CSV
category_totals_2018.to_csv(
    "andaman_nicobar_2018_category_analysis.csv",
    index=False
)

print("\nSaved: andaman_nicobar_2018_category_analysis.csv")
```

```python

import os
import json
import pandas as pd
import matplotlib.pyplot as plt

# STEP 1: LOAD ALL FOUR QUARTERS
base_path_2024 = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state\andhra-pradesh\2024"

records_2024 = []

for q in range(1, 5):
    file_path = os.path.join(base_path_2024, f"{q}.json")

    if os.path.exists(file_path):
        with open(file_path, "r", encoding="utf-8") as file:
            data = json.load(file)

        for item in data.get("data", {}).get("transactionData", []):
            category = item.get("name", "Unknown")
            count = sum(
                instrument.get("count", 0)
                for instrument in item.get("paymentInstruments", [])
            )

            records_2024.append({
                "Year": 2024,
                "Quarter": f"Q{q}",
                "Category": category,
                "Transaction_Count": count
            })
    else:
        print(f"Missing file: {file_path} — skipped")

andhra_2024_master_df = pd.DataFrame(records_2024)

if andhra_2024_master_df.empty:
    raise ValueError("2024 data load aagala. Folder path and JSON files check pannu.")

print("STEP 1: Master DataFrame")
display(andhra_2024_master_df)


# STEP 2: QUARTER-WISE CATEGORY COMPARISON
quarter_comparison_2024 = andhra_2024_master_df.pivot_table(
    index="Category",
    columns="Quarter",
    values="Transaction_Count",
    aggfunc="sum",
    fill_value=0
)

quarter_order_2024 = ["Q1", "Q2", "Q3", "Q4"]
quarter_comparison_2024 = quarter_comparison_2024.reindex(
    columns=[q for q in quarter_order_2024
             if q in quarter_comparison_2024.columns]
)

print("STEP 2: Quarter-wise Category Comparison")
display(quarter_comparison_2024)


# STEP 3: QUARTER-WISE CATEGORY COMPARISON CHART
quarter_colors_2024 = {
    "Q1": "#60A5FA",
    "Q2": "#34D399",
    "Q3": "#A78BFA",
    "Q4": "#F59E0B"
}

ax = quarter_comparison_2024.plot(
    kind="bar",
    figsize=(13, 6),
    color=[quarter_colors_2024[q]
           for q in quarter_comparison_2024.columns],
    width=0.8
)

ax.set_title(
    "Andhra Pradesh PhonePe Transactions by Category — 2024",
    color="white",
    fontsize=14
)
ax.set_xlabel("Payment Category", color="white")
ax.set_ylabel("Transaction Count", color="white")
ax.tick_params(axis="x", colors="white", rotation=35)
ax.tick_params(axis="y", colors="white")
ax.set_facecolor("#0B0F19")
ax.figure.set_facecolor("#0B0F19")
ax.legend(title="Quarter", labelcolor="white")
ax.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()


# STEP 4: QUARTER-WISE TOTALS AND HIGHEST QUARTER
quarter_totals_2024 = (
    andhra_2024_master_df.groupby("Quarter")["Transaction_Count"]
    .sum()
    .reindex([q for q in quarter_order_2024
              if q in andhra_2024_master_df["Quarter"].unique()])
)

print("STEP 4: Quarter-wise Transaction Totals")
display(quarter_totals_2024.to_frame(name="Total_Transactions"))

highest_quarter_2024 = quarter_totals_2024.idxmax()
highest_quarter_total_2024 = quarter_totals_2024.max()

print("Highest Transaction Quarter:", highest_quarter_2024)
print("Transaction Count:", f"{highest_quarter_total_2024:,}")


# STEP 5: QUARTER TOTALS CHART
quarter_bar_colors_2024 = [
    quarter_colors_2024[q] for q in quarter_totals_2024.index
]

ax = quarter_totals_2024.plot(
    kind="bar",
    figsize=(9, 5),
    color=quarter_bar_colors_2024
)

ax.set_title(
    "Total Transactions by Quarter — 2024",
    color="white"
)
ax.set_xlabel("Quarter", color="white")
ax.set_ylabel("Transaction Count", color="white")
ax.tick_params(axis="x", colors="white", rotation=0)
ax.tick_params(axis="y", colors="white")
ax.set_facecolor("#0B0F19")
ax.figure.set_facecolor("#0B0F19")
ax.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()


# STEP 6: PAYMENT CATEGORY TOTALS AND HIGHEST CATEGORY
category_totals_2024 = (
    andhra_2024_master_df.groupby("Category")["Transaction_Count"]
    .sum()
    .sort_values(ascending=False)
)

print("STEP 6: Category-wise Transaction Totals")
display(category_totals_2024.to_frame(name="Total_Transactions"))

highest_category_2024 = category_totals_2024.idxmax()
highest_category_total_2024 = category_totals_2024.max()

print("Most Used Payment Category:", highest_category_2024)
print("Transaction Count:", f"{highest_category_total_2024:,}")


# STEP 7: PAYMENT CATEGORY CHART
ax = category_totals_2024.plot(
    kind="bar",
    figsize=(11, 5),
    color="#60A5FA"
)

ax.set_title(
    "Payment Category Analysis — Andhra Pradesh 2024",
    color="white"
)
ax.set_xlabel("Payment Category", color="white")
ax.set_ylabel("Transaction Count", color="white")
ax.tick_params(axis="x", colors="white", rotation=35)
ax.tick_params(axis="y", colors="white")
ax.set_facecolor("#0B0F19")
ax.figure.set_facecolor("#0B0F19")
ax.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()


# STEP 8: SAVE DATASET AS CSV
csv_path_2024 = os.path.join(
    os.getcwd(),
    "andhra_pradesh_transactions_2024.csv"
)

andhra_2024_master_df.to_csv(csv_path_2024, index=False)

print("STEP 8: CSV saved successfully")
print("Saved file:", csv_path_2024)
print("Total rows:", len(andhra_2024_master_df))
print("Total transactions:", f"{andhra_2024_master_df['Transaction_Count'].sum():,}")
```

```python

import os
import json
import pandas as pd

# RELOAD MISSING YEARS: 2018 AND 2023

base_folder = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state\andhra-pradesh"

for year in [2018, 2023]:
    records = []
    year_path = os.path.join(base_folder, str(year))

    for q in range(1, 5):
        file_path = os.path.join(year_path, f"{q}.json")

        if os.path.exists(file_path):
            with open(file_path, "r", encoding="utf-8") as file:
                data = json.load(file)

            for item in data.get("data", {}).get("transactionData", []):
                count = sum(
                    instrument.get("count", 0)
                    for instrument in item.get("paymentInstruments", [])
                )

                records.append({
                    "Year": year,
                    "Quarter": f"Q{q}",
                    "Category": item.get("name", "Unknown"),
                    "Transaction_Count": count
                })
        else:
            print(f"{year} Q{q}: file missing, skipped")

    if records:
        globals()[f"andhra_{year}_master_df"] = pd.DataFrame(records)
        print(f"\n{year} loaded successfully")
        display(globals()[f"andhra_{year}_master_df"].head())
    else:
        print(f"\n{year}: data load aagala. Folder path check pannu.")
```

```python

# ANDHRA PRADESH: 2018–2026 YEAR-WISE COMPARISON

year_dfs = [
    globals()[f"andhra_{year}_master_df"]
    for year in range(2018, 2027)
]

all_years_df = pd.concat(year_dfs, ignore_index=True)

yearly_comparison = (
    all_years_df.groupby("Year")["Transaction_Count"]
    .sum()
    .sort_index()
)

print("YEAR-WISE TRANSACTION COMPARISON")
display(yearly_comparison.to_frame(name="Total_Transactions"))

# BEST PERFORMING FULL YEAR (2018–2025)
full_years = yearly_comparison.loc[
    (yearly_comparison.index >= 2018) &
    (yearly_comparison.index <= 2025)
]

best_year = full_years.idxmax()
best_total = full_years.max()

print("\nBEST PERFORMING FULL YEAR:", best_year)
print("TOTAL TRANSACTIONS:", f"{best_total:,}")

# YEAR-WISE COMPARISON CHART
colors = [
    "#A78BFA" if year == 2026 else
    "#F59E0B" if year == best_year else
    "#60A5FA"
    for year in yearly_comparison.index
]

ax = yearly_comparison.plot(
    kind="bar",
    figsize=(12, 6),
    color=colors
)

ax.set_title(
    "Andhra Pradesh PhonePe — Year-wise Transaction Comparison",
    color="white",
    fontsize=14
)
ax.set_xlabel("Year", color="white")
ax.set_ylabel("Total Transactions", color="white")
ax.tick_params(axis="x", colors="white", rotation=0)
ax.tick_params(axis="y", colors="white")
ax.set_facecolor("#0B0F19")
ax.figure.set_facecolor("#0B0F19")
ax.grid(axis="y", alpha=0.2)

plt.tight_layout()
plt.show()
```

```python

# ANDHRA PRADESH: FINAL INSIGHTS AND YEAR-WISE GROWTH

# STEP 1: YEAR-WISE TOTALS
yearly_comparison = (
    all_years_df.groupby("Year")["Transaction_Count"]
    .sum()
    .sort_index()
)

print("YEAR-WISE TOTAL TRANSACTIONS")
display(yearly_comparison.to_frame(name="Total_Transactions"))


# STEP 2: FULL-YEAR GROWTH (2019–2025)
full_year_totals = yearly_comparison.loc[2018:2025]

growth_df = full_year_totals.to_frame(
    name="Total_Transactions"
)

growth_df["Previous_Year_Transactions"] = (
    growth_df["Total_Transactions"].shift(1)
)

growth_df["Growth_Percentage"] = (
    (
        growth_df["Total_Transactions"]
        - growth_df["Previous_Year_Transactions"]
    )
    / growth_df["Previous_Year_Transactions"]
    * 100
).round(2)

print("YEAR-WISE GROWTH ANALYSIS")
display(growth_df)


# STEP 3: HIGHEST FULL-YEAR TRANSACTIONS
best_year = full_year_totals.idxmax()
best_total = full_year_totals.max()

print("Best Performing Full Year:", best_year)
print("Total Transactions:", f"{best_total:,}")


# STEP 4: HIGHEST GROWTH YEAR
valid_growth = growth_df["Growth_Percentage"].dropna()

if not valid_growth.empty:
    highest_growth_year = valid_growth.idxmax()
    highest_growth = valid_growth.max()

    print("Highest Growth Year:", highest_growth_year)
    print("Growth Percentage:", f"{highest_growth}%")
```

```python

import os
import json
import pandas as pd
import matplotlib.pyplot as plt

# STEP 1: LOAD ALL FOUR QUARTERS

base_path_2018 = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state\arunachal-pradesh\2018"

records_2018 = []

for q in range(1, 5):
    file_path = os.path.join(base_path_2018, f"{q}.json")

    if os.path.exists(file_path):
        with open(file_path, "r", encoding="utf-8") as file:
            data = json.load(file)

        for item in data.get("data", {}).get("transactionData", []):
            category = item.get("name", "Unknown")
            count = sum(
                instrument.get("count", 0)
                for instrument in item.get("paymentInstruments", [])
            )

            records_2018.append({
                "State": "Arunachal Pradesh",
                "Year": 2018,
                "Quarter": f"Q{q}",
                "Category": category,
                "Transaction_Count": count
            })
    else:
        print(f"Q{q} file missing — skipped")

arunachal_2018_master_df = pd.DataFrame(records_2018)

if arunachal_2018_master_df.empty:
    raise ValueError("Data load aagala. Folder path and JSON files check pannu.")

print("STEP 1: Master DataFrame")
display(arunachal_2018_master_df)


# STEP 2: QUARTER-WISE CATEGORY COMPARISON

quarter_order_2018 = ["Q1", "Q2", "Q3", "Q4"]

quarter_comparison_2018 = arunachal_2018_master_df.pivot_table(
    index="Category",
    columns="Quarter",
    values="Transaction_Count",
    aggfunc="sum",
    fill_value=0
)

quarter_comparison_2018 = quarter_comparison_2018.reindex(
    columns=[
        q for q in quarter_order_2018
        if q in quarter_comparison_2018.columns
    ]
)

print("STEP 2: Quarter-wise Category Comparison")
display(quarter_comparison_2018)


# STEP 3: QUARTER-WISE CATEGORY COMPARISON CHART

quarter_colors_2018 = {
    "Q1": "#60A5FA",
    "Q2": "#34D399",
    "Q3": "#A78BFA",
    "Q4": "#F59E0B"
}

ax = quarter_comparison_2018.plot(
    kind="bar",
    figsize=(13, 6),
    color=[
        quarter_colors_2018[q]
        for q in quarter_comparison_2018.columns
    ],
    width=0.8
)

ax.set_title(
    "Arunachal Pradesh PhonePe Transactions by Category — 2018",
    color="white"
)
ax.set_xlabel("Payment Category", color="white")
ax.set_ylabel("Transaction Count", color="white")
ax.tick_params(axis="x", colors="white", rotation=35)
ax.tick_params(axis="y", colors="white")
ax.set_facecolor("#0B0F19")
ax.figure.set_facecolor("#0B0F19")
ax.legend(title="Quarter", labelcolor="white")
ax.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()


# STEP 4: QUARTER-WISE TOTALS AND HIGHEST QUARTER

quarter_totals_2018 = (
    arunachal_2018_master_df.groupby("Quarter")["Transaction_Count"]
    .sum()
    .reindex([
        q for q in quarter_order_2018
        if q in arunachal_2018_master_df["Quarter"].unique()
    ])
)

print("STEP 4: Quarter-wise Transaction Totals")
display(quarter_totals_2018.to_frame(name="Total_Transactions"))

highest_quarter_2018 = quarter_totals_2018.idxmax()
highest_quarter_total_2018 = quarter_totals_2018.max()

print("Highest Transaction Quarter:", highest_quarter_2018)
print("Transaction Count:", f"{highest_quarter_total_2018:,}")


# STEP 5: QUARTER TOTALS CHART

ax = quarter_totals_2018.plot(
    kind="bar",
    figsize=(9, 5),
    color=[
        quarter_colors_2018[q]
        for q in quarter_totals_2018.index
    ]
)

ax.set_title("Total Transactions by Quarter — 2018", color="white")
ax.set_xlabel("Quarter", color="white")
ax.set_ylabel("Transaction Count", color="white")
ax.tick_params(axis="x", colors="white", rotation=0)
ax.tick_params(axis="y", colors="white")
ax.set_facecolor("#0B0F19")
ax.figure.set_facecolor("#0B0F19")
ax.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()


# STEP 6: CATEGORY TOTALS AND HIGHEST CATEGORY

category_totals_2018 = (
    arunachal_2018_master_df.groupby("Category")["Transaction_Count"]
    .sum()
    .sort_values(ascending=False)
)

print("STEP 6: Category-wise Transaction Totals")
display(category_totals_2018.to_frame(name="Total_Transactions"))

highest_category_2018 = category_totals_2018.idxmax()
highest_category_total_2018 = category_totals_2018.max()

print("Most Used Payment Category:", highest_category_2018)
print("Transaction Count:", f"{highest_category_total_2018:,}")


# STEP 7: PAYMENT CATEGORY CHART

ax = category_totals_2018.plot(
    kind="bar",
    figsize=(11, 5),
    color="#60A5FA"
)

ax.set_title(
    "Payment Category Analysis — Arunachal Pradesh 2018",
    color="white"
)
ax.set_xlabel("Payment Category", color="white")
ax.set_ylabel("Transaction Count", color="white")
ax.tick_params(axis="x", colors="white", rotation=35)
ax.tick_params(axis="y", colors="white")
ax.set_facecolor("#0B0F19")
ax.figure.set_facecolor("#0B0F19")
ax.grid(axis="y", alpha=0.2)
plt.tight_layout()
plt.show()


# STEP 8: SAVE DATASET AS CSV

csv_path_2018 = os.path.join(
    os.getcwd(),
    "arunachal_pradesh_transactions_2018.csv"
)

arunachal_2018_master_df.to_csv(csv_path_2018, index=False)

print("STEP 8: CSV saved successfully")
print("Saved file:", csv_path_2018)
print("Total rows:", len(arunachal_2018_master_df))
print(
    "Total transactions:",
    f"{arunachal_2018_master_df['Transaction_Count'].sum():,}"
)
```

```python

import os
import json
import pandas as pd

base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state\arunachal-pradesh\2020"

all_data = []

for quarter in range(1, 5):
    file_path = os.path.join(base_path, f"{quarter}.json")

    if os.path.exists(file_path):
        with open(file_path, "r") as file:
            data = json.load(file)

        transaction_data = data["data"]["transactionData"]

        for category in transaction_data:
            total_count = sum(
                instrument["count"]
                for instrument in category["paymentInstruments"]
            )

            all_data.append({
                "Year": 2020,
                "Quarter": f"Q{quarter}",
                "Category": category["name"],
                "Transactions": total_count
            })

    else:
        print(f"Quarter {quarter} file not found")

arunachal_2020_master_df = pd.DataFrame(all_data)

print("MASTER DATA CREATED!")
print("Total rows:", len(arunachal_2020_master_df))

display(arunachal_2020_master_df)
```

```python

import os
import json
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("dark_background")

# 2021 dataset path
base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state\arunachal-pradesh\2021"

all_data = []

# 1. Load JSON files and create master DataFrame
for quarter in range(1, 5):
    file_path = os.path.join(base_path, f"{quarter}.json")

    if os.path.exists(file_path):
        with open(file_path, "r") as file:
            data = json.load(file)

        for category in data["data"]["transactionData"]:
            total_count = sum(
                instrument["count"]
                for instrument in category["paymentInstruments"]
            )

            all_data.append({
                "Year": 2021,
                "Quarter": f"Q{quarter}",
                "Category": category["name"],
                "Transactions": total_count
            })
    else:
        print(f"Q{quarter} file not found")

arunachal_2021_master_df = pd.DataFrame(all_data)

if arunachal_2021_master_df.empty:
    print("No data loaded. Check the 2021 folder path.")
else:
    print("MASTER DATA")
    display(arunachal_2021_master_df)

    # 2. Quarter-wise category comparison
    quarter_comparison_2021 = arunachal_2021_master_df.pivot_table(
        index="Category",
        columns="Quarter",
        values="Transactions",
        aggfunc="sum",
        fill_value=0
    )

    print("QUARTER-WISE CATEGORY COMPARISON")
    display(quarter_comparison_2021)

    # 3. Category comparison chart
    quarter_colors = {
        "Q1": "#60A5FA",
        "Q2": "#34D399",
        "Q3": "#A78BFA",
        "Q4": "#F59E0B"
    }

    quarter_comparison_2021.plot(
        kind="bar",
        figsize=(12, 6),
        color=[
            quarter_colors[q]
            for q in quarter_comparison_2021.columns
        ]
    )

    plt.title("Arunachal Pradesh - 2021 Category Comparison")
    plt.xlabel("Payment Category")
    plt.ylabel("Transactions")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    # 4. Quarter totals and highest quarter
    quarter_totals_2021 = arunachal_2021_master_df.groupby(
        "Quarter"
    )["Transactions"].sum()

    print("QUARTER TOTALS")
    display(quarter_totals_2021)

    print(
        "Highest Quarter:",
        quarter_totals_2021.idxmax(),
        "-",
        f"{quarter_totals_2021.max():,} transactions"
    )

    # 5. Quarter totals chart
    quarter_totals_2021.plot(
        kind="bar",
        figsize=(8, 5),
        color=[
            quarter_colors[q]
            for q in quarter_totals_2021.index
        ]
    )

    plt.title("Arunachal Pradesh - 2021 Quarter Totals")
    plt.xlabel("Quarter")
    plt.ylabel("Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 6. Category totals and highest category
    category_totals_2021 = arunachal_2021_master_df.groupby(
        "Category"
    )["Transactions"].sum().sort_values(ascending=False)

    print("CATEGORY TOTALS")
    display(category_totals_2021)

    print(
        "Highest Payment Category:",
        category_totals_2021.idxmax(),
        "-",
        f"{category_totals_2021.max():,} transactions"
    )

    # 7. Payment category chart - different colours
    category_colors = [
        "#F472B6", "#22D3EE", "#FBBF24", "#A3E635",
        "#C084FC", "#FB7185", "#2DD4BF", "#818CF8"
    ]

    plt.figure(figsize=(11, 6))

    plt.bar(
        category_totals_2021.index,
        category_totals_2021.values,
        color=[
            category_colors[i % len(category_colors)]
            for i in range(len(category_totals_2021))
        ]
    )

    plt.title("Arunachal Pradesh - 2021 Payment Category Analysis")
    plt.xlabel("Payment Category")
    plt.ylabel("Transactions")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    # 8. Save master and comparison CSV files
    master_file = "arunachal_pradesh_transactions_2021.csv"
    comparison_file = "arunachal_pradesh_2021_quarter_category_comparison.csv"
    quarter_file = "arunachal_pradesh_2021_quarter_totals.csv"
    category_file = "arunachal_pradesh_2021_category_totals.csv"

    arunachal_2021_master_df.to_csv(master_file, index=False)
    quarter_comparison_2021.to_csv(comparison_file)
    quarter_totals_2021.to_csv(quarter_file, header=["Transactions"])
    category_totals_2021.to_csv(category_file, header=["Transactions"])

    # 9. Verify saved files
    print("\nSAVE VERIFICATION")

    for filename in [
        master_file,
        comparison_file,
        quarter_file,
        category_file
    ]:
        print(filename, "->", os.path.exists(filename))

    verified_df = pd.read_csv(master_file)

    print("Original master rows:", len(arunachal_2021_master_df))
    print("Saved master rows:", len(verified_df))
    print(
        "Master verification:",
        len(arunachal_2021_master_df) == len(verified_df)
    )

    print("All available 2021 analysis completed!")
```

```python

import os
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("dark_background")

# 1. Load all yearly master CSV files
all_years_data = []

for year in range(2018, 2027):
    filename = f"arunachal_pradesh_transactions_{year}.csv"

    if os.path.exists(filename):
        year_df = pd.read_csv(filename)
        all_years_data.append(year_df)
        print(f"{year} loaded: {len(year_df)} rows")
    else:
        print(f"{year} CSV not found: {filename}")

# 2. Combine all years into one master DataFrame
if all_years_data:
    arunachal_all_years_master_df = pd.concat(
        all_years_data,
        ignore_index=True
    )

    print("\nALL YEARS MASTER DATA")
    display(arunachal_all_years_master_df.head(10))

    print("Total rows:", len(arunachal_all_years_master_df))
    print(
        "Years included:",
        sorted(arunachal_all_years_master_df["Year"].unique())
    )

    # 3. Year-wise total transactions
    yearly_totals = arunachal_all_years_master_df.groupby(
        "Year"
    )["Transactions"].sum().sort_index()

    print("\nYEAR-WISE TRANSACTION TOTALS")
    display(yearly_totals)

    # 4. Year-wise transaction chart
    yearly_totals.plot(
        kind="bar",
        figsize=(12, 6),
        color="#60A5FA"
    )

    plt.title("Arunachal Pradesh - Year-wise Transactions (2018–2026)")
    plt.xlabel("Year")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

    # 5. Category-wise totals across all years
    yearly_category_comparison = arunachal_all_years_master_df.pivot_table(
        index="Year",
        columns="Category",
        values="Transactions",
        aggfunc="sum",
        fill_value=0
    ).sort_index()

    print("\nYEAR-WISE CATEGORY COMPARISON")
    display(yearly_category_comparison)

    # 6. Category trend chart
    yearly_category_comparison.plot(
        kind="line",
        marker="o",
        figsize=(12, 6)
    )

    plt.title("Arunachal Pradesh - Payment Category Trends")
    plt.xlabel("Year")
    plt.ylabel("Number of Transactions")
    plt.xticks(yearly_category_comparison.index)
    plt.tight_layout()
    plt.show()

    # 7. Highest and lowest year in loaded data
    highest_year = yearly_totals.idxmax()
    lowest_year = yearly_totals.idxmin()

    print(
        f"Highest transaction year: {highest_year} "
        f"({yearly_totals.max():,} transactions)"
    )
    print(
        f"Lowest transaction year: {lowest_year} "
        f"({yearly_totals.min():,} transactions)"
    )

    # 8. Save combined master and comparison CSV files
    master_file = "arunachal_pradesh_all_years_master.csv"
    yearly_file = "arunachal_pradesh_yearly_totals_2018_2026.csv"
    comparison_file = "arunachal_pradesh_yearly_category_comparison.csv"

    arunachal_all_years_master_df.to_csv(master_file, index=False)
    yearly_totals.to_csv(yearly_file, header=["Transactions"])
    yearly_category_comparison.to_csv(comparison_file)

    # 9. Verify saved files
    print("\nSAVE VERIFICATION")

    for filename in [master_file, yearly_file, comparison_file]:
        print(filename, "->", os.path.exists(filename))

    verified_master = pd.read_csv(master_file)

    print("Original rows:", len(arunachal_all_years_master_df))
    print("Saved rows:", len(verified_master))
    print(
        "Master verification:",
        len(arunachal_all_years_master_df) == len(verified_master)
    )

    print("\nYEAR-WISE COMPARISON COMPLETED!")

else:
    print(
        "No yearly CSV files found in the current working folder. "
        "Check where your yearly CSV files were saved."
    )
```

```python

import os
import json
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("dark_background")

# ASSAM - 2021 dataset path
base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state\assam\2021"

all_data = []

# 1. Load JSON files and create Master DataFrame
for quarter in range(1, 5):
    file_path = os.path.join(base_path, f"{quarter}.json")

    if os.path.exists(file_path):
        with open(file_path, "r") as file:
            data = json.load(file)

        for category in data["data"]["transactionData"]:
            total_count = sum(
                instrument["count"]
                for instrument in category["paymentInstruments"]
            )

            all_data.append({
                "State": "Assam",
                "Year": 2021,
                "Quarter": f"Q{quarter}",
                "Category": category["name"],
                "Transactions": total_count
            })
    else:
        print(f"Q{quarter} file not found")

assam_2021_master_df = pd.DataFrame(all_data)

if assam_2021_master_df.empty:
    print("No data loaded. Check the Assam 2021 folder path.")
else:
    print("ASSAM - 2021 MASTER DATA")
    display(assam_2021_master_df)

    # 2. Quarter-wise Category Comparison
    quarter_comparison_2021 = assam_2021_master_df.pivot_table(
        index="Category",
        columns="Quarter",
        values="Transactions",
        aggfunc="sum",
        fill_value=0
    )

    print("QUARTER-WISE CATEGORY COMPARISON")
    display(quarter_comparison_2021)

    # 3. Category Comparison Chart
    quarter_colors = {
        "Q1": "#60A5FA",
        "Q2": "#34D399",
        "Q3": "#A78BFA",
        "Q4": "#F59E0B"
    }

    quarter_comparison_2021.plot(
        kind="bar",
        figsize=(12, 6),
        color=[
            quarter_colors[q]
            for q in quarter_comparison_2021.columns
        ]
    )

    plt.title("Assam - 2021 Category Comparison")
    plt.xlabel("Payment Category")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    # 4. Quarter Totals and Highest Quarter
    quarter_totals_2021 = assam_2021_master_df.groupby(
        "Quarter"
    )["Transactions"].sum()

    print("QUARTER TOTALS")
    display(quarter_totals_2021)

    print(
        "Highest Quarter:",
        quarter_totals_2021.idxmax(),
        "-",
        f"{quarter_totals_2021.max():,} transactions"
    )

    # 5. Quarter Totals Chart
    quarter_totals_2021.plot(
        kind="bar",
        figsize=(8, 5),
        color=[
            quarter_colors[q]
            for q in quarter_totals_2021.index
        ]
    )

    plt.title("Assam - 2021 Quarter-wise Transactions")
    plt.xlabel("Quarter")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 6. Category Totals and Highest Category
    category_totals_2021 = assam_2021_master_df.groupby(
        "Category"
    )["Transactions"].sum().sort_values(ascending=False)

    print("CATEGORY TOTALS")
    display(category_totals_2021)

    print(
        "Highest Payment Category:",
        category_totals_2021.idxmax(),
        "-",
        f"{category_totals_2021.max():,} transactions"
    )

    # 7. Payment Category Chart - Different Colours
    category_colors = [
        "#F472B6", "#22D3EE", "#FBBF24", "#A3E635",
        "#C084FC", "#FB7185", "#2DD4BF", "#818CF8"
    ]

    plt.figure(figsize=(11, 6))

    plt.bar(
        category_totals_2021.index,
        category_totals_2021.values,
        color=[
            category_colors[i % len(category_colors)]
            for i in range(len(category_totals_2021))
        ]
    )

    plt.title("Assam - 2021 Payment Category Analysis")
    plt.xlabel("Payment Category")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    # 8. Save Master and Analysis CSV Files
    master_file = "assam_transactions_2021.csv"
    comparison_file = "assam_2021_quarter_category_comparison.csv"
    quarter_file = "assam_2021_quarter_totals.csv"
    category_file = "assam_2021_category_totals.csv"

    assam_2021_master_df.to_csv(master_file, index=False)
    quarter_comparison_2021.to_csv(comparison_file)
    quarter_totals_2021.to_csv(quarter_file, header=["Transactions"])
    category_totals_2021.to_csv(category_file, header=["Transactions"])

    # 9. Verify Saved Files
    print("\nSAVE VERIFICATION")

    for filename in [
        master_file,
        comparison_file,
        quarter_file,
        category_file
    ]:
        print(filename, "->", os.path.exists(filename))

    verified_df = pd.read_csv(master_file)

    print("Original master rows:", len(assam_2021_master_df))
    print("Saved master rows:", len(verified_df))
    print(
        "Master verification:",
        len(assam_2021_master_df) == len(verified_df)
    )

    print("ASSAM 2021 ANALYSIS COMPLETED!")
```

```python

import os
import json
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("dark_background")

# ASSAM - 2025 dataset path
base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state\assam\2025"

all_data = []

# 1. Load quarterly JSON files
for quarter in range(1, 5):
    file_path = os.path.join(base_path, f"{quarter}.json")

    if os.path.exists(file_path):
        with open(file_path, "r") as file:
            data = json.load(file)

        for category in data["data"]["transactionData"]:
            total_count = sum(
                instrument["count"]
                for instrument in category["paymentInstruments"]
            )

            all_data.append({
                "State": "Assam",
                "Year": 2025,
                "Quarter": f"Q{quarter}",
                "Category": category["name"],
                "Transactions": total_count
            })
    else:
        print(f"Q{quarter} file not found")

assam_2025_master_df = pd.DataFrame(all_data)

if assam_2025_master_df.empty:
    print("No data loaded. Check the Assam 2025 folder path.")
else:
    print("ASSAM - 2025 MASTER DATA")
    display(assam_2025_master_df)

    # 2. Quarter-wise Category Comparison
    quarter_comparison_2025 = assam_2025_master_df.pivot_table(
        index="Category",
        columns="Quarter",
        values="Transactions",
        aggfunc="sum",
        fill_value=0
    )

    print("QUARTER-WISE CATEGORY COMPARISON")
    display(quarter_comparison_2025)

    # 3. Category Comparison Chart
    quarter_colors = {
        "Q1": "#60A5FA",
        "Q2": "#34D399",
        "Q3": "#A78BFA",
        "Q4": "#F59E0B"
    }

    quarter_comparison_2025.plot(
        kind="bar",
        figsize=(12, 6),
        color=[
            quarter_colors[q]
            for q in quarter_comparison_2025.columns
        ]
    )

    plt.title("Assam - 2025 Category Comparison")
    plt.xlabel("Payment Category")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    # 4. Quarter Totals and Highest Quarter
    quarter_totals_2025 = assam_2025_master_df.groupby(
        "Quarter"
    )["Transactions"].sum()

    print("QUARTER TOTALS")
    display(quarter_totals_2025)

    print(
        "Highest Quarter:",
        quarter_totals_2025.idxmax(),
        "-",
        f"{quarter_totals_2025.max():,} transactions"
    )

    # 5. Quarter Totals Chart
    quarter_totals_2025.plot(
        kind="bar",
        figsize=(8, 5),
        color=[
            quarter_colors[q]
            for q in quarter_totals_2025.index
        ]
    )

    plt.title("Assam - 2025 Quarter-wise Transactions")
    plt.xlabel("Quarter")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 6. Category Totals and Highest Category
    category_totals_2025 = assam_2025_master_df.groupby(
        "Category"
    )["Transactions"].sum().sort_values(ascending=False)

    print("CATEGORY TOTALS")
    display(category_totals_2025)

    print(
        "Highest Payment Category:",
        category_totals_2025.idxmax(),
        "-",
        f"{category_totals_2025.max():,} transactions"
    )

    # 7. Payment Category Chart - Different Colours
    category_colors = [
        "#F472B6", "#22D3EE", "#FBBF24", "#A3E635",
        "#C084FC", "#FB7185", "#2DD4BF", "#818CF8"
    ]

    plt.figure(figsize=(11, 6))

    plt.bar(
        category_totals_2025.index,
        category_totals_2025.values,
        color=[
            category_colors[i % len(category_colors)]
            for i in range(len(category_totals_2025))
        ]
    )

    plt.title("Assam - 2025 Payment Category Analysis")
    plt.xlabel("Payment Category")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    # 8. Save CSV Files
    master_file = "assam_transactions_2025.csv"
    comparison_file = "assam_2025_quarter_category_comparison.csv"
    quarter_file = "assam_2025_quarter_totals.csv"
    category_file = "assam_2025_category_totals.csv"

    assam_2025_master_df.to_csv(master_file, index=False)
    quarter_comparison_2025.to_csv(comparison_file)
    quarter_totals_2025.to_csv(quarter_file, header=["Transactions"])
    category_totals_2025.to_csv(category_file, header=["Transactions"])

    # 9. Verify Saved Files
    print("\nSAVE VERIFICATION")

    for filename in [
        master_file,
        comparison_file,
        quarter_file,
        category_file
    ]:
        print(filename, "->", os.path.exists(filename))

    verified_df = pd.read_csv(master_file)

    print("Original master rows:", len(assam_2025_master_df))
    print("Saved master rows:", len(verified_df))
    print(
        "Master verification:",
        len(assam_2025_master_df) == len(verified_df)
    )

    print("ASSAM 2025 ANALYSIS COMPLETED!")
```

```python

import os
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("dark_background")

# 1. Load Assam master CSV files for all years
all_years_data = []

for year in range(2018, 2027):
    filename = f"assam_transactions_{year}.csv"

    if os.path.exists(filename):
        df = pd.read_csv(filename)
        all_years_data.append(df)
        print(f"{year} loaded successfully")
    else:
        print(f"{year} CSV not found: {filename}")

# 2. Combine all years
if all_years_data:
    assam_all_years_df = pd.concat(
        all_years_data,
        ignore_index=True
    )

    print("\nASSAM - ALL YEARS MASTER DATA")
    display(assam_all_years_df.head())

    # 3. Calculate total transactions for each year
    yearly_totals = (
        assam_all_years_df
        .groupby("Year")["Transactions"]
        .sum()
        .sort_index()
    )

    yearly_summary = yearly_totals.reset_index()
    yearly_summary.columns = ["Year", "Total_Transactions"]

    print("\nYEAR-WISE TRANSACTION TOTALS")
    display(yearly_summary)

    # 4. Find the best-performing year
    best_year = yearly_summary.loc[
        yearly_summary["Total_Transactions"].idxmax()
    ]

    print("\nBEST-PERFORMING YEAR:", int(best_year["Year"]))
    print(
        "Total Transactions:",
        f"{int(best_year['Total_Transactions']):,}"
    )

    # 5. Year-wise comparison chart
    plt.figure(figsize=(12, 6))

    bar_colors = [
        "#F472B6" if year == best_year["Year"] else "#60A5FA"
        for year in yearly_summary["Year"]
    ]

    plt.bar(
        yearly_summary["Year"].astype(str),
        yearly_summary["Total_Transactions"],
        color=bar_colors
    )

    plt.title("Assam - Year-wise Transaction Performance")
    plt.xlabel("Year")
    plt.ylabel("Total Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 6. Save combined data and yearly summary
    assam_all_years_df.to_csv(
        "assam_all_years_master.csv",
        index=False
    )

    yearly_summary.to_csv(
        "assam_yearly_transaction_summary.csv",
        index=False
    )

    # 7. Verify saved files
    print("\nFILE VERIFICATION")
    print(
        "Combined master saved:",
        os.path.exists("assam_all_years_master.csv")
    )
    print(
        "Yearly summary saved:",
        os.path.exists("assam_yearly_transaction_summary.csv")
    )

else:
    print(
        "No Assam CSV files found. Check your Jupyter working folder."
    )
```

```python

import os
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("dark_background")

# 1. Load Bihar master CSV files from 2018 to 2026
all_years_data = []

for year in range(2018, 2027):
    file_name = f"bihar_transactions_{year}.csv"

    if os.path.exists(file_name):
        df = pd.read_csv(file_name)
        all_years_data.append(df)
        print(f"{year}: Loaded successfully - {len(df)} rows")
    else:
        print(f"{year}: File not found - {file_name}")

if not all_years_data:
    print("No Bihar master CSV files found. Check your folder.")
else:
    # 2. Combine all years
    bihar_all_years_master = pd.concat(
        all_years_data, ignore_index=True
    )

    print("\nBIHAR - ALL YEARS MASTER DATA")
    display(bihar_all_years_master)

    # 3. Year-wise transaction totals
    yearly_summary = (
        bihar_all_years_master
        .groupby("Year")
        .agg(
            Total_Transactions=("Transactions", "sum"),
            Quarters_Available=("Quarter", "nunique"),
            Categories=("Category", "nunique")
        )
        .reset_index()
    )

    yearly_summary["Data_Status"] = yearly_summary["Year"].apply(
        lambda year: (
            "Partial/current year - check available quarters"
            if year == 2026
            else "Historical year - verify all 4 quarters"
        )
    )

    print("\nYEAR-WISE TRANSACTION SUMMARY")
    display(yearly_summary)

    # 4. Compare historical years separately from 2026
    historical_years = yearly_summary[
        (yearly_summary["Year"] < 2026) &
        (yearly_summary["Quarters_Available"] == 4)
    ]

    if not historical_years.empty:
        highest_year = historical_years.loc[
            historical_years["Total_Transactions"].idxmax()
        ]
        lowest_year = historical_years.loc[
            historical_years["Total_Transactions"].idxmin()
        ]

        print("\nHISTORICAL YEAR COMPARISON (2018-2025)")
        print(
            "Highest year:",
            int(highest_year["Year"]),
            "-",
            f"{int(highest_year['Total_Transactions']):,} transactions"
        )
        print(
            "Lowest year:",
            int(lowest_year["Year"]),
            "-",
            f"{int(lowest_year['Total_Transactions']):,} transactions"
        )
    else:
        print("No historical years with all 4 quarters found.")

    # 5. Year-wise trend chart
    plt.figure(figsize=(12, 6))

    plt.plot(
        yearly_summary["Year"],
        yearly_summary["Total_Transactions"],
        marker="o",
        linewidth=2
    )

    plt.title("Bihar - Year-wise Transaction Trend (2018-2026)")
    plt.xlabel("Year")
    plt.ylabel("Total Transactions")
    plt.xticks(yearly_summary["Year"])
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

    # 6. Historical comparison chart (complete 4-quarter years only)
    if not historical_years.empty:
        plt.figure(figsize=(12, 6))

        plt.bar(
            historical_years["Year"].astype(str),
            historical_years["Total_Transactions"],
            color=[
                "#60A5FA" if year != highest_year["Year"]
                else "#34D399"
                for year in historical_years["Year"]
            ]
        )

        plt.title("Bihar - Annual Transactions (Complete Years)")
        plt.xlabel("Year")
        plt.ylabel("Total Transactions")
        plt.tight_layout()
        plt.show()

    # 7. Save combined master and yearly summary
    bihar_all_years_master.to_csv(
        "bihar_all_years_master.csv", index=False
    )
    yearly_summary.to_csv(
        "bihar_yearly_transaction_summary.csv", index=False
    )

    # 8. Verify saved files
    print("\nSAVE VERIFICATION")

    for filename in [
        "bihar_all_years_master.csv",
        "bihar_yearly_transaction_summary.csv"
    ]:
        print(filename, "->", os.path.exists(filename))

    saved_master = pd.read_csv("bihar_all_years_master.csv")
    saved_summary = pd.read_csv("bihar_yearly_transaction_summary.csv")

    print("Combined master rows:", len(bihar_all_years_master))
    print("Saved master rows:", len(saved_master))
    print("Master verification:", len(saved_master) == len(bihar_all_years_master))

    print("Yearly summary rows:", len(yearly_summary))
    print("Saved summary rows:", len(saved_summary))
    print("Summary verification:", len(saved_summary) == len(yearly_summary))

    print("\nBIHAR YEAR-WISE COMPARISON COMPLETED!")
```

```python

import os
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("dark_background")

# 1. Load Bihar yearly master CSV files
all_data = []

for year in range(2018, 2027):
    file_name = f"bihar_transactions_{year}.csv"

    if os.path.exists(file_name):
        df = pd.read_csv(file_name)
        all_data.append(df)
        print(f"{year} loaded successfully")
    else:
        print(f"{file_name} not found")

if not all_data:
    print("No Bihar yearly master files found. Check your folder.")
else:
    # 2. Combine all years
    bihar_all_years_df = pd.concat(all_data, ignore_index=True)

    # 3. Category-wise transaction totals for each year
    category_yearly = bihar_all_years_df.pivot_table(
        index="Category",
        columns="Year",
        values="Transactions",
        aggfunc="sum",
        fill_value=0
    )

    print("\nBIHAR - CATEGORY-WISE YEARLY COMPARISON")
    display(category_yearly)

    # 4. Category totals across all loaded years
    category_totals = (
        bihar_all_years_df.groupby("Category")["Transactions"]
        .sum()
        .sort_values(ascending=False)
    )

    print("\nTOTAL TRANSACTIONS BY CATEGORY")
    display(category_totals)

    if not category_totals.empty:
        print(
            "Highest category across loaded years:",
            category_totals.idxmax(),
            "-",
            f"{category_totals.max():,} transactions"
        )

    # 5. Category yearly trend chart
    category_yearly.T.plot(
        kind="line",
        marker="o",
        figsize=(13, 7)
    )

    plt.title("Bihar - Category-wise Transaction Trend (2018-2026)")
    plt.xlabel("Year")
    plt.ylabel("Number of Transactions")
    plt.xticks(category_yearly.columns, rotation=45)
    plt.grid(alpha=0.3)
    plt.legend(title="Payment Category", bbox_to_anchor=(1.02, 1), loc="upper left")
    plt.tight_layout()
    plt.show()

    # 6. Highest category in each year
    yearly_category_winner = (
        bihar_all_years_df.groupby(["Year", "Category"])["Transactions"]
        .sum()
        .reset_index()
    )

    idx = yearly_category_winner.groupby("Year")["Transactions"].idxmax()
    yearly_winners = yearly_category_winner.loc[idx].sort_values("Year")

    print("\nHIGHEST TRANSACTION CATEGORY IN EACH YEAR")
    display(yearly_winners)

    # 7. Save comparison files
    category_yearly.to_csv("bihar_category_yearly_comparison.csv")
    category_totals.to_csv("bihar_category_totals_2018_2026.csv")
    yearly_winners.to_csv("bihar_highest_category_each_year.csv", index=False)

    # 8. Verify saved files
    print("\nSAVE VERIFICATION")

    for filename in [
        "bihar_category_yearly_comparison.csv",
        "bihar_category_totals_2018_2026.csv",
        "bihar_highest_category_each_year.csv"
    ]:
        print(filename, "->", os.path.exists(filename))

    print("\nBIHAR CATEGORY-WISE YEARLY COMPARISON COMPLETED!")
```

```python

import os
import json
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("dark_background")

# CHANDIGARH - 2018 DATASET PATH
base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state\chandigarh\2018"

all_data = []

# 1. LOAD ALL FOUR QUARTERS
for quarter in range(1, 5):
    file_path = os.path.join(base_path, f"{quarter}.json")

    if os.path.exists(file_path):
        with open(file_path, "r") as file:
            data = json.load(file)

        for category in data["data"]["transactionData"]:
            total_count = sum(
                instrument["count"]
                for instrument in category["paymentInstruments"]
            )

            all_data.append({
                "State": "Chandigarh",
                "Year": 2018,
                "Quarter": f"Q{quarter}",
                "Category": category["name"],
                "Transactions": total_count
            })

        print(f"Q{quarter} loaded successfully")
    else:
        print(f"Q{quarter} file not found")

# 2. MASTER DATAFRAME
chandigarh_2018_master_df = pd.DataFrame(all_data)

if chandigarh_2018_master_df.empty:
    print("No data loaded. Please check the folder path.")
else:
    print("\nCHANDIGARH - 2018 MASTER DATA")
    display(chandigarh_2018_master_df)

    # 3. QUARTER-WISE CATEGORY COMPARISON
    quarter_comparison_2018 = chandigarh_2018_master_df.pivot_table(
        index="Category",
        columns="Quarter",
        values="Transactions",
        aggfunc="sum",
        fill_value=0
    )

    print("\nQUARTER-WISE CATEGORY COMPARISON")
    display(quarter_comparison_2018)

    quarter_colors = {
        "Q1": "#60A5FA",
        "Q2": "#34D399",
        "Q3": "#A78BFA",
        "Q4": "#F59E0B"
    }

    quarter_comparison_2018.plot(
        kind="bar",
        figsize=(12, 6),
        color=[
            quarter_colors[q]
            for q in quarter_comparison_2018.columns
        ]
    )

    plt.title("Chandigarh - 2018 Category Comparison")
    plt.xlabel("Payment Category")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    # 4. QUARTER TOTALS AND HIGHEST QUARTER
    quarter_totals_2018 = chandigarh_2018_master_df.groupby(
        "Quarter"
    )["Transactions"].sum()

    print("\nQUARTER TOTALS")
    display(quarter_totals_2018)

    print(
        "Highest Quarter:",
        quarter_totals_2018.idxmax(),
        "-",
        f"{quarter_totals_2018.max():,} transactions"
    )

    quarter_totals_2018.plot(
        kind="bar",
        figsize=(8, 5),
        color=[
            quarter_colors[q]
            for q in quarter_totals_2018.index
        ]
    )

    plt.title("Chandigarh - 2018 Quarter-wise Transactions")
    plt.xlabel("Quarter")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 5. CATEGORY TOTALS AND HIGHEST CATEGORY
    category_totals_2018 = chandigarh_2018_master_df.groupby(
        "Category"
    )["Transactions"].sum().sort_values(ascending=False)

    print("\nCATEGORY TOTALS")
    display(category_totals_2018)

    print(
        "Highest Payment Category:",
        category_totals_2018.idxmax(),
        "-",
        f"{category_totals_2018.max():,} transactions"
    )

    # 6. CATEGORY CHART WITH DIFFERENT COLOURS
    category_colors = [
        "#F472B6", "#22D3EE", "#FBBF24", "#A3E635",
        "#C084FC", "#FB7185", "#2DD4BF", "#818CF8"
    ]

    plt.figure(figsize=(11, 6))

    plt.bar(
        category_totals_2018.index,
        category_totals_2018.values,
        color=[
            category_colors[i % len(category_colors)]
            for i in range(len(category_totals_2018))
        ]
    )

    plt.title("Chandigarh - 2018 Payment Category Analysis")
    plt.xlabel("Payment Category")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    # 7. SAVE FOUR CSV FILES
    master_file = "chandigarh_transactions_2018.csv"
    comparison_file = "chandigarh_2018_quarter_category_comparison.csv"
    quarter_file = "chandigarh_2018_quarter_totals.csv"
    category_file = "chandigarh_2018_category_totals.csv"

    chandigarh_2018_master_df.to_csv(master_file, index=False)
    quarter_comparison_2018.to_csv(comparison_file)
    quarter_totals_2018.to_csv(
        quarter_file, header=["Transactions"]
    )
    category_totals_2018.to_csv(
        category_file, header=["Transactions"]
    )

    # 8. VERIFY SAVED FILES
    print("\nSAVE VERIFICATION")

    for filename in [
        master_file,
        comparison_file,
        quarter_file,
        category_file
    ]:
        print(filename, "->", os.path.exists(filename))

    saved_df = pd.read_csv(master_file)

    print("\nOriginal master rows:", len(chandigarh_2018_master_df))
    print("Saved master rows:", len(saved_df))
    print(
        "Master verification:",
        len(chandigarh_2018_master_df) == len(saved_df)
    )

    print("\nCHANDIGARH 2018 ANALYSIS COMPLETED!")
```

```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("dark_background")

# Chandigarh 2020 dataset path
base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state\chandigarh\2020"

all_data = []

# Load Q1 to Q4
for quarter in range(1, 5):
    file_path = os.path.join(base_path, f"{quarter}.json")

    if os.path.exists(file_path):
        with open(file_path, "r") as file:
            data = json.load(file)

        for category in data["data"]["transactionData"]:
            total_count = sum(
                instrument["count"]
                for instrument in category["paymentInstruments"]
            )

            all_data.append({
                "State": "Chandigarh",
                "Year": 2020,
                "Quarter": f"Q{quarter}",
                "Category": category["name"],
                "Transactions": total_count
            })

        print(f"Q{quarter} loaded successfully")
    else:
        print(f"Q{quarter} file not found")

# Create master DataFrame
chandigarh_2020_master_df = pd.DataFrame(all_data)

if chandigarh_2020_master_df.empty:
    print("No data loaded. Please check the folder path.")

else:
    print("\n========== CHANDIGARH 2020 MASTER DATA ==========")
    display(chandigarh_2020_master_df)

    # Quarter and category comparison
    pivot = chandigarh_2020_master_df.pivot_table(
        index="Category",
        columns="Quarter",
        values="Transactions",
        aggfunc="sum",
        fill_value=0
    )

    print("\n========== QUARTER-WISE CATEGORY COMPARISON ==========")
    display(pivot)

    quarter_colors = {
        "Q1": "#60A5FA",
        "Q2": "#34D399",
        "Q3": "#A78BFA",
        "Q4": "#F59E0B"
    }

    available_quarters = [
        q for q in ["Q1", "Q2", "Q3", "Q4"]
        if q in pivot.columns
    ]

    pivot.plot(
        kind="bar",
        figsize=(13, 6),
        color=[quarter_colors[q] for q in available_quarters]
    )
    plt.title("Chandigarh 2020 - Category Transactions by Quarter")
    plt.xlabel("Payment Category")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=45, ha="right")
    plt.legend(title="Quarter")
    plt.tight_layout()
    plt.show()

    # Quarter totals
    quarter_totals = (
        chandigarh_2020_master_df
        .groupby("Quarter")["Transactions"]
        .sum()
        .reindex(["Q1", "Q2", "Q3", "Q4"])
        .dropna()
    )

    print("\n========== QUARTER TOTALS ==========")
    display(quarter_totals)

    if not quarter_totals.empty:
        print(
            f"Highest transaction quarter: {quarter_totals.idxmax()} "
            f"({int(quarter_totals.max()):,} transactions)"
        )

        quarter_totals.plot(
            kind="bar",
            figsize=(9, 5),
            color=[quarter_colors[q] for q in quarter_totals.index]
        )
        plt.title("Chandigarh 2020 - Total Transactions by Quarter")
        plt.xlabel("Quarter")
        plt.ylabel("Number of Transactions")
        plt.xticks(rotation=0)
        plt.tight_layout()
        plt.show()

    # Category totals
    category_totals = (
        chandigarh_2020_master_df
        .groupby("Category")["Transactions"]
        .sum()
        .sort_values(ascending=False)
    )

    print("\n========== CATEGORY TOTALS ==========")
    display(category_totals)

    if not category_totals.empty:
        print(
            f"Highest transaction category: {category_totals.idxmax()} "
            f"({int(category_totals.max()):,} transactions)"
        )

        category_colors = [
            "#F472B6", "#22D3EE", "#FBBF24", "#A3E635",
            "#C084FC", "#FB7185", "#2DD4BF", "#818CF8"
        ]

        category_totals.plot(
            kind="bar",
            figsize=(12, 6),
            color=[
                category_colors[i % len(category_colors)]
                for i in range(len(category_totals))
            ]
        )
        plt.title("Chandigarh 2020 - Transactions by Category")
        plt.xlabel("Payment Category")
        plt.ylabel("Number of Transactions")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()

    # Save CSV files
    master_file = "chandigarh_transactions_2020.csv"
    comparison_file = "chandigarh_2020_quarter_category_comparison.csv"
    quarter_file = "chandigarh_2020_quarter_totals.csv"
    category_file = "chandigarh_2020_category_totals.csv"

    chandigarh_2020_master_df.to_csv(master_file, index=False)
    pivot.to_csv(comparison_file)
    quarter_totals.rename("Transactions").to_csv(quarter_file)
    category_totals.rename("Transactions").to_csv(category_file)

    print("\n========== CSV FILE VERIFICATION ==========")

    for filename in [
        master_file,
        comparison_file,
        quarter_file,
        category_file
    ]:
        print(f"{filename}: {os.path.exists(filename)}")

    saved_master = pd.read_csv(master_file)

    print("Original rows:", len(chandigarh_2020_master_df))
    print("Saved rows:", len(saved_master))
    print(
        "Master verification:",
        len(chandigarh_2020_master_df) == len(saved_master)
    )
```

```python

import os
import json
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# 1. Folder path
base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state\dadra-&-nagar-haveli-&-daman-&-diu"

state_name = "Dadra & Nagar Haveli and Daman & Diu"
years = range(2019, 2027)
quarters = range(1, 5)

records = []
missing_files = []

# 2. Load all JSON files: 2019 to 2026
for year in years:
    for quarter in quarters:

        file_path = os.path.join(
            base_path, str(year), f"{quarter}.json"
        )

        if not os.path.exists(file_path):
            missing_files.append(f"{year} Q{quarter}")
            print(f"Missing: {year} Q{quarter}")
            continue

        try:
            with open(file_path, "r", encoding="utf-8") as file:
                data = json.load(file)

            transaction_data = data["data"]["transactionData"]

            for item in transaction_data:
                category = item["name"]

                total_transactions = sum(
                    instrument["count"]
                    for instrument in item["paymentInstruments"]
                )

                records.append({
                    "State": state_name,
                    "Year": year,
                    "Quarter": f"Q{quarter}",
                    "Category": category,
                    "Transactions": total_transactions
                })

            print(f"Loaded: {year} Q{quarter}")

        except (KeyError, ValueError, TypeError) as error:
            print(f"Error in {year} Q{quarter}: {error}")
            missing_files.append(f"{year} Q{quarter} (read error)")

# 3. Create master DataFrame
df = pd.DataFrame(
    records,
    columns=["State", "Year", "Quarter", "Category", "Transactions"]
)

if df.empty:
    print("\nNo data loaded. Check the folder path.")
else:
    print("\n========== MASTER DATA ==========")
    display(df)

    # 4. Year-wise total transactions
    yearly_total = df.groupby("Year")["Transactions"].sum()

    print("\n========== YEAR-WISE TOTAL ==========")
    display(yearly_total.rename("Transactions").reset_index())

    highest_year = yearly_total.idxmax()
    print("Highest transaction year:", highest_year)
    print("Total transactions:", f"{yearly_total.max():,}")

    plt.style.use("dark_background")
    yearly_total.plot(
        kind="bar",
        figsize=(12, 5),
        title="Year-wise Transactions (2019–2026)"
    )
    plt.xlabel("Year")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 5. Quarter-wise comparison by year
    quarter_order = ["Q1", "Q2", "Q3", "Q4"]

    quarter_year = df.pivot_table(
        index="Year",
        columns="Quarter",
        values="Transactions",
        aggfunc="sum"
    ).reindex(columns=quarter_order)

    print("\n========== QUARTER-WISE COMPARISON ==========")
    display(quarter_year)

    quarter_year.plot(
        kind="bar",
        figsize=(13, 6),
        title="Quarter-wise Transactions by Year"
    )
    plt.xlabel("Year")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 6. Total transactions for each quarter across all available years
    quarter_totals_all_years = (
        df.groupby("Quarter")["Transactions"]
        .sum()
        .reindex(quarter_order)
        .dropna()
    )

    print("\n========== TOTAL BY QUARTER ==========")
    display(
        quarter_totals_all_years
        .rename("Transactions")
        .reset_index()
    )

    if not quarter_totals_all_years.empty:
        print(
            "Highest quarter overall:",
            quarter_totals_all_years.idxmax()
        )

    # 7. Category-wise totals
    category_totals = (
        df.groupby("Category")["Transactions"]
        .sum()
        .sort_values(ascending=False)
    )

    print("\n========== CATEGORY TOTALS ==========")
    display(category_totals.rename("Transactions").reset_index())

    if not category_totals.empty:
        print("Most-used category:", category_totals.idxmax())
        print(
            "Category transactions:",
            f"{category_totals.max():,}"
        )

        category_totals.plot(
            kind="bar",
            figsize=(12, 5),
            title="Payment Category Totals (2019–2026)"
        )
        plt.xlabel("Category")
        plt.ylabel("Number of Transactions")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()

    # 8. Category-wise comparison by year
    category_year = df.pivot_table(
        index="Category",
        columns="Year",
        values="Transactions",
        aggfunc="sum"
    )

    print("\n========== CATEGORY BY YEAR ==========")
    display(category_year)

    category_year.plot(
        kind="bar",
        figsize=(15, 6),
        title="Category-wise Transactions by Year"
    )
    plt.xlabel("Category")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    # 9. Highest year for each category
    highest_year_by_category = (
        df.groupby(["Category", "Year"])["Transactions"]
        .sum()
        .reset_index()
        .loc[
            lambda x: x.groupby("Category")["Transactions"].idxmax()
        ]
        .sort_values("Transactions", ascending=False)
    )

    print("\n========== HIGHEST YEAR BY CATEGORY ==========")
    display(highest_year_by_category)

    # 10. Save all CSV files
    df.to_csv(
        "dadra_daman_diu_master_2019_2026.csv",
        index=False
    )

    yearly_total.rename("Transactions").to_csv(
        "dadra_daman_diu_yearly_totals_2019_2026.csv"
    )

    quarter_year.to_csv(
        "dadra_daman_diu_quarter_year_comparison_2019_2026.csv"
    )

    quarter_totals_all_years.rename("Transactions").to_csv(
        "dadra_daman_diu_quarter_totals_2019_2026.csv"
    )

    category_totals.rename("Transactions").to_csv(
        "dadra_daman_diu_category_totals_2019_2026.csv"
    )

    category_year.to_csv(
        "dadra_daman_diu_category_year_comparison_2019_2026.csv"
    )

    highest_year_by_category.to_csv(
        "dadra_daman_diu_highest_year_by_category_2019_2026.csv",
        index=False
    )

    # 11. Verify master CSV
    saved_df = pd.read_csv(
        "dadra_daman_diu_master_2019_2026.csv"
    )

    print("\n========== VERIFICATION ==========")
    print("Master rows:", len(df))
    print("Saved rows:", len(saved_df))
    print("Master verification:", len(df) == len(saved_df))
    print("Missing files:", missing_files)

    print("\nAll available data processed!")
    print("CSV files saved successfully.")
```

```python

import os
import json
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# 1. Folder path
base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state\gujarat"

state_name = "Gujarat"
records = []
missing_files = []

# 2. Load all JSON files: 2018–2026
for year in range(2018, 2027):
    for quarter in range(1, 5):
        file_path = os.path.join(base_path, str(year), f"{quarter}.json")

        if not os.path.exists(file_path):
            missing_files.append(f"{year} Q{quarter}")
            print(f"Missing: {year} Q{quarter}")
            continue

        try:
            with open(file_path, "r", encoding="utf-8") as file:
                data = json.load(file)

            for item in data["data"]["transactionData"]:
                total = sum(
                    instrument["count"]
                    for instrument in item["paymentInstruments"]
                )

                records.append({
                    "State": state_name,
                    "Year": year,
                    "Quarter": f"Q{quarter}",
                    "Category": item["name"],
                    "Transactions": total
                })

            print(f"Loaded: {year} Q{quarter}")

        except (KeyError, ValueError, TypeError) as error:
            print(f"Error in {year} Q{quarter}: {error}")
            missing_files.append(f"{year} Q{quarter} (read error)")

# 3. Create master DataFrame
df = pd.DataFrame(
    records,
    columns=["State", "Year", "Quarter", "Category", "Transactions"]
)

if df.empty:
    print("No data loaded. Check the folder path.")
else:
    print("\n========== MASTER DATA ==========")
    display(df)

    # 4. Year-wise totals and highest complete year
    yearly_total = df.groupby("Year")["Transactions"].sum()
    quarter_counts = df.groupby("Year")["Quarter"].nunique()
    complete_years = quarter_counts[quarter_counts == 4].index
    complete_totals = yearly_total[yearly_total.index.isin(complete_years)]

    print("\n========== YEAR-WISE TOTALS ==========")
    display(yearly_total.rename("Transactions").reset_index())

    if not complete_totals.empty:
        highest_year = complete_totals.idxmax()
        print("Highest complete year:", highest_year)
        print("Transactions:", f"{complete_totals.max():,}")
    else:
        print("No complete year available for comparison.")

    plt.style.use("dark_background")
    yearly_total.plot(
        kind="bar", figsize=(12, 5),
        title="Gujarat Year-wise Transactions (2018–2026)"
    )
    plt.xlabel("Year")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 5. Quarter-wise comparison by year
    quarter_order = ["Q1", "Q2", "Q3", "Q4"]
    quarter_year = df.pivot_table(
        index="Year", columns="Quarter",
        values="Transactions", aggfunc="sum"
    ).reindex(columns=quarter_order)

    print("\n========== QUARTER-WISE COMPARISON ==========")
    display(quarter_year)

    quarter_year.plot(
        kind="bar", figsize=(13, 6),
        title="Gujarat Quarter-wise Transactions by Year"
    )
    plt.xlabel("Year")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 6. Total transactions by quarter across available years
    quarter_totals = (
        df.groupby("Quarter")["Transactions"]
        .sum().reindex(quarter_order).dropna()
    )

    print("\n========== TOTAL BY QUARTER ==========")
    display(quarter_totals.rename("Transactions").reset_index())

    if not quarter_totals.empty:
        print("Highest quarter overall:", quarter_totals.idxmax())

    # 7. Category totals
    category_totals = (
        df.groupby("Category")["Transactions"]
        .sum().sort_values(ascending=False)
    )

    print("\n========== CATEGORY TOTALS ==========")
    display(category_totals.rename("Transactions").reset_index())

    if not category_totals.empty:
        print("Most-used category:", category_totals.idxmax())

        category_totals.plot(
            kind="bar", figsize=(12, 5),
            title="Gujarat Payment Category Totals (2018–2026)"
        )
        plt.xlabel("Category")
        plt.ylabel("Number of Transactions")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()

    # 8. Category-wise comparison by year
    category_year = df.pivot_table(
        index="Category", columns="Year",
        values="Transactions", aggfunc="sum"
    )

    print("\n========== CATEGORY BY YEAR ==========")
    display(category_year)

    category_year.plot(
        kind="bar", figsize=(15, 6),
        title="Gujarat Category-wise Transactions by Year"
    )
    plt.xlabel("Category")
    plt.ylabel("Number of Transactions")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    # 9. Highest year per category
    highest_year_category = (
        df.groupby(["Category", "Year"])["Transactions"]
        .sum().reset_index()
    )

    highest_year_category = highest_year_category.loc[
        highest_year_category.groupby("Category")["Transactions"].idxmax()
    ]

    print("\n========== HIGHEST YEAR BY CATEGORY ==========")
    display(highest_year_category)

    # 10. Save CSV files
    df.to_csv("gujarat_master_2018_2026.csv", index=False)
    yearly_total.rename("Transactions").to_csv(
        "gujarat_yearly_totals_2018_2026.csv"
    )
    quarter_year.to_csv("gujarat_quarter_year_comparison_2018_2026.csv")
    quarter_totals.rename("Transactions").to_csv(
        "gujarat_quarter_totals_2018_2026.csv"
    )
    category_totals.rename("Transactions").to_csv(
        "gujarat_category_totals_2018_2026.csv"
    )
    category_year.to_csv("gujarat_category_year_comparison_2018_2026.csv")
    highest_year_category.to_csv(
        "gujarat_highest_year_by_category_2018_2026.csv",
        index=False
    )

    # 11. Verify saved master CSV
    saved_df = pd.read_csv("gujarat_master_2018_2026.csv")

    print("\n========== VERIFICATION ==========")
    print("Master rows:", len(df))
    print("Saved rows:", len(saved_df))
    print("Master verification:", len(df) == len(saved_df))
    print("Missing files:", missing_files)
    print("\nGujarat analysis completed!")

```

```python

import os
import json
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# 1. Folder path
base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state\jammu-&-kashmir"

state_name = "Jammu & Kashmir"
records = []
missing_files = []

# 2. Load all JSON files: 2018–2026
for year in range(2018, 2027):
    for quarter in range(1, 5):
        file_path = os.path.join(base_path, str(year), f"{quarter}.json")

        if not os.path.exists(file_path):
            missing_files.append(f"{year} Q{quarter}")
            print(f"Missing: {year} Q{quarter}")
            continue

        try:
            with open(file_path, "r", encoding="utf-8") as file:
                data = json.load(file)

            for item in data["data"]["transactionData"]:
                total = sum(
                    instrument["count"]
                    for instrument in item["paymentInstruments"]
                )

                records.append({
                    "State": state_name,
                    "Year": year,
                    "Quarter": f"Q{quarter}",
                    "Category": item["name"],
                    "Transactions": total
                })

            print(f"Loaded: {year} Q{quarter}")

        except (KeyError, ValueError, TypeError) as error:
            print(f"Error in {year} Q{quarter}: {error}")
            missing_files.append(f"{year} Q{quarter} (read error)")

# 3. Create master DataFrame
df = pd.DataFrame(
    records,
    columns=["State", "Year", "Quarter", "Category", "Transactions"]
)

if df.empty:
    print("No data loaded. Check the folder path.")
else:
    print("\n========== MASTER DATA ==========")
    display(df)

    # 4. Year-wise totals
    yearly_total = df.groupby("Year")["Transactions"].sum()
    quarters_per_year = df.groupby("Year")["Quarter"].nunique()

    year_df = pd.DataFrame({
        "Transactions": yearly_total,
        "Quarters_Available": quarters_per_year
    }).reset_index()

    print("\n========== YEAR-WISE TOTALS ==========")
    display(year_df)

    # Compare only years with four quarters
    complete_years = year_df[year_df["Quarters_Available"] == 4]

    if not complete_years.empty:
        highest = complete_years.loc[
            complete_years["Transactions"].idxmax()
        ]
        print("Highest complete year:", int(highest["Year"]))
        print("Transactions:", f'{int(highest["Transactions"]):,}')
    else:
        print("No complete years available.")

    plt.style.use("dark_background")
    yearly_total.plot(
        kind="bar", figsize=(12, 5),
        title="Jammu & Kashmir Year-wise Transactions"
    )
    plt.xlabel("Year")
    plt.ylabel("Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 5. Quarter totals across available years
    quarter_order = ["Q1", "Q2", "Q3", "Q4"]
    quarter_totals = (
        df.groupby("Quarter")["Transactions"]
        .sum()
        .reindex(quarter_order)
        .dropna()
    )

    print("\n========== QUARTER TOTALS ==========")
    display(
        quarter_totals.rename("Transactions")
        .reset_index()
        .sort_values("Transactions", ascending=False)
        .reset_index(drop=True)
    )

    if not quarter_totals.empty:
        print("Highest quarter:", quarter_totals.idxmax())

    # 6. Quarter-wise comparison by year
    quarter_year = df.pivot_table(
        index="Year",
        columns="Quarter",
        values="Transactions",
        aggfunc="sum"
    ).reindex(columns=quarter_order)

    print("\n========== QUARTER BY YEAR ==========")
    display(quarter_year)

    quarter_year.plot(
        kind="bar", figsize=(13, 6),
        title="Jammu & Kashmir Quarter-wise Comparison"
    )
    plt.xlabel("Year")
    plt.ylabel("Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 7. Category totals
    category_totals = (
        df.groupby("Category")["Transactions"]
        .sum()
        .sort_values(ascending=False)
    )

    print("\n========== CATEGORY TOTALS ==========")
    display(category_totals.rename("Transactions").reset_index())

    if not category_totals.empty:
        print("Most-used category:", category_totals.idxmax())

        category_totals.plot(
            kind="bar", figsize=(12, 5),
            title="Jammu & Kashmir Category Totals"
        )
        plt.xlabel("Category")
        plt.ylabel("Transactions")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()

    # 8. Category-wise comparison by year
    category_year = df.pivot_table(
        index="Category",
        columns="Year",
        values="Transactions",
        aggfunc="sum"
    )

    print("\n========== CATEGORY BY YEAR ==========")
    display(category_year)

    # 9. Save CSV files
    df.to_csv("jammu_kashmir_master_2018_2026.csv", index=False)
    year_df.to_csv("jammu_kashmir_yearly_totals_2018_2026.csv", index=False)
    quarter_year.to_csv("jammu_kashmir_quarter_year_2018_2026.csv")
    category_totals.to_csv("jammu_kashmir_category_totals_2018_2026.csv")
    category_year.to_csv("jammu_kashmir_category_year_2018_2026.csv")

    # 10. Verify saved master file
    saved_df = pd.read_csv("jammu_kashmir_master_2018_2026.csv")

    print("\n========== VERIFICATION ==========")
    print("Master rows:", len(df))
    print("Saved rows:", len(saved_df))
    print("Master verification:", len(df) == len(saved_df))
    print("Missing files:", missing_files)
    print("\nJammu & Kashmir analysis completed!")
```

```python

import os
import json
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# 1. Folder path
base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state\nagaland"

state_name = "Nagaland"
records = []
loaded_quarters = {}
missing_files = []

plt.style.use("dark_background")

# 2. Load JSON files: 2018–2026
for year in range(2018, 2027):
    loaded_quarters[year] = set()

    for quarter in range(1, 5):
        file_path = os.path.join(base_path, str(year), f"{quarter}.json")

        if not os.path.exists(file_path):
            missing_files.append(f"{year} Q{quarter}")
            print(f"Missing: {year} Q{quarter}")
            continue

        try:
            with open(file_path, "r", encoding="utf-8") as file:
                data = json.load(file)

            transaction_data = data["data"]["transactionData"]
            loaded_quarters[year].add(quarter)

            for item in transaction_data:
                total = sum(
                    instrument["count"]
                    for instrument in item["paymentInstruments"]
                )

                records.append({
                    "State": state_name,
                    "Year": year,
                    "Quarter": f"Q{quarter}",
                    "Category": item["name"],
                    "Transactions": total
                })

            print(f"Loaded: {year} Q{quarter}")

        except (KeyError, ValueError, TypeError) as error:
            print(f"Error in {year} Q{quarter}: {error}")
            missing_files.append(f"{year} Q{quarter} (read error)")

# 3. Create master DataFrame
df = pd.DataFrame(
    records,
    columns=["State", "Year", "Quarter", "Category", "Transactions"]
)

if df.empty:
    print("\nNo data loaded. Check the folder path.")
else:
    print("\n========== MASTER DATA ==========")
    display(df)

    # 4. Year-wise totals
    yearly_total = df.groupby("Year")["Transactions"].sum()

    year_df = pd.DataFrame({"Year": list(range(2018, 2027))})
    year_df["Transactions"] = year_df["Year"].map(yearly_total)
    year_df["Quarters_Available"] = year_df["Year"].map(
        lambda year: len(loaded_quarters[year])
    )

    print("\n========== YEAR-WISE TOTALS ==========")
    display(year_df)

    # Compare only years with all four quarter files
    complete_years = year_df[
        year_df["Quarters_Available"] == 4
    ].dropna(subset=["Transactions"])

    if not complete_years.empty:
        highest = complete_years.loc[
            complete_years["Transactions"].idxmax()
        ]
        print("\nHighest complete year:", int(highest["Year"]))
        print("Transactions:", f'{int(highest["Transactions"]):,}')
    else:
        print("\nNo complete years available for comparison.")

    yearly_total.plot(
        kind="bar", figsize=(12, 5),
        title="Nagaland Year-wise Transactions"
    )
    plt.xlabel("Year")
    plt.ylabel("Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 5. Quarter totals across available data
    quarter_order = ["Q1", "Q2", "Q3", "Q4"]
    quarter_totals = (
        df.groupby("Quarter")["Transactions"]
        .sum()
        .reindex(quarter_order)
        .dropna()
    )

    print("\n========== QUARTER TOTALS ==========")
    display(
        quarter_totals.rename("Transactions")
        .reset_index()
        .sort_values("Transactions", ascending=False)
        .reset_index(drop=True)
    )

    if not quarter_totals.empty:
        print("Highest quarter in summed available data:", quarter_totals.idxmax())

    # 6. Quarter-wise comparison by year
    quarter_year = df.pivot_table(
        index="Year",
        columns="Quarter",
        values="Transactions",
        aggfunc="sum"
    ).reindex(columns=quarter_order)

    print("\n========== QUARTER BY YEAR ==========")
    display(quarter_year)

    quarter_year.plot(
        kind="bar", figsize=(13, 6),
        title="Nagaland Quarter-wise Comparison"
    )
    plt.xlabel("Year")
    plt.ylabel("Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 7. Category totals
    category_totals = (
        df.groupby("Category")["Transactions"]
        .sum()
        .sort_values(ascending=False)
    )

    print("\n========== CATEGORY TOTALS ==========")
    display(category_totals.rename("Transactions").reset_index())

    if not category_totals.empty:
        print("Most-used category:", category_totals.idxmax())

        category_totals.plot(
            kind="bar", figsize=(12, 5),
            title="Nagaland Category Totals"
        )
        plt.xlabel("Category")
        plt.ylabel("Transactions")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()

    # 8. Category-wise comparison by year
    category_year = df.pivot_table(
        index="Category",
        columns="Year",
        values="Transactions",
        aggfunc="sum"
    )

    print("\n========== CATEGORY BY YEAR ==========")
    display(category_year)

    # 9. Save CSV files
    df.to_csv("nagaland_master_2018_2026.csv", index=False)
    year_df.to_csv("nagaland_yearly_totals_2018_2026.csv", index=False)
    quarter_year.to_csv("nagaland_quarter_year_2018_2026.csv")
    category_totals.to_csv("nagaland_category_totals_2018_2026.csv")
    category_year.to_csv("nagaland_category_year_2018_2026.csv")

    # 10. Verify saved data
    saved_df = pd.read_csv("nagaland_master_2018_2026.csv")

    print("\n========== VERIFICATION ==========")
    print("Master rows:", len(df))
    print("Saved rows:", len(saved_df))
    print("Master verification:", len(df) == len(saved_df))
    print("Missing files:", missing_files)
    print("\nNagaland analysis completed!")
```

```python

import os
import json
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# 1. Folder path
base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state\west-bengal"

state_name = "West Bengal"
records = []
loaded_quarters = {}
missing_files = []

plt.style.use("dark_background")

# 2. Load JSON files: 2018–2026
for year in range(2018, 2027):
    loaded_quarters[year] = set()

    for quarter in range(1, 5):
        file_path = os.path.join(base_path, str(year), f"{quarter}.json")

        if not os.path.exists(file_path):
            missing_files.append(f"{year} Q{quarter}")
            print(f"Missing: {year} Q{quarter}")
            continue

        try:
            with open(file_path, "r", encoding="utf-8") as file:
                data = json.load(file)

            transaction_data = data["data"]["transactionData"]
            loaded_quarters[year].add(quarter)

            for item in transaction_data:
                total = sum(
                    instrument["count"]
                    for instrument in item["paymentInstruments"]
                )

                records.append({
                    "State": state_name,
                    "Year": year,
                    "Quarter": f"Q{quarter}",
                    "Category": item["name"],
                    "Transactions": total
                })

            print(f"Loaded: {year} Q{quarter}")

        except (KeyError, ValueError, TypeError) as error:
            print(f"Error in {year} Q{quarter}: {error}")
            missing_files.append(f"{year} Q{quarter} (read error)")

# 3. Master DataFrame
df = pd.DataFrame(
    records,
    columns=["State", "Year", "Quarter", "Category", "Transactions"]
)

if df.empty:
    print("\nNo data loaded. Check the folder path.")
else:
    print("\n========== MASTER DATA ==========")
    display(df)

    # 4. Year-wise totals
    yearly_total = df.groupby("Year")["Transactions"].sum()

    year_df = pd.DataFrame({"Year": list(range(2018, 2027))})
    year_df["Transactions"] = year_df["Year"].map(yearly_total)
    year_df["Quarters_Available"] = year_df["Year"].map(
        lambda year: len(loaded_quarters[year])
    )

    print("\n========== YEAR-WISE TOTALS ==========")
    display(year_df)

    # Compare only years with all four quarters
    complete_years = year_df[
        year_df["Quarters_Available"] == 4
    ].dropna(subset=["Transactions"])

    if not complete_years.empty:
        highest = complete_years.loc[
            complete_years["Transactions"].idxmax()
        ]
        print("\nHighest complete year:", int(highest["Year"]))
        print("Transactions:", f'{int(highest["Transactions"]):,}')
    else:
        print("\nNo complete years available for comparison.")

    yearly_total.plot(
        kind="bar",
        figsize=(12, 5),
        title="West Bengal Year-wise Transactions"
    )
    plt.xlabel("Year")
    plt.ylabel("Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 5. Quarter totals
    quarter_order = ["Q1", "Q2", "Q3", "Q4"]

    quarter_totals = (
        df.groupby("Quarter")["Transactions"]
        .sum()
        .reindex(quarter_order)
        .dropna()
    )

    print("\n========== QUARTER TOTALS ==========")
    display(
        quarter_totals.rename("Transactions")
        .reset_index()
        .sort_values("Transactions", ascending=False)
        .reset_index(drop=True)
    )

    if not quarter_totals.empty:
        print(
            "Highest quarter in summed available data:",
            quarter_totals.idxmax()
        )

    # 6. Quarter by year
    quarter_year = df.pivot_table(
        index="Year",
        columns="Quarter",
        values="Transactions",
        aggfunc="sum"
    ).reindex(columns=quarter_order)

    print("\n========== QUARTER BY YEAR ==========")
    display(quarter_year)

    quarter_year.plot(
        kind="bar",
        figsize=(13, 6),
        title="West Bengal Quarter-wise Comparison"
    )
    plt.xlabel("Year")
    plt.ylabel("Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 7. Category totals
    category_totals = (
        df.groupby("Category")["Transactions"]
        .sum()
        .sort_values(ascending=False)
    )

    print("\n========== CATEGORY TOTALS ==========")
    display(category_totals.rename("Transactions").reset_index())

    if not category_totals.empty:
        print("Most-used category:", category_totals.idxmax())

        category_totals.plot(
            kind="bar",
            figsize=(12, 5),
            title="West Bengal Category Totals"
        )
        plt.xlabel("Category")
        plt.ylabel("Transactions")
        plt.xticks(rotation=45, ha="right")
        plt.tight_layout()
        plt.show()

    # 8. Category by year
    category_year = df.pivot_table(
        index="Category",
        columns="Year",
        values="Transactions",
        aggfunc="sum"
    )

    print("\n========== CATEGORY BY YEAR ==========")
    display(category_year)

    # 9. Save CSV files
    df.to_csv("west_bengal_master_2018_2026.csv", index=False)
    year_df.to_csv("west_bengal_yearly_totals_2018_2026.csv", index=False)
    quarter_year.to_csv("west_bengal_quarter_year_2018_2026.csv")
    category_totals.to_csv("west_bengal_category_totals_2018_2026.csv")
    category_year.to_csv("west_bengal_category_year_2018_2026.csv")

    # 10. Verify saved data
    saved_df = pd.read_csv("west_bengal_master_2018_2026.csv")

    print("\n========== VERIFICATION ==========")
    print("Master rows:", len(df))
    print("Saved rows:", len(saved_df))
    print("Master verification:", len(df) == len(saved_df))
    print("Missing files:", missing_files)
    print("\nWest Bengal analysis completed!")
```

```python

import os
import glob
import pandas as pd
from IPython.display import display

# 1. CSV files save aana folder
folder_path = os.getcwd()

# 2. State master CSV files mattum select pannu
csv_files = glob.glob(
    os.path.join(folder_path, "*_master_2018_2026.csv")
)

print("Master CSV files found:", len(csv_files))

# 3. Combine all state files
dataframes = []

for file in csv_files:
    try:
        temp_df = pd.read_csv(file)

        required_columns = [
            "State", "Year", "Quarter", "Category", "Transactions"
        ]

        if all(col in temp_df.columns for col in required_columns):
            dataframes.append(temp_df)
            print("Loaded:", os.path.basename(file))
        else:
            print("Skipped - columns mismatch:", os.path.basename(file))

    except Exception as error:
        print("Error:", os.path.basename(file), error)

# 4. Create India master dataset
if dataframes:
    india_df = pd.concat(dataframes, ignore_index=True)

    # Remove exact duplicate rows, if any
    india_df = india_df.drop_duplicates()

    print("\n========== INDIA MASTER DATA ==========")
    display(india_df.head(10))

    print("\nTotal rows:", len(india_df))
    print("States found:", india_df["State"].nunique())
    print("State names:", sorted(india_df["State"].unique()))

    # 5. State-wise transaction totals
    state_totals = (
        india_df.groupby("State")["Transactions"]
        .sum()
        .sort_values(ascending=False)
    )

    print("\n========== STATE-WISE TOTALS ==========")
    display(state_totals.reset_index())

    print("Top state:", state_totals.idxmax())
    print("Transactions:", f"{int(state_totals.max()):,}")

    # 6. State comparison chart
    state_totals.plot(
        kind="bar",
        figsize=(15, 6),
        title="India - State-wise PhonePe Transactions"
    )
    import matplotlib.pyplot as plt
    plt.xlabel("State")
    plt.ylabel("Transactions")
    plt.xticks(rotation=75, ha="right")
    plt.tight_layout()
    plt.show()

    # 7. Category-wise India totals
    category_totals = (
        india_df.groupby("Category")["Transactions"]
        .sum()
        .sort_values(ascending=False)
    )

    print("\n========== INDIA CATEGORY TOTALS ==========")
    display(category_totals.reset_index())

    print("Top category:", category_totals.idxmax())

    # 8. Year-wise India totals
    yearly_totals = (
        india_df.groupby("Year")["Transactions"]
        .sum()
        .sort_index()
    )

    print("\n========== INDIA YEARLY TOTALS ==========")
    display(yearly_totals.reset_index())

    yearly_totals.plot(
        kind="bar",
        figsize=(12, 5),
        title="India - Year-wise PhonePe Transactions"
    )
    plt.xlabel("Year")
    plt.ylabel("Transactions")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    # 9. Save combined datasets
    india_df.to_csv("india_master_2018_2026.csv", index=False)
    state_totals.to_csv("india_state_totals_2018_2026.csv")
    category_totals.to_csv("india_category_totals_2018_2026.csv")
    yearly_totals.to_csv("india_yearly_totals_2018_2026.csv")

    # 10. Verify saved file
    saved_india_df = pd.read_csv("india_master_2018_2026.csv")

    print("\n========== VERIFICATION ==========")
    print("Original rows:", len(india_df))
    print("Saved rows:", len(saved_india_df))
    print("Verification:", len(india_df) == len(saved_india_df))
    print("\nIndia-level analysis completed!")

else:
    print("Master CSV files kidaikkala.")
    print("State master CSV files save aana folder-la notebook open panni run pannu.")
```

```python

import os
import json
import pandas as pd

base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state"

states_to_process = {
    "andaman-&-nicobar-islands": "Andaman & Nicobar Islands",
    "andhra-pradesh": "Andhra Pradesh",
    "arunachal-pradesh": "Arunachal Pradesh",
    "assam": "Assam",
    "bihar": "Bihar",
    "chhattisgarh": "Chhattisgarh",
    "dadra-&-nagar-haveli-&-daman-&-diu": "Dadra & Nagar Haveli and Daman & Diu"
}

for state_folder, state_name in states_to_process.items():

    slug = state_folder.replace("-", "_").replace("&", "").replace("__", "_")
    slug = "_".join(slug.split())

    output_file = f"{slug}_master_2018_2026.csv"

    # Existing file irundha skip pannuvom
    if os.path.exists(output_file):
        print("Already exists, skipped:", output_file)
        continue

    state_path = os.path.join(base_path, state_folder)

    if not os.path.isdir(state_path):
        print("State folder not found:", state_folder)
        continue

    rows = []

    for year in range(2018, 2027):
        for quarter in range(1, 5):
            json_path = os.path.join(
                state_path, str(year), f"{quarter}.json"
            )

            if not os.path.exists(json_path):
                continue

            try:
                with open(json_path, "r", encoding="utf-8") as f:
                    data = json.load(f)

                transaction_data = (
                    data.get("data", {}).get("transactionData", [])
                )

                for item in transaction_data:
                    category = item.get("name", "Unknown")

                    count = sum(
                        instrument.get("count", 0)
                        for instrument in item.get(
                            "paymentInstruments", []
                        )
                    )

                    rows.append({
                        "State": state_name,
                        "Year": year,
                        "Quarter": f"Q{quarter}",
                        "Category": category,
                        "Transactions": count
                    })

            except (json.JSONDecodeError, OSError) as error:
                print("Error reading:", json_path, error)

    if rows:
        state_df = pd.DataFrame(rows)
        state_df.to_csv(output_file, index=False)

        print(
            f"Created: {output_file} | "
            f"Rows: {len(state_df)} | "
            f"Years: {state_df['Year'].nunique()}"
        )
    else:
        print("No transaction data found:", state_name)

print("\nMissing-state processing finished!")
```

```python

import os
import glob
import pandas as pd

base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state"

# Dataset-la irukkura ella state folders
state_folders = sorted([
    name for name in os.listdir(base_path)
    if os.path.isdir(os.path.join(base_path, name))
])

# Current folder-la irukkura master CSV files
csv_files = glob.glob("*_master_2018_2026.csv")

# CSV filename -> state data mapping
csv_map = {}

for file in csv_files:
    filename = os.path.basename(file)

    if filename.startswith(("india_", "phonepe_india_")):
        continue

    slug = filename.replace("_master_2018_2026.csv", "")
    csv_map[slug] = file

results = []

for state in state_folders:
    # State folder name and CSV slug compare
    expected_slug = state.replace("-", "_").replace("&", "")
    expected_slug = "_".join(expected_slug.split())

    # Known filename variations
    aliases = {
        "andaman_&_nicobar_islands": "andaman_nicobar_islands",
        "dadra_&_nagar_haveli_&_daman_&_diu":
            "dadra_nagar_haveli_daman_diu",
        "jammu_&_kashmir": "jammu_kashmir",
        "tamil_nadu": "tamil_nadu",
        "uttar_pradesh": "uttar_pradesh",
        "west_bengal": "west_bengal",
        "madhya_pradesh": "madhya_pradesh",
        "himachal_pradesh": "himachal_pradesh",
        "andhra_pradesh": "andhra_pradesh",
        "arunachal_pradesh": "arunachal_pradesh"
    }

    slug = aliases.get(state, expected_slug)
    csv_file = csv_map.get(slug)

    if not csv_file:
        results.append({
            "State_Folder": state,
            "CSV_Status": "MISSING",
            "Rows": 0,
            "Years": 0,
            "Transactions": 0
        })
        continue

    try:
        df = pd.read_csv(csv_file)

        required = {
            "State", "Year", "Quarter", "Category", "Transactions"
        }

        if not required.issubset(df.columns):
            status = "INVALID COLUMNS"
            rows = years = total = 0
        else:
            status = "OK" if not df.empty else "EMPTY"
            rows = len(df)
            years = df["Year"].nunique()
            total = df["Transactions"].sum()

        results.append({
            "State_Folder": state,
            "CSV_Status": status,
            "Rows": rows,
            "Years": years,
            "Transactions": total
        })

    except Exception as e:
        results.append({
            "State_Folder": state,
            "CSV_Status": "READ ERROR",
            "Rows": 0,
            "Years": 0,
            "Transactions": 0
        })

check_df = pd.DataFrame(results)

print("Total dataset folders:", len(state_folders))
print("CSV found:", (check_df["CSV_Status"] == "OK").sum())
print("Missing / needs checking:", (check_df["CSV_Status"] != "OK").sum())

display(check_df)

```

```python

import os
import glob
import pandas as pd
from IPython.display import display

required_columns = [
    "State", "Year", "Quarter", "Category", "Transactions"
]

# State master files collect pannuvom
all_files = glob.glob("*_master_2018_2026.csv")

# Existing combined India files exclude pannuvom
state_files = [
    f for f in all_files
    if not os.path.basename(f).startswith("india_")
    and not os.path.basename(f).startswith("phonepe_india_")
]

dataframes = []
loaded_files = []

for file in state_files:
    try:
        df = pd.read_csv(file)

        if all(col in df.columns for col in required_columns):
            dataframes.append(df[required_columns])
            loaded_files.append(file)
        else:
            print("Skipped - columns mismatch:", file)

    except Exception as e:
        print("Error reading:", file, e)

if dataframes:
    india_df = pd.concat(dataframes, ignore_index=True)
    india_df = india_df.drop_duplicates()

    # State totals
    state_totals = (
        india_df.groupby("State", as_index=False)["Transactions"]
        .sum()
        .sort_values("Transactions", ascending=False)
    )

    # Year totals
    yearly_totals = (
        india_df.groupby("Year", as_index=False)["Transactions"]
        .sum()
        .sort_values("Year")
    )

    # Category totals
    category_totals = (
        india_df.groupby("Category", as_index=False)["Transactions"]
        .sum()
        .sort_values("Transactions", ascending=False)
    )

    # Save final files
    india_df.to_csv("phonepe_india_master_2018_2026.csv", index=False)
    state_totals.to_csv("phonepe_india_state_totals_2018_2026.csv", index=False)
    yearly_totals.to_csv("phonepe_india_yearly_totals_2018_2026.csv", index=False)
    category_totals.to_csv("phonepe_india_category_totals_2018_2026.csv", index=False)

    print("========== FINAL DATASET ==========")
    print("State CSV files loaded:", len(loaded_files))
    print("Unique states/UT names:", india_df["State"].nunique())
    print("Total rows:", len(india_df))

    print("\nTop 10 states:")
    display(state_totals.head(10))

    print("\nYearly totals:")
    display(yearly_totals)

    print("\nCategory totals:")
    display(category_totals)

    print("\nSaved final India datasets successfully!")

else:
    print("No valid state master files found.")
```

```python
import os
import json
import re
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# STEP 1: MERCHANT INDIA PATH
# ============================================================

base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\merchant\country\india"

output_path = os.path.join(
    base_path, "merchant_analysis_2018_2026"
)

chart_path = os.path.join(output_path, "charts")
year_comparison_path = os.path.join(
    output_path, "year_wise_comparisons"
)

os.makedirs(chart_path, exist_ok=True)
os.makedirs(year_comparison_path, exist_ok=True)

years = list(range(2018, 2027))
quarters = ["Q1", "Q2", "Q3", "Q4"]

print("MERCHANT INDIA ANALYSIS")
print("Years:", years)
print("Quarters:", quarters)
print("Folder exists:", os.path.exists(base_path))


# ============================================================
# STEP 2: NESTED JSON DATA-VAI FLATTEN PANNA FUNCTION
# ============================================================

def flatten_json(data, prefix=""):
    result = {}

    if isinstance(data, dict):

        for key, value in data.items():
            new_key = f"{prefix}_{key}" if prefix else str(key)

            result.update(
                flatten_json(value, new_key)
            )

    elif isinstance(data, list):

        for index, value in enumerate(data):
            new_key = f"{prefix}_{index}" if prefix else str(index)

            result.update(
                flatten_json(value, new_key)
            )

    else:

        if prefix:
            result[prefix] = data

    return result


# ============================================================
# STEP 3: 2018-2026 Q1, Q2, Q3, Q4 JSON FILES READ
# ============================================================

rows = []
missing_files = []
read_errors = []

for year in years:

    for quarter_number in range(1, 5):

        file_path = os.path.join(
            base_path,
            str(year),
            f"{quarter_number}.json"
        )

        if not os.path.exists(file_path):

            missing_files.append(
                f"{year}-Q{quarter_number}"
            )
            continue

        try:

            with open(
                file_path,
                "r",
                encoding="utf-8"
            ) as file:

                json_data = json.load(file)

            actual_data = json_data.get(
                "data", json_data
            )

            row = flatten_json(actual_data)

            row["Year"] = year
            row["Quarter"] = f"Q{quarter_number}"

            rows.append(row)

            print(
                f"Read successfully: {year} Q{quarter_number}"
            )

        except Exception as error:

            read_errors.append({
                "Year": year,
                "Quarter": f"Q{quarter_number}",
                "Error": str(error)
            })


if len(rows) == 0:
    raise ValueError(
        "JSON files read aagala. Merchant folder path-ai check pannu."
    )


# ============================================================
# STEP 4: MASTER DATAFRAME CREATE
# ============================================================

master_df = pd.DataFrame(rows)

first_columns = ["Year", "Quarter"]

other_columns = [
    column
    for column in master_df.columns
    if column not in first_columns
]

master_df = master_df[
    first_columns + other_columns
]

master_df["Year"] = pd.to_numeric(
    master_df["Year"], errors="coerce"
)

print("\n========== MASTER DATA ==========")
print("Rows:", master_df.shape[0])
print("Columns:", master_df.shape[1])
print("Years available:", sorted(
    master_df["Year"].dropna().unique().tolist()
))
print("Quarters available:", sorted(
    master_df["Quarter"].unique().tolist()
))

display(master_df.head(10))


# ============================================================
# STEP 5: NUMERIC MERCHANT METRICS IDENTIFY
# ============================================================

metric_columns = []

for column in master_df.columns:

    if column in ["Year", "Quarter"]:
        continue

    # Metadata columns-ai comparison metrics-ah edukkama iruppom
    lower_name = column.lower()

    if any(word in lower_name for word in [
        "timestamp", "datetime"
    ]):
        continue

    if lower_name.endswith("_from") or lower_name.endswith("_to"):
        continue

    converted = pd.to_numeric(
        master_df[column], errors="coerce"
    )

    if converted.notna().any():

        master_df[column] = converted
        metric_columns.append(column)


print("\n========== NUMERIC METRICS ==========")
print("Total numeric metrics:", len(metric_columns))
print(metric_columns)

if len(metric_columns) == 0:
    raise ValueError(
        "Numeric metrics kidaikkala. JSON structure inspect pannanum."
    )


# ============================================================
# STEP 6: EACH YEAR - Q1 VS Q2, Q2 VS Q3, Q3 VS Q4
# ============================================================

def calculate_difference(current_value, previous_value):

    if pd.notna(current_value) and pd.notna(previous_value):
        return current_value - previous_value

    return None


def calculate_percentage(current_value, previous_value):

    if (
        pd.notna(current_value)
        and pd.notna(previous_value)
        and previous_value != 0
    ):
        return (
            (current_value - previous_value)
            / previous_value
        ) * 100

    return None


def create_quarter_comparison(data, year, metrics):

    year_df = data[data["Year"] == year]

    comparison_rows = []

    for metric in metrics:

        quarter_values = {}

        for quarter in quarters:

            selected = year_df.loc[
                year_df["Quarter"] == quarter,
                metric
            ]

            if not selected.empty:
                quarter_values[quarter] = selected.iloc[0]
            else:
                quarter_values[quarter] = None

        q1 = quarter_values["Q1"]
        q2 = quarter_values["Q2"]
        q3 = quarter_values["Q3"]
        q4 = quarter_values["Q4"]

        comparison_rows.append({

            "Year": year,
            "Metric": metric,

            "Q1": q1,
            "Q2": q2,
            "Q3": q3,
            "Q4": q4,

            "Q2_minus_Q1": calculate_difference(q2, q1),
            "Q2_change_percent": calculate_percentage(q2, q1),

            "Q3_minus_Q2": calculate_difference(q3, q2),
            "Q3_change_percent": calculate_percentage(q3, q2),

            "Q4_minus_Q3": calculate_difference(q4, q3),
            "Q4_change_percent": calculate_percentage(q4, q3),

            "Q4_minus_Q1": calculate_difference(q4, q1)
        })

    return pd.DataFrame(comparison_rows)


all_year_comparisons = []

for year in years:

    year_data = master_df[
        master_df["Year"] == year
    ]

    if year_data.empty:
        print(f"\n{year}: Data illa; skipped.")
        continue

    year_comparison = create_quarter_comparison(
        master_df, year, metric_columns
    )

    all_year_comparisons.append(year_comparison)

    print(f"\n========== {year} Q1-Q4 COMPARISON ==========")
    display(year_comparison)

    # Ovvoru year-um separate CSV
    year_csv = os.path.join(
        year_comparison_path,
        f"merchant_{year}_quarter_comparison.csv"
    )

    year_comparison.to_csv(
        year_csv, index=False
    )

    print("Saved:", year_csv)


# All years quarter comparison-ai combine pannuvom
if all_year_comparisons:

    all_year_comparison_df = pd.concat(
        all_year_comparisons,
        ignore_index=True
    )

else:
    all_year_comparison_df = pd.DataFrame()


# ============================================================
# STEP 7: OVERALL 2018-2026 COMPARISON
# SAME QUARTER BETWEEN YEARS
# ============================================================

overall_rows = []

available_years = sorted(
    master_df["Year"].dropna().astype(int).unique().tolist()
)

for metric in metric_columns:

    for quarter in quarters:

        previous_year_value = None
        previous_year = None

        for year in available_years:

            selected = master_df.loc[
                (master_df["Year"] == year)
                & (master_df["Quarter"] == quarter),
                metric
            ]

            current_value = (
                selected.iloc[0]
                if not selected.empty
                else None
            )

            overall_rows.append({

                "Metric": metric,
                "Quarter": quarter,
                "Year": year,
                "Value": current_value,

                "Previous_Year": previous_year,

                "Previous_Year_Value": previous_year_value,

                "Difference_From_Previous_Year":
                    calculate_difference(
                        current_value,
                        previous_year_value
                    ),

                "Change_Percent_From_Previous_Year":
                    calculate_percentage(
                        current_value,
                        previous_year_value
                    )
            })

            previous_year = year
            previous_year_value = current_value


overall_comparison_df = pd.DataFrame(overall_rows)

print("\n========== OVERALL 2018-2026 COMPARISON ==========")
print(
    "Same quarter across years; previous-year difference included."
)
display(overall_comparison_df.head(30))


# ============================================================
# STEP 8: BEST OBSERVED YEAR & QUARTER FOR EACH METRIC
# ============================================================

best_rows = []

for metric in metric_columns:

    valid_rows = master_df[
        master_df[metric].notna()
    ]

    if valid_rows.empty:
        continue

    best_index = valid_rows[metric].idxmax()
    best_row = valid_rows.loc[best_index]

    best_rows.append({

        "Metric": metric,
        "Highest_Value": best_row[metric],
        "Year": int(best_row["Year"]),
        "Quarter": best_row["Quarter"]

    })


best_comparison_df = pd.DataFrame(best_rows)

print("\n========== HIGHEST OBSERVED VALUE PER METRIC ==========")
display(best_comparison_df)


# ============================================================
# STEP 9: SAVE MASTER & ALL COMPARISON CSV FILES
# ============================================================

master_file = os.path.join(
    output_path,
    "merchant_india_master_2018_2026.csv"
)

all_years_file = os.path.join(
    output_path,
    "merchant_india_quarter_comparison_2018_2026.csv"
)

overall_file = os.path.join(
    output_path,
    "merchant_india_overall_year_comparison.csv"
)

best_file = os.path.join(
    output_path,
    "merchant_india_highest_observed_values.csv"
)

master_df.to_csv(master_file, index=False)

all_year_comparison_df.to_csv(
    all_years_file, index=False
)

overall_comparison_df.to_csv(
    overall_file, index=False
)

best_comparison_df.to_csv(
    best_file, index=False
)

print("\n========== CSV SAVE STATUS ==========")

for file_path in [
    master_file,
    all_years_file,
    overall_file,
    best_file
]:
    print(
        os.path.basename(file_path),
        "Saved:",
        os.path.exists(file_path)
    )


# ============================================================
# STEP 10: EACH YEAR & EACH METRIC CHARTS SAVE
# ============================================================

yearly_charts_created = 0

for year in available_years:

    year_df = master_df[
        master_df["Year"] == year
    ]

    for metric in metric_columns:

        values = []

        for quarter in quarters:

            selected = year_df.loc[
                year_df["Quarter"] == quarter,
                metric
            ]

            values.append(
                selected.iloc[0]
                if not selected.empty
                else None
            )

        plot_df = pd.DataFrame({
            "Quarter": quarters,
            "Value": values
        })

        plot_df["Value"] = pd.to_numeric(
            plot_df["Value"], errors="coerce"
        )

        plot_df = plot_df.dropna()

        if plot_df.empty:
            continue

        safe_metric = re.sub(
            r"[^A-Za-z0-9_-]+",
            "_",
            metric
        )

        plt.figure(figsize=(8, 5))

        plt.plot(
            plot_df["Quarter"],
            plot_df["Value"],
            marker="o",
            linewidth=2
        )

        plt.title(
            f"Merchant India - {metric} - {year}"
        )
        plt.xlabel("Quarter")
        plt.ylabel(metric)
        plt.grid(True, linestyle="--", alpha=0.5)
        plt.tight_layout()

        year_chart_folder = os.path.join(
            chart_path, str(year)
        )

        os.makedirs(
            year_chart_folder, exist_ok=True
        )

        chart_file = os.path.join(
            year_chart_folder,
            f"{safe_metric}_{year}.png"
        )

        plt.savefig(chart_file, dpi=150)
        plt.close()

        yearly_charts_created += 1


# ============================================================
# STEP 11: OVERALL CHARTS - 2018 TO 2026
# EACH YEAR GETS ITS OWN LINE
# ============================================================

overall_charts_created = 0

for metric in metric_columns:

    plt.figure(figsize=(10, 6))

    has_values = False

    for year in available_years:

        year_df = master_df[
            master_df["Year"] == year
        ]

        values = []

        for quarter in quarters:

            selected = year_df.loc[
                year_df["Quarter"] == quarter,
                metric
            ]

            values.append(
                selected.iloc[0]
                if not selected.empty
                else None
            )

        values = pd.to_numeric(
            pd.Series(values), errors="coerce"
        )

        if values.notna().any():

            plt.plot(
                quarters,
                values,
                marker="o",
                linewidth=1.5,
                label=str(year)
            )

            has_values = True

    if has_values:

        safe_metric = re.sub(
            r"[^A-Za-z0-9_-]+",
            "_",
            metric
        )

        plt.title(
            f"Merchant India - {metric} - Quarterly Comparison 2018-2026"
        )
        plt.xlabel("Quarter")
        plt.ylabel(metric)
        plt.legend(title="Year")
        plt.grid(True, linestyle="--", alpha=0.5)
        plt.tight_layout()

        chart_file = os.path.join(
            chart_path,
            f"{safe_metric}_overall_2018_2026.png"
        )

        plt.savefig(chart_file, dpi=150)
        plt.show()
        plt.close()

        overall_charts_created += 1

    else:
        plt.close()


# ============================================================
# STEP 12: VERIFY SAVED FILES BY READING THEM AGAIN
# ============================================================

verified_master = pd.read_csv(master_file)

verified_year_comparison = pd.read_csv(
    all_years_file
)

verified_overall = pd.read_csv(
    overall_file
)

verified_best = pd.read_csv(
    best_file
)

print("\n========== FINAL VERIFICATION ==========")

print("Master rows:", verified_master.shape[0])
print("Master columns:", verified_master.shape[1])

print(
    "Years found:",
    sorted(verified_master["Year"].dropna().unique().tolist())
)

print(
    "Quarters found:",
    sorted(verified_master["Quarter"].unique().tolist())
)

print(
    "Year-wise comparison rows:",
    verified_year_comparison.shape[0]
)

print(
    "Overall comparison rows:",
    verified_overall.shape[0]
)

print(
    "Best-value comparison rows:",
    verified_best.shape[0]
)

print("Year-wise charts saved:", yearly_charts_created)
print("Overall charts saved:", overall_charts_created)

print("\nMissing files:", missing_files)
print("Read errors:", read_errors)

print("\nOutput folder:", output_path)
print("Charts folder:", chart_path)

print("\nVerified master preview:")
display(verified_master.head())

print("\nVerified overall comparison preview:")
display(verified_overall.head())

print("\nVerified highest-value preview:")
display(verified_best.head())
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# ==================================================
# STEP 1: EXISTING MERCHANT MASTER CSV
# ==================================================

base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\merchant\country\india"

output_path = os.path.join(
    base_path, "merchant_analysis_2018_2026"
)

master_file = os.path.join(
    output_path, "merchant_india_master_2018_2026.csv"
)

chart_path = os.path.join(output_path, "charts")
os.makedirs(chart_path, exist_ok=True)

# Read existing processed data
df = pd.read_csv(master_file)

df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df["Quarter"] = df["Quarter"].astype(str)

quarter_order = ["Q1", "Q2", "Q3", "Q4"]

metric_columns = [
    col for col in df.select_dtypes(include="number").columns
    if col != "Year" and df[col].notna().any()
]

if not metric_columns:
    raise ValueError("Numeric metric kidaikkala. Master CSV check pannu.")

print("Master CSV loaded:", True)
print("Rows:", len(df))
print("Metrics:", metric_columns)


# ==================================================
# STEP 2: PROFESSIONAL DARK THEME
# ==================================================

BG = "#0B1020"
PANEL = "#111A2E"
GRID = "#26344D"
TEXT = "#E6EDF7"
MUTED = "#9FB0C8"
CYAN = "#38BDF8"
BLUE = "#6366F1"

YEAR_COLORS = [
    "#38BDF8", "#A78BFA", "#34D399",
    "#FBBF24", "#FB7185", "#22D3EE",
    "#C084FC", "#4ADE80", "#F97316"
]

def format_count(value, position=None):
    if abs(value) >= 1_000_000:
        return f"{value / 1_000_000:.1f}M"
    if abs(value) >= 1_000:
        return f"{value / 1_000:.0f}K"
    return f"{value:.0f}"


def style_axis(ax):
    ax.set_facecolor(PANEL)
    ax.tick_params(colors=MUTED, labelsize=10)
    ax.xaxis.label.set_color(TEXT)
    ax.yaxis.label.set_color(TEXT)
    ax.title.set_color(TEXT)

    for spine in ax.spines.values():
        spine.set_color(GRID)

    ax.grid(
        True,
        color=GRID,
        linestyle="--",
        linewidth=0.7,
        alpha=0.75
    )

    ax.set_axisbelow(True)
    ax.yaxis.set_major_formatter(FuncFormatter(format_count))


def style_figure(fig, ax, title, subtitle):
    fig.patch.set_facecolor(BG)

    ax.set_title(
        title,
        fontsize=17,
        fontweight="bold",
        color=TEXT,
        loc="left",
        pad=24
    )

    ax.text(
        0,
        1.02,
        subtitle,
        transform=ax.transAxes,
        fontsize=10,
        color=MUTED,
        va="bottom"
    )

    style_axis(ax)


# ==================================================
# STEP 3: YEAR-WISE Q1-Q4 DARK CHARTS
# ==================================================

charts_created = 0

available_years = sorted(
    df["Year"].dropna().astype(int).unique()
)

for year in available_years:

    year_df = df[df["Year"] == year].copy()
    year_folder = os.path.join(chart_path, str(year))
    os.makedirs(year_folder, exist_ok=True)

    for metric in metric_columns:

        quarter_values = []

        for quarter in quarter_order:
            selected = year_df.loc[
                year_df["Quarter"] == quarter,
                metric
            ]

            quarter_values.append(
                selected.iloc[0] if not selected.empty else np.nan
            )

        values = np.array(quarter_values, dtype=float)

        if np.isnan(values).all():
            continue

        fig, ax = plt.subplots(figsize=(12, 6.5))

        x = np.arange(len(quarter_order))

        # Soft glow effect
        ax.plot(
            x, values,
            color=CYAN,
            linewidth=7,
            alpha=0.10,
            solid_capstyle="round"
        )

        # Main line
        ax.plot(
            x, values,
            color=CYAN,
            linewidth=2.8,
            marker="o",
            markersize=8,
            markerfacecolor=BG,
            markeredgecolor=CYAN,
            markeredgewidth=2.2,
            label="Registered Merchants"
        )

        # Values on data points
        for i, value in enumerate(values):
            if not np.isnan(value):
                ax.annotate(
                    f"{value:,.0f}",
                    (x[i], value),
                    textcoords="offset points",
                    xytext=(0, 13),
                    ha="center",
                    fontsize=9,
                    color=TEXT,
                    fontweight="bold"
                )

        ax.set_xticks(x)
        ax.set_xticklabels(
            quarter_order,
            color=TEXT,
            fontsize=11
        )

        ax.set_xlabel("Quarter", labelpad=12)
        ax.set_ylabel("Registered Merchant Count", labelpad=12)

        style_figure(
            fig,
            ax,
            f"MERCHANT PERFORMANCE | {year}",
            "INDIA  •  AGGREGATED MERCHANT DATA  •  QUARTERLY TREND"
        )

        ax.legend(
            facecolor=PANEL,
            edgecolor=GRID,
            labelcolor=TEXT,
            loc="upper left"
        )

        # Leave room for title and value labels
        ax.margins(y=0.22)
        fig.tight_layout(pad=2.2)

        safe_metric = "".join(
            char if char.isalnum() or char in "_-" else "_"
            for char in metric
        )

        file_name = f"{safe_metric}_{year}.png"
        save_file = os.path.join(year_folder, file_name)

        fig.savefig(
            save_file,
            dpi=220,
            facecolor=fig.get_facecolor(),
            bbox_inches="tight"
        )

        plt.close(fig)
        charts_created += 1

        print("Saved:", save_file)


# ==================================================
# STEP 4: OVERALL 2018-2026 COMPARISON CHART
# ==================================================

for metric in metric_columns:

    fig, ax = plt.subplots(figsize=(13, 7))

    x = np.arange(4)
    plotted_years = 0

    for index, year in enumerate(available_years):

        year_df = df[df["Year"] == year]

        values = []

        for quarter in quarter_order:
            selected = year_df.loc[
                year_df["Quarter"] == quarter,
                metric
            ]

            values.append(
                selected.iloc[0] if not selected.empty else np.nan
            )

        values = np.array(values, dtype=float)

        if np.isnan(values).all():
            continue

        color = YEAR_COLORS[index % len(YEAR_COLORS)]

        ax.plot(
            x,
            values,
            color=color,
            linewidth=2.2,
            marker="o",
            markersize=5.5,
            alpha=0.95,
            label=str(year)
        )

        plotted_years += 1

    ax.set_xticks(x)
    ax.set_xticklabels(
        quarter_order,
        color=TEXT,
        fontsize=11
    )

    ax.set_xlabel("Quarter", labelpad=12)
    ax.set_ylabel("Registered Merchant Count", labelpad=12)

    style_figure(
        fig,
        ax,
        "INDIA MERCHANTS | 2018–2026",
        "YEAR-ON-YEAR QUARTERLY COMPARISON  •  DARK ANALYTICS THEME"
    )

    ax.legend(
        title="Year",
        ncol=3,
        facecolor=PANEL,
        edgecolor=GRID,
        labelcolor=TEXT,
        title_fontsize=10,
        fontsize=9,
        loc="upper left"
    )

    ax.margins(y=0.12)
    fig.tight_layout(pad=2.2)

    safe_metric = "".join(
        char if char.isalnum() or char in "_-" else "_"
        for char in metric
    )

    overall_file = os.path.join(
        chart_path,
        f"{safe_metric}_overall_2018_2026.png"
    )

    fig.savefig(
        overall_file,
        dpi=240,
        facecolor=fig.get_facecolor(),
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)

    charts_created += 1

    print("Overall chart saved:", overall_file)


# ==================================================
# STEP 5: VERIFY DARK CHARTS
# ==================================================

print("\n========== DARK THEME CHART VERIFICATION ==========")
print("Master CSV exists:", os.path.exists(master_file))
print("Year-wise charts created:", charts_created - len(metric_columns))
print("Overall charts created:", len(metric_columns))
print("Chart folder exists:", os.path.exists(chart_path))

for year in available_years:
    year_folder = os.path.join(chart_path, str(year))

    if os.path.exists(year_folder):
        png_files = [
            file for file in os.listdir(year_folder)
            if file.endswith(".png")
        ]
        print(f"{year}: {len(png_files)} chart(s) saved")

print("\nAll dark-theme charts saved in:")
print(chart_path)
```

```python
import os
import json
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# ============================================================
# STEP 1: PATHS AND FOLDERS
# ============================================================

india_path = (
    r"C:\Users\HP\OneDrive\Documents\fita project"
    r"\pulse-main\pulse-main\data\aggregated\merchant\country\india"
)

state_path = os.path.join(india_path, "state")

output_path = os.path.join(
    india_path, "merchant_state_analysis_2018_2026"
)

state_master_path = os.path.join(
    output_path, "state_masters"
)

comparison_path = os.path.join(
    output_path, "year_wise_comparisons"
)

chart_path = os.path.join(
    output_path, "charts"
)

for folder in [
    output_path,
    state_master_path,
    comparison_path,
    chart_path
]:
    os.makedirs(folder, exist_ok=True)

years = list(range(2018, 2027))
quarters = ["Q1", "Q2", "Q3", "Q4"]

print("MERCHANT STATE ANALYSIS")
print("State folder exists:", os.path.exists(state_path))


# ============================================================
# STEP 2: FIND ALL STATE/UT FOLDERS
# ============================================================

state_folders = sorted([
    folder
    for folder in os.listdir(state_path)
    if os.path.isdir(os.path.join(state_path, folder))
])

def format_state_name(folder_name):
    name = folder_name.replace("-", " ").title()

    # Common official-style names
    replacements = {
        "Andaman & Nicobar Islands": "Andaman & Nicobar Islands",
        "Dadra & Nagar Haveli & Daman & Diu":
            "Dadra and Nagar Haveli and Daman and Diu",
        "Jammu & Kashmir": "Jammu & Kashmir"
    }

    return replacements.get(name, name)


print("Total state/UT folders:", len(state_folders))
print(state_folders)


# ============================================================
# STEP 3: READ EVERY STATE, YEAR AND QUARTER JSON
# ============================================================

all_rows = []
missing_files = []
read_errors = []

for state_folder in state_folders:

    state_name = format_state_name(state_folder)

    for year in years:

        for q_num in range(1, 5):

            file_path = os.path.join(
                state_path,
                state_folder,
                str(year),
                f"{q_num}.json"
            )

            if not os.path.exists(file_path):
                missing_files.append({
                    "State": state_name,
                    "Year": year,
                    "Quarter": f"Q{q_num}",
                    "File": file_path
                })
                continue

            try:

                with open(
                    file_path,
                    "r",
                    encoding="utf-8"
                ) as file:
                    json_data = json.load(file)

                payload = json_data.get("data", {})
                aggregated = payload.get("aggregated", {})

                registered_count = aggregated.get(
                    "registeredCount"
                )

                if registered_count is None:
                    read_errors.append({
                        "State": state_name,
                        "Year": year,
                        "Quarter": f"Q{q_num}",
                        "Error": "registeredCount field missing"
                    })
                    continue

                all_rows.append({
                    "State": state_name,
                    "StateFolder": state_folder,
                    "Year": year,
                    "Quarter": f"Q{q_num}",
                    "QuarterNumber": q_num,
                    "RegisteredMerchants": registered_count
                })

            except Exception as error:

                read_errors.append({
                    "State": state_name,
                    "Year": year,
                    "Quarter": f"Q{q_num}",
                    "Error": str(error)
                })


if not all_rows:
    raise ValueError(
        "Merchant state JSON data read aagala. Folder path check pannu."
    )


# ============================================================
# STEP 4: CREATE COMBINED MASTER DATAFRAME
# ============================================================

master_df = pd.DataFrame(all_rows)

master_df["Year"] = pd.to_numeric(
    master_df["Year"], errors="coerce"
)

master_df["QuarterNumber"] = pd.to_numeric(
    master_df["QuarterNumber"], errors="coerce"
)

master_df["RegisteredMerchants"] = pd.to_numeric(
    master_df["RegisteredMerchants"], errors="coerce"
)

master_df = master_df.sort_values(
    ["State", "Year", "QuarterNumber"]
).reset_index(drop=True)

print("\n========== MERCHANT STATE MASTER ==========")
print("Rows:", len(master_df))
print("Columns:", master_df.columns.tolist())
print("States/UTs with data:", master_df["State"].nunique())
print(
    "Years available:",
    sorted(master_df["Year"].dropna().unique().tolist())
)

display(master_df.head(10))


# ============================================================
# STEP 5: SAVE INDIVIDUAL STATE MASTER CSV FILES
# ============================================================

individual_files_created = 0

for state_folder in state_folders:

    state_name = format_state_name(state_folder)

    state_df = master_df[
        master_df["StateFolder"] == state_folder
    ].copy()

    state_df = state_df[
        [
            "State",
            "Year",
            "Quarter",
            "QuarterNumber",
            "RegisteredMerchants"
        ]
    ]

    safe_name = re.sub(
        r"[^A-Za-z0-9_-]+",
        "_",
        state_folder
    )

    state_csv = os.path.join(
        state_master_path,
        f"{safe_name}_merchant_master_2018_2026.csv"
    )

    state_df.to_csv(state_csv, index=False)
    individual_files_created += 1

print("\nIndividual state master CSVs created:", individual_files_created)


# ============================================================
# STEP 6: QUARTER-WISE COMPARISON FOR EACH STATE AND YEAR
# Q1 vs Q2, Q2 vs Q3, Q3 vs Q4
# ============================================================

def calculate_difference(current, previous):
    if pd.notna(current) and pd.notna(previous):
        return current - previous
    return np.nan


def calculate_percentage(current, previous):
    if (
        pd.notna(current)
        and pd.notna(previous)
        and previous != 0
    ):
        return (current - previous) / previous * 100
    return np.nan


comparison_rows = []

for state_folder in state_folders:

    state_name = format_state_name(state_folder)

    state_df = master_df[
        master_df["StateFolder"] == state_folder
    ]

    for year in years:

        year_df = state_df[
            state_df["Year"] == year
        ]

        if year_df.empty:
            continue

        values = {}

        for quarter in quarters:

            selected = year_df.loc[
                year_df["Quarter"] == quarter,
                "RegisteredMerchants"
            ]

            values[quarter] = (
                selected.iloc[0]
                if not selected.empty
                else np.nan
            )

        q1 = values["Q1"]
        q2 = values["Q2"]
        q3 = values["Q3"]
        q4 = values["Q4"]

        comparison_rows.append({
            "State": state_name,
            "Year": year,
            "Q1": q1,
            "Q2": q2,
            "Q3": q3,
            "Q4": q4,

            "Q2_minus_Q1": calculate_difference(q2, q1),
            "Q2_change_percent": calculate_percentage(q2, q1),

            "Q3_minus_Q2": calculate_difference(q3, q2),
            "Q3_change_percent": calculate_percentage(q3, q2),

            "Q4_minus_Q3": calculate_difference(q4, q3),
            "Q4_change_percent": calculate_percentage(q4, q3),

            "Q4_minus_Q1": calculate_difference(q4, q1)
        })


comparison_df = pd.DataFrame(comparison_rows)

comparison_file = os.path.join(
    output_path,
    "merchant_state_quarter_comparison_2018_2026.csv"
)

comparison_df.to_csv(comparison_file, index=False)

print("\n========== STATE QUARTER COMPARISON ==========")
print("Comparison rows:", len(comparison_df))
display(comparison_df.head(12))


# ============================================================
# STEP 7: SAVE A SEPARATE COMPARISON CSV FOR EACH YEAR
# ============================================================

for year in years:

    year_compare_df = comparison_df[
        comparison_df["Year"] == year
    ].copy()

    year_csv = os.path.join(
        comparison_path,
        f"merchant_state_quarter_comparison_{year}.csv"
    )

    year_compare_df.to_csv(year_csv, index=False)

    print(
        year,
        "comparison rows:",
        len(year_compare_df),
        "| Saved:",
        os.path.exists(year_csv)
    )


# ============================================================
# STEP 8: CREATE Q4 YEAR-END COMPARISON BY STATE
# ============================================================

q4_df = master_df[
    master_df["Quarter"] == "Q4"
].copy()

q4_comparison_file = os.path.join(
    output_path,
    "merchant_state_q4_yearly_comparison.csv"
)

q4_df.to_csv(q4_comparison_file, index=False)

print("\n========== STATE YEAR-END Q4 DATA ==========")
display(q4_df.head(10))


# ============================================================
# STEP 9: FIND LATEST QUARTER AVAILABLE FOR ALL STATES
# FOR A FAIR STATE RANKING
# ============================================================

state_count = len(state_folders)

coverage_df = (
    master_df.dropna(subset=["RegisteredMerchants"])
    .groupby(["Year", "Quarter", "QuarterNumber"])["State"]
    .nunique()
    .reset_index(name="StateCount")
)

complete_periods = coverage_df[
    coverage_df["StateCount"] == state_count
].copy()

if not complete_periods.empty:

    selected_period = complete_periods.sort_values(
        ["Year", "QuarterNumber"]
    ).iloc[-1]

    ranking_year = int(selected_period["Year"])
    ranking_quarter = selected_period["Quarter"]

    ranking_note = "Latest quarter available for every state/UT"

else:

    # Fallback: period with the largest state coverage
    selected_period = coverage_df.sort_values(
        ["StateCount", "Year", "QuarterNumber"]
    ).iloc[-1]

    ranking_year = int(selected_period["Year"])
    ranking_quarter = selected_period["Quarter"]

    ranking_note = "Largest available coverage; some states may be missing"


ranking_df = master_df[
    (master_df["Year"] == ranking_year)
    & (master_df["Quarter"] == ranking_quarter)
].copy()

ranking_df = ranking_df.dropna(
    subset=["RegisteredMerchants"]
).sort_values(
    "RegisteredMerchants",
    ascending=False
).reset_index(drop=True)

ranking_df["Rank"] = np.arange(1, len(ranking_df) + 1)

ranking_df = ranking_df[
    [
        "Rank",
        "State",
        "Year",
        "Quarter",
        "RegisteredMerchants"
    ]
]

ranking_file = os.path.join(
    output_path,
    "merchant_state_ranking_latest_common_quarter.csv"
)

ranking_df.to_csv(ranking_file, index=False)

print("\n========== MERCHANT STATE RANKING ==========")
print("Ranking period:", ranking_year, ranking_quarter)
print("Ranking basis:", ranking_note)
print("States in ranking:", len(ranking_df))
display(ranking_df.head(10))


# ============================================================
# STEP 10: PROFESSIONAL DARK THEME SETTINGS
# ============================================================

BG = "#0B1020"
PANEL = "#111A2E"
GRID = "#26344D"
TEXT = "#E6EDF7"
MUTED = "#9FB0C8"
CYAN = "#38BDF8"
GREEN = "#34D399"

YEAR_COLORS = [
    "#38BDF8", "#A78BFA", "#34D399",
    "#FBBF24", "#FB7185", "#22D3EE",
    "#C084FC", "#4ADE80", "#F97316"
]


def apply_dark_theme(fig, ax, title, subtitle, ylabel):

    fig.patch.set_facecolor(BG)
    ax.set_facecolor(PANEL)

    ax.set_title(
        title,
        fontsize=15,
        fontweight="bold",
        color=TEXT,
        loc="left",
        pad=25
    )

    ax.text(
        0,
        1.02,
        subtitle,
        transform=ax.transAxes,
        color=MUTED,
        fontsize=9,
        va="bottom"
    )

    ax.set_xlabel("Quarter", color=TEXT, labelpad=10)
    ax.set_ylabel(ylabel, color=TEXT, labelpad=10)

    ax.tick_params(colors=MUTED)

    ax.yaxis.set_major_formatter(
        FuncFormatter(
            lambda value, position:
            f"{value / 1_000_000:.1f}M"
            if abs(value) >= 1_000_000
            else f"{value / 1_000:.0f}K"
            if abs(value) >= 1_000
            else f"{value:.0f}"
        )
    )

    ax.grid(
        True,
        color=GRID,
        linestyle="--",
        linewidth=0.7,
        alpha=0.8
    )

    for spine in ax.spines.values():
        spine.set_color(GRID)

    ax.set_axisbelow(True)


# ============================================================
# STEP 11: CREATE A DARK QUARTERLY TREND CHART FOR EACH STATE
# EACH LINE REPRESENTS A YEAR
# ============================================================

state_charts_created = 0

for state_folder in state_folders:

    state_name = format_state_name(state_folder)

    state_df = master_df[
        master_df["StateFolder"] == state_folder
    ].copy()

    if state_df.empty:
        continue

    fig, ax = plt.subplots(figsize=(11, 6.5))
    plotted_lines = 0

    for index, year in enumerate(years):

        year_df = state_df[
            state_df["Year"] == year
        ]

        values = []

        for quarter in quarters:

            selected = year_df.loc[
                year_df["Quarter"] == quarter,
                "RegisteredMerchants"
            ]

            values.append(
                selected.iloc[0]
                if not selected.empty
                else np.nan
            )

        values = np.array(values, dtype=float)

        if np.isnan(values).all():
            continue

        color = YEAR_COLORS[index % len(YEAR_COLORS)]

        ax.plot(
            range(4),
            values,
            marker="o",
            linewidth=2,
            markersize=5,
            color=color,
            label=str(year)
        )

        plotted_lines += 1

    if plotted_lines == 0:
        plt.close(fig)
        continue

    ax.set_xticks(range(4))
    ax.set_xticklabels(quarters)

    apply_dark_theme(
        fig,
        ax,
        f"MERCHANT PERFORMANCE | {state_name.upper()}",
        "AGGREGATED MERCHANT DATA  •  QUARTERLY TREND BY YEAR",
        "Registered Merchants"
    )

    ax.legend(
        title="Year",
        ncol=3,
        facecolor=PANEL,
        edgecolor=GRID,
        labelcolor=TEXT,
        title_fontsize=9,
        fontsize=8
    )

    ax.margins(y=0.15)
    fig.tight_layout(pad=2)

    safe_name = re.sub(
        r"[^A-Za-z0-9_-]+",
        "_",
        state_folder
    )

    chart_file = os.path.join(
        chart_path,
        f"{safe_name}_merchant_quarterly_2018_2026.png"
    )

    fig.savefig(
        chart_file,
        dpi=220,
        facecolor=fig.get_facecolor(),
        bbox_inches="tight"
    )

    plt.close(fig)
    state_charts_created += 1


print("\nState-wise dark charts created:", state_charts_created)


# ============================================================
# STEP 12: OVERALL STATE RANKING DARK BAR CHART
# ============================================================

if not ranking_df.empty:

    plot_df = ranking_df.head(10).sort_values(
        "RegisteredMerchants",
        ascending=True
    )

    fig, ax = plt.subplots(figsize=(12, 7))

    fig.patch.set_facecolor(BG)
    ax.set_facecolor(PANEL)

    bar_colors = [
        GREEN if state == ranking_df.iloc[0]["State"] else CYAN
        for state in plot_df["State"]
    ]

    bars = ax.barh(
        plot_df["State"],
        plot_df["RegisteredMerchants"],
        color=bar_colors,
        height=0.65
    )

    for bar, value in zip(
        bars,
        plot_df["RegisteredMerchants"]
    ):
        ax.annotate(
            f"{value:,.0f}",
            (bar.get_width(), bar.get_y() + bar.get_height() / 2),
            xytext=(7, 0),
            textcoords="offset points",
            va="center",
            color=TEXT,
            fontsize=9
        )

    ax.set_title(
        "TOP 10 STATES | REGISTERED MERCHANTS",
        fontsize=16,
        fontweight="bold",
        color=TEXT,
        loc="left",
        pad=25
    )

    ax.text(
        0,
        1.02,
        f"INDIA  •  {ranking_year} {ranking_quarter}  •  {ranking_note.upper()}",
        transform=ax.transAxes,
        color=MUTED,
        fontsize=8
    )

    ax.set_xlabel("Registered Merchants", color=TEXT, labelpad=10)

    ax.tick_params(colors=MUTED)
    ax.xaxis.set_major_formatter(
        FuncFormatter(
            lambda value, position:
            f"{value / 1_000_000:.1f}M"
            if value >= 1_000_000
            else f"{value / 1_000:.0f}K"
        )
    )

    ax.grid(
        axis="x",
        color=GRID,
        linestyle="--",
        linewidth=0.7
    )

    for spine in ax.spines.values():
        spine.set_color(GRID)

    ax.set_axisbelow(True)
    ax.margins(x=0.2)

    fig.tight_layout(pad=2)

    ranking_chart_file = os.path.join(
        chart_path,
        "merchant_state_top10_ranking_dark.png"
    )

    fig.savefig(
        ranking_chart_file,
        dpi=240,
        facecolor=fig.get_facecolor(),
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)

    print("Ranking chart saved:", ranking_chart_file)


# ============================================================
# STEP 13: TOP 10 STATE TRENDS USING YEAR-END Q4 VALUES
# ============================================================

q4_data = master_df[
    master_df["Quarter"] == "Q4"
].copy()

if not q4_data.empty and not ranking_df.empty:

    top_state_names = ranking_df.head(10)["State"].tolist()

    fig, ax = plt.subplots(figsize=(13, 7))
    fig.patch.set_facecolor(BG)
    ax.set_facecolor(PANEL)

    trends_created = 0

    for index, state_name in enumerate(top_state_names):

        state_q4 = q4_data[
            q4_data["State"] == state_name
        ].sort_values("Year")

        if state_q4.empty:
            continue

        ax.plot(
            state_q4["Year"],
            state_q4["RegisteredMerchants"],
            marker="o",
            linewidth=2,
            markersize=4,
            color=YEAR_COLORS[index % len(YEAR_COLORS)],
            label=state_name
        )

        trends_created += 1

    ax.set_title(
        "TOP STATES | YEAR-END MERCHANT TREND",
        fontsize=16,
        fontweight="bold",
        color=TEXT,
        loc="left",
        pad=25
    )

    ax.text(
        0,
        1.02,
        "INDIA  •  Q4 VALUES BY YEAR  •  2018–2026",
        transform=ax.transAxes,
        color=MUTED,
        fontsize=9
    )

    ax.set_xlabel("Year", color=TEXT, labelpad=10)
    ax.set_ylabel("Registered Merchants", color=TEXT, labelpad=10)
    ax.tick_params(colors=MUTED)

    ax.yaxis.set_major_formatter(
        FuncFormatter(
            lambda value, position: f"{value / 1_000_000:.1f}M"
        )
    )

    ax.grid(
        True,
        color=GRID,
        linestyle="--",
        linewidth=0.7
    )

    ax.legend(
        title="State",
        ncol=2,
        facecolor=PANEL,
        edgecolor=GRID,
        labelcolor=TEXT,
        title_fontsize=9,
        fontsize=8
    )

    for spine in ax.spines.values():
        spine.set_color(GRID)

    ax.set_axisbelow(True)
    fig.tight_layout(pad=2)

    trend_chart_file = os.path.join(
        chart_path,
        "merchant_top10_state_year_end_trend_dark.png"
    )

    fig.savefig(
        trend_chart_file,
        dpi=240,
        facecolor=fig.get_facecolor(),
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)

    print("Top-state trend chart saved:", trend_chart_file)


# ============================================================
# STEP 14: SAVE MISSING FILES AND READ ERRORS
# ============================================================

missing_df = pd.DataFrame(missing_files)
errors_df = pd.DataFrame(read_errors)

missing_file_report = os.path.join(
    output_path, "merchant_state_missing_quarters.csv"
)

error_file_report = os.path.join(
    output_path, "merchant_state_read_errors.csv"
)

missing_df.to_csv(missing_file_report, index=False)
errors_df.to_csv(error_file_report, index=False)


# ============================================================
# STEP 15: SAVE MASTER AND VERIFY OUTPUT FILES
# ============================================================

master_file = os.path.join(
    output_path,
    "merchant_state_master_2018_2026.csv"
)

master_df.to_csv(master_file, index=False)

verified_master = pd.read_csv(master_file)
verified_comparison = pd.read_csv(comparison_file)
verified_q4 = pd.read_csv(q4_comparison_file)
verified_ranking = pd.read_csv(ranking_file)

print("\n========== FINAL VERIFICATION ==========")

print("State folders found:", len(state_folders))
print("Individual state CSVs created:", individual_files_created)

print("Combined master rows:", len(verified_master))
print("States in master:", verified_master["State"].nunique())

print("Quarter comparison rows:", len(verified_comparison))
print("Q4 year-wise rows:", len(verified_q4))
print("Ranking rows:", len(verified_ranking))

print("State charts created:", state_charts_created)
print("Missing JSON files:", len(missing_df))
print("Read/schema errors:", len(errors_df))

print("\nMissing-file report:", missing_file_report)
print("Read-error report:", error_file_report)

print("\nOutput folder:")
print(output_path)

print("\nVerified master preview:")
display(verified_master.head(10))

print("\nVerified state ranking:")
display(verified_ranking.head(10))
```

```python
import os
import json
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# =====================================================
# STEP 1: PATHS
# =====================================================

base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\merchant\country\india\state"

output_path = os.path.join(
    base_path, "merchant_state_analysis_2018_2026"
)
state_csv_path = os.path.join(output_path, "state_masters")
compare_path = os.path.join(output_path, "year_wise_comparisons")
chart_path = os.path.join(output_path, "charts")

for folder in [output_path, state_csv_path, compare_path, chart_path]:
    os.makedirs(folder, exist_ok=True)

years = range(2018, 2027)
quarters = ["Q1", "Q2", "Q3", "Q4"]

state_folders = sorted([
    folder for folder in os.listdir(base_path)
    if os.path.isdir(os.path.join(base_path, folder))
    and folder != "merchant_state_analysis_2018_2026"
])

def state_display_name(name):
    name = name.replace("-", " ").title()
    if name == "Dadra & Nagar Haveli & Daman & Diu":
        return "Dadra and Nagar Haveli and Daman and Diu"
    return name

print("Merchant states found:", len(state_folders))


# =====================================================
# STEP 2: READ ALL STATES, YEARS, QUARTERS
# =====================================================

rows = []
missing_files = []
read_errors = []

for state_folder in state_folders:
    state_name = state_display_name(state_folder)

    for year in years:
        for q_num in range(1, 5):

            file_path = os.path.join(
                base_path, state_folder, str(year), f"{q_num}.json"
            )

            if not os.path.exists(file_path):
                missing_files.append(
                    [state_name, year, f"Q{q_num}"]
                )
                continue

            try:
                with open(file_path, "r", encoding="utf-8") as file:
                    json_data = json.load(file)

                registered = (
                    json_data.get("data", {})
                    .get("aggregated", {})
                    .get("registeredCount")
                )

                if registered is None:
                    read_errors.append(
                        [state_name, year, f"Q{q_num}",
                         "registeredCount missing"]
                    )
                    continue

                rows.append({
                    "State": state_name,
                    "StateFolder": state_folder,
                    "Year": year,
                    "Quarter": f"Q{q_num}",
                    "QuarterNumber": q_num,
                    "RegisteredMerchants": registered
                })

            except Exception as error:
                read_errors.append(
                    [state_name, year, f"Q{q_num}", str(error)]
                )

if not rows:
    raise ValueError("JSON files-la valid merchant data kidaikkala.")

master_df = pd.DataFrame(rows)

master_df["Year"] = pd.to_numeric(master_df["Year"])
master_df["QuarterNumber"] = pd.to_numeric(master_df["QuarterNumber"])
master_df["RegisteredMerchants"] = pd.to_numeric(
    master_df["RegisteredMerchants"], errors="coerce"
)

master_df = master_df.sort_values(
    ["State", "Year", "QuarterNumber"]
).reset_index(drop=True)

print("\n========== MERCHANT STATE MASTER ==========")
print("Rows:", len(master_df))
print("States/UTs with data:", master_df["State"].nunique())
print("Years:", sorted(master_df["Year"].unique()))
display(master_df.head(10))


# =====================================================
# STEP 3: SAVE COMBINED + INDIVIDUAL STATE CSV FILES
# =====================================================

master_file = os.path.join(
    output_path, "merchant_state_master_2018_2026.csv"
)
master_df.to_csv(master_file, index=False)

for state_folder in state_folders:
    individual_df = master_df[
        master_df["StateFolder"] == state_folder
    ][[
        "State", "Year", "Quarter",
        "QuarterNumber", "RegisteredMerchants"
    ]]

    safe_name = re.sub(r"[^A-Za-z0-9_-]+", "_", state_folder)

    individual_file = os.path.join(
        state_csv_path,
        f"{safe_name}_merchant_master_2018_2026.csv"
    )
    individual_df.to_csv(individual_file, index=False)

print("Combined master saved:", os.path.exists(master_file))
print("Individual state CSVs:", len(os.listdir(state_csv_path)))


# =====================================================
# STEP 4: QUARTER-TO-QUARTER COMPARISON
# =====================================================

def difference(current, previous):
    if pd.notna(current) and pd.notna(previous):
        return current - previous
    return np.nan

def pct_change(current, previous):
    if pd.notna(current) and pd.notna(previous) and previous != 0:
        return (current - previous) / previous * 100
    return np.nan

comparison_rows = []

for state_folder in state_folders:
    state_name = state_display_name(state_folder)

    state_df = master_df[
        master_df["StateFolder"] == state_folder
    ]

    for year in years:
        year_df = state_df[state_df["Year"] == year]

        if year_df.empty:
            continue

        values = {}
        for quarter in quarters:
            selected = year_df.loc[
                year_df["Quarter"] == quarter,
                "RegisteredMerchants"
            ]
            values[quarter] = (
                selected.iloc[0] if not selected.empty else np.nan
            )

        q1, q2, q3, q4 = [values[q] for q in quarters]

        comparison_rows.append({
            "State": state_name,
            "Year": year,
            "Q1": q1, "Q2": q2, "Q3": q3, "Q4": q4,
            "Q2_minus_Q1": difference(q2, q1),
            "Q2_change_percent": pct_change(q2, q1),
            "Q3_minus_Q2": difference(q3, q2),
            "Q3_change_percent": pct_change(q3, q2),
            "Q4_minus_Q3": difference(q4, q3),
            "Q4_change_percent": pct_change(q4, q3)
        })

comparison_df = pd.DataFrame(comparison_rows)

comparison_file = os.path.join(
    output_path, "merchant_state_quarter_comparison_2018_2026.csv"
)
comparison_df.to_csv(comparison_file, index=False)

for year in years:
    year_comparison = comparison_df[
        comparison_df["Year"] == year
    ]
    year_file = os.path.join(
        compare_path, f"merchant_state_comparison_{year}.csv"
    )
    year_comparison.to_csv(year_file, index=False)

print("\n========== QUARTER COMPARISON ==========")
print("Comparison rows:", len(comparison_df))
display(comparison_df.head(10))


# =====================================================
# STEP 5: PROFESSIONAL DARK THEME
# =====================================================

BG = "#0B1020"
PANEL = "#111A2E"
GRID = "#26344D"
TEXT = "#E6EDF7"
MUTED = "#9FB0C8"
CYAN = "#38BDF8"

year_colors = [
    "#38BDF8", "#A78BFA", "#34D399",
    "#FBBF24", "#FB7185", "#22D3EE",
    "#C084FC", "#4ADE80", "#F97316"
]

def style_chart(fig, ax, title, subtitle, ylabel):
    fig.patch.set_facecolor(BG)
    ax.set_facecolor(PANEL)

    ax.set_title(
        title, fontsize=15, fontweight="bold",
        color=TEXT, loc="left", pad=25
    )
    ax.text(
        0, 1.02, subtitle, transform=ax.transAxes,
        color=MUTED, fontsize=9, va="bottom"
    )

    ax.set_xlabel("Quarter", color=TEXT, labelpad=10)
    ax.set_ylabel(ylabel, color=TEXT, labelpad=10)
    ax.tick_params(colors=MUTED)

    ax.yaxis.set_major_formatter(
        FuncFormatter(
            lambda x, pos:
            f"{x/1_000_000:.1f}M" if abs(x) >= 1_000_000
            else f"{x/1_000:.0f}K" if abs(x) >= 1_000
            else f"{x:.0f}"
        )
    )

    ax.grid(
        True, color=GRID, linestyle="--",
        linewidth=0.7, alpha=0.8
    )

    for spine in ax.spines.values():
        spine.set_color(GRID)

    ax.set_axisbelow(True)


# =====================================================
# STEP 6: STATE-WISE CHARTS — SAVE AND DISPLAY
# =====================================================

charts_created = 0
available_years = sorted(master_df["Year"].unique())

for state_folder in state_folders:
    state_name = state_display_name(state_folder)

    state_df = master_df[
        master_df["StateFolder"] == state_folder
    ]

    fig, ax = plt.subplots(figsize=(11, 6))

    lines_created = 0

    for index, year in enumerate(available_years):
        year_df = state_df[state_df["Year"] == year]

        values = []
        for quarter in quarters:
            selected = year_df.loc[
                year_df["Quarter"] == quarter,
                "RegisteredMerchants"
            ]
            values.append(
                selected.iloc[0] if not selected.empty else np.nan
            )

        values = np.array(values, dtype=float)

        if np.isnan(values).all():
            continue

        ax.plot(
            range(4), values,
            marker="o", linewidth=2, markersize=5,
            color=year_colors[index % len(year_colors)],
            label=str(year)
        )
        lines_created += 1

    if lines_created == 0:
        plt.close(fig)
        continue

    ax.set_xticks(range(4))
    ax.set_xticklabels(quarters)

    style_chart(
        fig, ax,
        f"MERCHANT PERFORMANCE | {state_name.upper()}",
        "AGGREGATED MERCHANT DATA • YEAR-WISE QUARTER TREND",
        "Registered Merchants"
    )

    ax.legend(
        title="Year", ncol=3,
        facecolor=PANEL, edgecolor=GRID,
        labelcolor=TEXT, fontsize=8
    )

    ax.margins(y=0.15)
    fig.tight_layout(pad=2)

    safe_name = re.sub(r"[^A-Za-z0-9_-]+", "_", state_folder)
    chart_file = os.path.join(
        chart_path, f"{safe_name}_merchant_quarterly_dark.png"
    )

    fig.savefig(
        chart_file, dpi=200,
        facecolor=fig.get_facecolor(),
        bbox_inches="tight"
    )

    # IMPORTANT: show chart inside Jupyter Notebook
    plt.show()
    plt.close(fig)

    charts_created += 1
    print(f"Chart displayed and saved: {state_name}")


# =====================================================
# STEP 7: TOP 10 STATES — LATEST COMMON QUARTER
# =====================================================

coverage = (
    master_df.groupby(["Year", "Quarter", "QuarterNumber"])["State"]
    .nunique()
    .reset_index(name="StateCount")
)

total_states = master_df["State"].nunique()

common_periods = coverage[
    coverage["StateCount"] == total_states
]

if not common_periods.empty:
    chosen = common_periods.sort_values(
        ["Year", "QuarterNumber"]
    ).iloc[-1]
else:
    chosen = coverage.sort_values(
        ["StateCount", "Year", "QuarterNumber"]
    ).iloc[-1]

ranking_year = int(chosen["Year"])
ranking_quarter = chosen["Quarter"]

ranking_df = master_df[
    (master_df["Year"] == ranking_year)
    & (master_df["Quarter"] == ranking_quarter)
].copy()

ranking_df = ranking_df.sort_values(
    "RegisteredMerchants", ascending=False
).reset_index(drop=True)

ranking_df.insert(0, "Rank", range(1, len(ranking_df) + 1))

ranking_file = os.path.join(
    output_path, "merchant_state_ranking.csv"
)
ranking_df.to_csv(ranking_file, index=False)

print("\n========== STATE RANKING ==========")
print("Ranking period:", ranking_year, ranking_quarter)
display(ranking_df.head(10))


# =====================================================
# STEP 8: SHOW TOP 10 RANKING CHART
# =====================================================

top10 = ranking_df.head(10).sort_values(
    "RegisteredMerchants", ascending=True
)

fig, ax = plt.subplots(figsize=(12, 7))

ax.barh(
    top10["State"], top10["RegisteredMerchants"],
    color=CYAN, height=0.65
)

style_chart(
    fig, ax,
    "TOP 10 STATES | REGISTERED MERCHANTS",
    f"INDIA • {ranking_year} {ranking_quarter}",
    "Registered Merchants"
)

fig.tight_layout(pad=2)

ranking_chart = os.path.join(
    chart_path, "merchant_state_top10_dark.png"
)
fig.savefig(
    ranking_chart, dpi=220,
    facecolor=fig.get_facecolor(),
    bbox_inches="tight"
)

plt.show()
plt.close(fig)


# =====================================================
# STEP 9: MISSING FILES / ERRORS AND FINAL VERIFICATION
# =====================================================

pd.DataFrame(
    missing_files, columns=["State", "Year", "Quarter"]
).to_csv(
    os.path.join(output_path, "missing_quarters.csv"),
    index=False
)

pd.DataFrame(
    read_errors, columns=["State", "Year", "Quarter", "Error"]
).to_csv(
    os.path.join(output_path, "read_errors.csv"),
    index=False
)

verified_master = pd.read_csv(master_file)
verified_comparison = pd.read_csv(comparison_file)
verified_ranking = pd.read_csv(ranking_file)

chart_files = [
    file for file in os.listdir(chart_path)
    if file.endswith(".png")
]

print("\n========== FINAL VERIFICATION ==========")
print("States found:", len(state_folders))
print("Master rows:", len(verified_master))
print("States in master:", verified_master["State"].nunique())
print("Comparison rows:", len(verified_comparison))
print("Ranked states:", len(verified_ranking))
print("Charts saved:", len(chart_files))
print("State charts displayed:", charts_created)
print("Missing JSON files:", len(missing_files))
print("Read/schema errors:", len(read_errors))
print("Output folder:", output_path)
print("Chart folder:", chart_path)

display(verified_master.head())
display(verified_ranking.head(10))
```

```python
import os
import json
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
from IPython.display import display

# ==========================================================
# STEP 1: FOLDER PATHS
# ==========================================================

india_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\merchant\country\india"

state_path = os.path.join(india_path, "state")

# Output folder state folder-kku veliya irukkum
output_path = os.path.join(
    india_path, "merchant_state_analysis_2018_2026"
)

state_csv_path = os.path.join(output_path, "state_masters")
compare_path = os.path.join(output_path, "year_wise_comparisons")
chart_path = os.path.join(output_path, "charts")

for folder in [output_path, state_csv_path, compare_path, chart_path]:
    os.makedirs(folder, exist_ok=True)

years = list(range(2018, 2027))
quarters = ["Q1", "Q2", "Q3", "Q4"]

print("MERCHANT STATE ANALYSIS STARTED", flush=True)
print("State path exists:", os.path.exists(state_path), flush=True)

if not os.path.exists(state_path):
    raise FileNotFoundError("Merchant state folder path incorrect.")

state_folders = sorted([
    folder for folder in os.listdir(state_path)
    if os.path.isdir(os.path.join(state_path, folder))
])

def state_name(folder):
    name = folder.replace("-", " ").title()

    if name == "Dadra & Nagar Haveli & Daman & Diu":
        name = "Dadra and Nagar Haveli and Daman and Diu"

    return name

print("Total state/UT folders:", len(state_folders), flush=True)


# ==========================================================
# STEP 2: READ ALL STATE JSON FILES
# ==========================================================

rows = []
missing_files = []
read_errors = []

for state_folder in state_folders:

    current_state = state_name(state_folder)

    for year in years:
        for q_num in range(1, 5):

            file_path = os.path.join(
                state_path,
                state_folder,
                str(year),
                f"{q_num}.json"
            )

            if not os.path.exists(file_path):
                missing_files.append({
                    "State": current_state,
                    "Year": year,
                    "Quarter": f"Q{q_num}"
                })
                continue

            try:
                with open(file_path, "r", encoding="utf-8") as file:
                    json_data = json.load(file)

                registered_count = (
                    json_data.get("data", {})
                    .get("aggregated", {})
                    .get("registeredCount")
                )

                if registered_count is None:
                    read_errors.append({
                        "State": current_state,
                        "Year": year,
                        "Quarter": f"Q{q_num}",
                        "Error": "registeredCount field missing"
                    })
                    continue

                rows.append({
                    "State": current_state,
                    "StateFolder": state_folder,
                    "Year": year,
                    "Quarter": f"Q{q_num}",
                    "QuarterNumber": q_num,
                    "RegisteredMerchants": registered_count
                })

            except Exception as error:
                read_errors.append({
                    "State": current_state,
                    "Year": year,
                    "Quarter": f"Q{q_num}",
                    "Error": str(error)
                })

    print("Processed state:", current_state, flush=True)

if not rows:
    raise ValueError("Valid Merchant JSON data kidaikkala.")

master_df = pd.DataFrame(rows)

master_df["Year"] = pd.to_numeric(master_df["Year"])
master_df["QuarterNumber"] = pd.to_numeric(master_df["QuarterNumber"])
master_df["RegisteredMerchants"] = pd.to_numeric(
    master_df["RegisteredMerchants"], errors="coerce"
)

master_df = master_df.sort_values(
    ["State", "Year", "QuarterNumber"]
).reset_index(drop=True)

print("\n========== MASTER DATA ==========")
print("Total rows:", len(master_df))
print("States with data:", master_df["State"].nunique())
print("Years:", sorted(master_df["Year"].unique()))
display(master_df.head(10))


# ==========================================================
# STEP 3: SAVE COMBINED MASTER CSV
# ==========================================================

master_file = os.path.join(
    output_path, "merchant_state_master_2018_2026.csv"
)

master_df.to_csv(master_file, index=False)

# Each state-ku separate CSV
for state_folder in state_folders:

    state_df = master_df[
        master_df["StateFolder"] == state_folder
    ][[
        "State", "Year", "Quarter",
        "QuarterNumber", "RegisteredMerchants"
    ]]

    safe_name = re.sub(
        r"[^A-Za-z0-9_-]+", "_", state_folder
    )

    state_df.to_csv(
        os.path.join(
            state_csv_path,
            f"{safe_name}_merchant_master.csv"
        ),
        index=False
    )

print("\nCombined master saved:", os.path.exists(master_file))
print("Separate state CSVs saved:", len(state_folders))


# ==========================================================
# STEP 4: QUARTER-TO-QUARTER COMPARISON
# ==========================================================

def difference(current, previous):
    if pd.notna(current) and pd.notna(previous):
        return current - previous
    return np.nan

def percentage_change(current, previous):
    if pd.notna(current) and pd.notna(previous) and previous != 0:
        return (current - previous) / previous * 100
    return np.nan

comparison_rows = []

for state_folder in state_folders:

    current_state = state_name(state_folder)

    state_df = master_df[
        master_df["StateFolder"] == state_folder
    ]

    for year in years:

        year_df = state_df[state_df["Year"] == year]

        if year_df.empty:
            continue

        values = {}

        for quarter in quarters:
            selected = year_df.loc[
                year_df["Quarter"] == quarter,
                "RegisteredMerchants"
            ]

            values[quarter] = (
                selected.iloc[0] if not selected.empty else np.nan
            )

        q1, q2, q3, q4 = [values[q] for q in quarters]

        comparison_rows.append({
            "State": current_state,
            "Year": year,
            "Q1": q1,
            "Q2": q2,
            "Q3": q3,
            "Q4": q4,

            "Q2_minus_Q1": difference(q2, q1),
            "Q2_change_percent": percentage_change(q2, q1),

            "Q3_minus_Q2": difference(q3, q2),
            "Q3_change_percent": percentage_change(q3, q2),

            "Q4_minus_Q3": difference(q4, q3),
            "Q4_change_percent": percentage_change(q4, q3)
        })

comparison_df = pd.DataFrame(comparison_rows)

comparison_file = os.path.join(
    output_path,
    "merchant_state_quarter_comparison_2018_2026.csv"
)

comparison_df.to_csv(comparison_file, index=False)

# Ovvoru year-kum separate comparison CSV
for year in years:

    year_df = comparison_df[
        comparison_df["Year"] == year
    ]

    year_df.to_csv(
        os.path.join(
            compare_path,
            f"merchant_state_comparison_{year}.csv"
        ),
        index=False
    )

print("\n========== QUARTER COMPARISON ==========")
print("Comparison records:", len(comparison_df))
display(comparison_df.head(10))


# ==========================================================
# STEP 5: FIND LATEST QUARTER WITH ALL STATES
# ==========================================================

state_count = master_df["State"].nunique()

coverage = (
    master_df.groupby(["Year", "Quarter", "QuarterNumber"])["State"]
    .nunique()
    .reset_index(name="StatesAvailable")
)

complete_periods = coverage[
    coverage["StatesAvailable"] == state_count
]

if not complete_periods.empty:
    chosen = complete_periods.sort_values(
        ["Year", "QuarterNumber"]
    ).iloc[-1]
else:
    chosen = coverage.sort_values(
        ["StatesAvailable", "Year", "QuarterNumber"]
    ).iloc[-1]

ranking_year = int(chosen["Year"])
ranking_quarter = chosen["Quarter"]

ranking_df = master_df[
    (master_df["Year"] == ranking_year) &
    (master_df["Quarter"] == ranking_quarter)
].copy()

ranking_df = ranking_df.dropna(
    subset=["RegisteredMerchants"]
).sort_values(
    "RegisteredMerchants", ascending=False
).reset_index(drop=True)

ranking_df.insert(0, "Rank", range(1, len(ranking_df) + 1))

ranking_file = os.path.join(
    output_path, "merchant_state_ranking.csv"
)

ranking_df.to_csv(ranking_file, index=False)

print("\n========== STATE RANKING ==========")
print("Ranking period:", ranking_year, ranking_quarter)
print("States in ranking:", len(ranking_df))
display(ranking_df.head(10))


# ==========================================================
# STEP 6: PROFESSIONAL DARK THEME
# ==========================================================

BG = "#0B1020"
PANEL = "#111A2E"
GRID = "#26344D"
TEXT = "#E6EDF7"
MUTED = "#9FB0C8"
CYAN = "#38BDF8"

year_colors = [
    "#38BDF8", "#A78BFA", "#34D399",
    "#FBBF24", "#FB7185", "#22D3EE",
    "#C084FC", "#4ADE80", "#F97316"
]

def style_chart(fig, ax, title, subtitle, ylabel):

    fig.patch.set_facecolor(BG)
    ax.set_facecolor(PANEL)

    ax.set_title(
        title,
        fontsize=15,
        fontweight="bold",
        color=TEXT,
        loc="left",
        pad=25
    )

    ax.text(
        0, 1.02, subtitle,
        transform=ax.transAxes,
        color=MUTED,
        fontsize=9,
        va="bottom"
    )

    ax.set_xlabel("Quarter", color=TEXT, labelpad=10)
    ax.set_ylabel(ylabel, color=TEXT, labelpad=10)
    ax.tick_params(colors=MUTED)

    ax.yaxis.set_major_formatter(
        FuncFormatter(
            lambda value, position:
            f"{value / 1_000_000:.1f}M"
            if abs(value) >= 1_000_000
            else f"{value / 1_000:.0f}K"
            if abs(value) >= 1_000
            else f"{value:.0f}"
        )
    )

    ax.grid(
        True,
        color=GRID,
        linestyle="--",
        linewidth=0.7,
        alpha=0.8
    )

    for spine in ax.spines.values():
        spine.set_color(GRID)

    ax.set_axisbelow(True)


# ==========================================================
# STEP 7: CREATE & DISPLAY CHARTS FOR ALL STATES
# ==========================================================

charts_created = 0
available_years = sorted(master_df["Year"].unique())

for state_folder in state_folders:

    current_state = state_name(state_folder)

    state_df = master_df[
        master_df["StateFolder"] == state_folder
    ]

    fig, ax = plt.subplots(figsize=(10, 5.5))
    lines_created = 0

    for index, year in enumerate(available_years):

        year_df = state_df[state_df["Year"] == year]
        values = []

        for quarter in quarters:
            selected = year_df.loc[
                year_df["Quarter"] == quarter,
                "RegisteredMerchants"
            ]

            values.append(
                selected.iloc[0] if not selected.empty else np.nan
            )

        values = np.array(values, dtype=float)

        if np.isnan(values).all():
            continue

        ax.plot(
            range(4),
            values,
            marker="o",
            linewidth=2,
            markersize=4,
            color=year_colors[index % len(year_colors)],
            label=str(year)
        )

        lines_created += 1

    if lines_created == 0:
        plt.close(fig)
        continue

    ax.set_xticks(range(4))
    ax.set_xticklabels(quarters)

    style_chart(
        fig,
        ax,
        f"MERCHANT PERFORMANCE | {current_state.upper()}",
        "INDIA  •  QUARTERLY REGISTERED MERCHANT TREND",
        "Registered Merchants"
    )

    ax.legend(
        title="Year",
        ncol=3,
        facecolor=PANEL,
        edgecolor=GRID,
        labelcolor=TEXT,
        fontsize=8
    )

    ax.margins(y=0.15)
    fig.tight_layout(pad=2)

    safe_name = re.sub(
        r"[^A-Za-z0-9_-]+", "_", state_folder
    )

    chart_file = os.path.join(
        chart_path,
        f"{safe_name}_merchant_dark.png"
    )

    fig.savefig(
        chart_file,
        dpi=150,
        facecolor=fig.get_facecolor(),
        bbox_inches="tight"
    )

    # Notebook-la chart display aagum
    plt.show()
    plt.close(fig)

    charts_created += 1
    print(
        f"Chart displayed and saved: {current_state}",
        flush=True
    )


# ==========================================================
# STEP 8: TOP 10 STATES BAR CHART
# ==========================================================

top10 = ranking_df.head(10).sort_values(
    "RegisteredMerchants", ascending=True
)

fig, ax = plt.subplots(figsize=(11, 6.5))

fig.patch.set_facecolor(BG)
ax.set_facecolor(PANEL)

colors = [
    "#34D399" if state == ranking_df.iloc[0]["State"]
    else CYAN
    for state in top10["State"]
]

bars = ax.barh(
    top10["State"],
    top10["RegisteredMerchants"],
    color=colors,
    height=0.65
)

for bar, value in zip(bars, top10["RegisteredMerchants"]):
    ax.annotate(
        f"{value:,.0f}",
        (bar.get_width(), bar.get_y() + bar.get_height() / 2),
        xytext=(6, 0),
        textcoords="offset points",
        va="center",
        color=TEXT,
        fontsize=9
    )

ax.set_title(
    "TOP 10 STATES | REGISTERED MERCHANTS",
    fontsize=16,
    fontweight="bold",
    color=TEXT,
    loc="left",
    pad=25
)

ax.text(
    0, 1.02,
    f"INDIA  •  {ranking_year} {ranking_quarter}",
    transform=ax.transAxes,
    color=MUTED,
    fontsize=9
)

ax.set_xlabel("Registered Merchants", color=TEXT)
ax.tick_params(colors=MUTED)

ax.xaxis.set_major_formatter(
    FuncFormatter(
        lambda value, position:
        f"{value / 1_000_000:.1f}M"
        if value >= 1_000_000
        else f"{value / 1_000:.0f}K"
    )
)

ax.grid(axis="x", color=GRID, linestyle="--", linewidth=0.7)

for spine in ax.spines.values():
    spine.set_color(GRID)

ax.set_axisbelow(True)
ax.margins(x=0.20)
fig.tight_layout(pad=2)

ranking_chart = os.path.join(
    chart_path, "merchant_state_top10_dark.png"
)

fig.savefig(
    ranking_chart,
    dpi=200,
    facecolor=fig.get_facecolor(),
    bbox_inches="tight"
)

plt.show()
plt.close(fig)


# ==========================================================
# STEP 9: SAVE MISSING FILE & ERROR REPORTS
# ==========================================================

pd.DataFrame(
    missing_files,
    columns=["State", "Year", "Quarter"]
).to_csv(
    os.path.join(output_path, "missing_quarters.csv"),
    index=False
)

pd.DataFrame(read_errors).to_csv(
    os.path.join(output_path, "read_errors.csv"),
    index=False
)


# ==========================================================
# STEP 10: FINAL VERIFICATION
# ==========================================================

verified_master = pd.read_csv(master_file)
verified_comparison = pd.read_csv(comparison_file)
verified_ranking = pd.read_csv(ranking_file)

chart_files = [
    file for file in os.listdir(chart_path)
    if file.endswith(".png")
]

print("\n========== FINAL VERIFICATION ==========")
print("State folders found:", len(state_folders))
print("State CSV files:", len(os.listdir(state_csv_path)))
print("Master rows:", len(verified_master))
print("States in master:", verified_master["State"].nunique())
print("Comparison records:", len(verified_comparison))
print("Ranking rows:", len(verified_ranking))
print("State charts displayed:", charts_created)
print("Total chart PNG files:", len(chart_files))
print("Missing JSON files:", len(missing_files))
print("Read/schema errors:", len(read_errors))

print("\nOutput folder:")
print(output_path)

print("\nCharts folder:")
print(chart_path)

print("\nVerified state ranking:")
display(verified_ranking.head(10))

print("\nMerchant State Analysis Completed!")
```

```python
import os
import json
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
from IPython.display import display

# ==========================================================
# STEP 1: PATHS
# ==========================================================

base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\user\country\india"

output_path = os.path.join(
    base_path, "user_india_analysis_2018_2026"
)

chart_path = os.path.join(output_path, "charts")
comparison_path = os.path.join(output_path, "year_wise_comparisons")

os.makedirs(output_path, exist_ok=True)
os.makedirs(chart_path, exist_ok=True)
os.makedirs(comparison_path, exist_ok=True)

years = list(range(2018, 2027))
quarters = ["Q1", "Q2", "Q3", "Q4"]

print("USER INDIA ANALYSIS STARTED")
print("Folder exists:", os.path.exists(base_path))


# ==========================================================
# STEP 2: READ JSON FILES FOR ALL YEARS AND QUARTERS
# ==========================================================

rows = []
missing_files = []
read_errors = []

for year in years:
    for q_num in range(1, 5):

        file_path = os.path.join(
            base_path, str(year), f"{q_num}.json"
        )

        if not os.path.exists(file_path):
            missing_files.append({
                "Year": year,
                "Quarter": f"Q{q_num}",
                "File": file_path
            })
            continue

        try:
            with open(file_path, "r", encoding="utf-8") as file:
                json_data = json.load(file)

            registered_count = (
                json_data.get("data", {})
                .get("aggregated", {})
                .get("registeredCount")
            )

            if registered_count is None:
                read_errors.append({
                    "Year": year,
                    "Quarter": f"Q{q_num}",
                    "Error": "registeredCount field missing"
                })
                continue

            rows.append({
                "Year": year,
                "Quarter": f"Q{q_num}",
                "QuarterNumber": q_num,
                "RegisteredUsers": registered_count
            })

            print(f"Read successfully: {year} Q{q_num}")

        except Exception as error:
            read_errors.append({
                "Year": year,
                "Quarter": f"Q{q_num}",
                "Error": str(error)
            })

if not rows:
    raise ValueError(
        "User JSON data read aagala. Folder path-ai check pannu."
    )


# ==========================================================
# STEP 3: CREATE MASTER DATAFRAME
# ==========================================================

master_df = pd.DataFrame(rows)

master_df["Year"] = pd.to_numeric(
    master_df["Year"], errors="coerce"
)

master_df["QuarterNumber"] = pd.to_numeric(
    master_df["QuarterNumber"], errors="coerce"
)

master_df["RegisteredUsers"] = pd.to_numeric(
    master_df["RegisteredUsers"], errors="coerce"
)

master_df = master_df.sort_values(
    ["Year", "QuarterNumber"]
).reset_index(drop=True)

print("\n========== USER INDIA MASTER DATA ==========")
print("Rows:", len(master_df))
print("Columns:", master_df.columns.tolist())
print("Years:", sorted(master_df["Year"].unique().tolist()))
print("Quarters:", sorted(master_df["Quarter"].unique().tolist()))

display(master_df)


# ==========================================================
# STEP 4: QUARTER-TO-QUARTER COMPARISON
# Each year separately: Q1->Q2, Q2->Q3, Q3->Q4
# ==========================================================

def calculate_difference(current, previous):
    if pd.notna(current) and pd.notna(previous):
        return current - previous
    return np.nan


def calculate_percentage(current, previous):
    if (
        pd.notna(current)
        and pd.notna(previous)
        and previous != 0
    ):
        return (current - previous) / previous * 100
    return np.nan


comparison_rows = []

for year in years:

    year_df = master_df[master_df["Year"] == year]

    if year_df.empty:
        continue

    values = {}

    for quarter in quarters:
        selected = year_df.loc[
            year_df["Quarter"] == quarter,
            "RegisteredUsers"
        ]

        values[quarter] = (
            selected.iloc[0] if not selected.empty else np.nan
        )

    q1, q2, q3, q4 = [values[q] for q in quarters]

    comparison_rows.append({
        "Year": year,
        "Q1": q1,
        "Q2": q2,
        "Q3": q3,
        "Q4": q4,

        "Q2_minus_Q1": calculate_difference(q2, q1),
        "Q2_change_percent": calculate_percentage(q2, q1),

        "Q3_minus_Q2": calculate_difference(q3, q2),
        "Q3_change_percent": calculate_percentage(q3, q2),

        "Q4_minus_Q3": calculate_difference(q4, q3),
        "Q4_change_percent": calculate_percentage(q4, q3),

        "Q4_minus_Q1": calculate_difference(q4, q1)
    })

comparison_df = pd.DataFrame(comparison_rows)

print("\n========== YEAR-WISE QUARTER COMPARISON ==========")
display(comparison_df)


# ==========================================================
# STEP 5: OVERALL YEAR-ON-YEAR COMPARISON
# Compare same quarter across years
# ==========================================================

overall_rows = []

available_years = sorted(
    master_df["Year"].dropna().astype(int).unique()
)

for quarter in quarters:

    previous_year = np.nan
    previous_value = np.nan

    for year in available_years:

        selected = master_df.loc[
            (master_df["Year"] == year)
            & (master_df["Quarter"] == quarter),
            "RegisteredUsers"
        ]

        current_value = (
            selected.iloc[0] if not selected.empty else np.nan
        )

        overall_rows.append({
            "Quarter": quarter,
            "Year": year,
            "RegisteredUsers": current_value,
            "PreviousYear": previous_year,
            "PreviousYearValue": previous_value,
            "Difference": calculate_difference(
                current_value, previous_value
            ),
            "ChangePercent": calculate_percentage(
                current_value, previous_value
            )
        })

        previous_year = year
        previous_value = current_value

overall_df = pd.DataFrame(overall_rows)

print("\n========== OVERALL YEAR COMPARISON ==========")
display(overall_df.head(20))


# ==========================================================
# STEP 6: HIGHEST OBSERVED REGISTERED USER COUNT
# ==========================================================

valid_data = master_df.dropna(
    subset=["RegisteredUsers"]
)

best_row = valid_data.loc[
    valid_data["RegisteredUsers"].idxmax()
]

best_summary = pd.DataFrame([{
    "Metric": "RegisteredUsers",
    "HighestObservedCount": int(best_row["RegisteredUsers"]),
    "Year": int(best_row["Year"]),
    "Quarter": best_row["Quarter"]
}])

print("\n========== HIGHEST OBSERVED USER COUNT ==========")
display(best_summary)


# ==========================================================
# STEP 7: SAVE ALL CSV FILES
# ==========================================================

master_file = os.path.join(
    output_path, "user_india_master_2018_2026.csv"
)

comparison_file = os.path.join(
    output_path, "user_india_quarter_comparison_2018_2026.csv"
)

overall_file = os.path.join(
    output_path, "user_india_overall_year_comparison.csv"
)

best_file = os.path.join(
    output_path, "user_india_highest_observed_value.csv"
)

master_df.to_csv(master_file, index=False)
comparison_df.to_csv(comparison_file, index=False)
overall_df.to_csv(overall_file, index=False)
best_summary.to_csv(best_file, index=False)

# Separate comparison file for each year
for year in years:
    one_year = comparison_df[
        comparison_df["Year"] == year
    ]

    one_year_file = os.path.join(
        comparison_path,
        f"user_india_{year}_quarter_comparison.csv"
    )

    one_year.to_csv(one_year_file, index=False)

print("\n========== CSV SAVE STATUS ==========")

for file_path in [
    master_file, comparison_file,
    overall_file, best_file
]:
    print(
        os.path.basename(file_path),
        "Saved:", os.path.exists(file_path)
    )


# ==========================================================
# STEP 8: PROFESSIONAL DARK THEME SETTINGS
# ==========================================================

BG = "#0B1020"
PANEL = "#111A2E"
GRID = "#26344D"
TEXT = "#E6EDF7"
MUTED = "#9FB0C8"
CYAN = "#38BDF8"

YEAR_COLORS = [
    "#38BDF8", "#A78BFA", "#34D399",
    "#FBBF24", "#FB7185", "#22D3EE",
    "#C084FC", "#4ADE80", "#F97316"
]


def style_chart(fig, ax, title, subtitle, ylabel):

    fig.patch.set_facecolor(BG)
    ax.set_facecolor(PANEL)

    ax.set_title(
        title,
        fontsize=16,
        fontweight="bold",
        color=TEXT,
        loc="left",
        pad=25
    )

    ax.text(
        0, 1.02, subtitle,
        transform=ax.transAxes,
        color=MUTED,
        fontsize=9,
        va="bottom"
    )

    ax.set_xlabel("Quarter", color=TEXT, labelpad=10)
    ax.set_ylabel(ylabel, color=TEXT, labelpad=10)

    ax.tick_params(colors=MUTED)

    ax.yaxis.set_major_formatter(
        FuncFormatter(
            lambda value, position:
            f"{value / 1_000_000:.1f}M"
            if abs(value) >= 1_000_000
            else f"{value / 1_000:.0f}K"
            if abs(value) >= 1_000
            else f"{value:.0f}"
        )
    )

    ax.grid(
        True,
        color=GRID,
        linestyle="--",
        linewidth=0.7,
        alpha=0.8
    )

    for spine in ax.spines.values():
        spine.set_color(GRID)

    ax.set_axisbelow(True)


# ==========================================================
# STEP 9: YEAR-WISE QUARTER TREND CHARTS
# ==========================================================

year_charts_created = 0

for year in available_years:

    year_df = master_df[
        master_df["Year"] == year
    ]

    values = []

    for quarter in quarters:
        selected = year_df.loc[
            year_df["Quarter"] == quarter,
            "RegisteredUsers"
        ]

        values.append(
            selected.iloc[0] if not selected.empty else np.nan
        )

    values = np.array(values, dtype=float)

    if np.isnan(values).all():
        continue

    fig, ax = plt.subplots(figsize=(11, 6))

    x = np.arange(4)

    # Soft glowing line
    ax.plot(
        x, values,
        color=CYAN,
        linewidth=8,
        alpha=0.10
    )

    ax.plot(
        x, values,
        color=CYAN,
        linewidth=2.8,
        marker="o",
        markersize=8,
        markeredgewidth=2,
        markeredgecolor=CYAN,
        markerfacecolor=BG
    )

    for i, value in enumerate(values):
        if not np.isnan(value):
            ax.annotate(
                f"{value:,.0f}",
                (x[i], value),
                textcoords="offset points",
                xytext=(0, 12),
                ha="center",
                color=TEXT,
                fontsize=9,
                fontweight="bold"
            )

    ax.set_xticks(x)
    ax.set_xticklabels(quarters)

    style_chart(
        fig, ax,
        f"USER ANALYTICS | {year}",
        "INDIA  •  REGISTERED USERS  •  QUARTERLY TREND",
        "Registered Users"
    )

    ax.margins(y=0.20)
    fig.tight_layout(pad=2)

    chart_file = os.path.join(
        chart_path, f"user_india_{year}_dark.png"
    )

    fig.savefig(
        chart_file,
        dpi=200,
        facecolor=fig.get_facecolor(),
        bbox_inches="tight"
    )

    # Notebook-la chart display aagum
    plt.show()
    plt.close(fig)

    year_charts_created += 1
    print(f"Chart displayed and saved: {year}")


# ==========================================================
# STEP 10: OVERALL QUARTERLY COMPARISON CHART
# One line per year
# ==========================================================

fig, ax = plt.subplots(figsize=(13, 7))

for index, year in enumerate(available_years):

    year_df = master_df[master_df["Year"] == year]
    values = []

    for quarter in quarters:
        selected = year_df.loc[
            year_df["Quarter"] == quarter,
            "RegisteredUsers"
        ]

        values.append(
            selected.iloc[0] if not selected.empty else np.nan
        )

    values = np.array(values, dtype=float)

    if np.isnan(values).all():
        continue

    ax.plot(
        range(4),
        values,
        marker="o",
        linewidth=2,
        markersize=5,
        color=YEAR_COLORS[index % len(YEAR_COLORS)],
        label=str(year)
    )

ax.set_xticks(range(4))
ax.set_xticklabels(quarters)

style_chart(
    fig, ax,
    "INDIA REGISTERED USERS | 2018–2026",
    "OVERALL YEAR-WISE QUARTER COMPARISON",
    "Registered Users"
)

ax.legend(
    title="Year",
    ncol=3,
    facecolor=PANEL,
    edgecolor=GRID,
    labelcolor=TEXT,
    fontsize=9
)

fig.tight_layout(pad=2)

overall_chart = os.path.join(
    chart_path, "user_india_overall_2018_2026_dark.png"
)

fig.savefig(
    overall_chart,
    dpi=220,
    facecolor=fig.get_facecolor(),
    bbox_inches="tight"
)

plt.show()
plt.close(fig)

print("Overall chart displayed and saved.")


# ==========================================================
# STEP 11: SAVE MISSING / ERROR REPORTS
# ==========================================================

pd.DataFrame(
    missing_files,
    columns=["Year", "Quarter", "File"]
).to_csv(
    os.path.join(output_path, "user_india_missing_quarters.csv"),
    index=False
)

pd.DataFrame(
    read_errors
).to_csv(
    os.path.join(output_path, "user_india_read_errors.csv"),
    index=False
)


# ==========================================================
# STEP 12: RE-READ SAVED CSV FILES AND VERIFY
# ==========================================================

verified_master = pd.read_csv(master_file)
verified_comparison = pd.read_csv(comparison_file)
verified_overall = pd.read_csv(overall_file)
verified_best = pd.read_csv(best_file)

saved_charts = [
    file for file in os.listdir(chart_path)
    if file.endswith(".png")
]

print("\n========== FINAL VERIFICATION ==========")
print("Master rows:", len(verified_master))
print("Comparison rows:", len(verified_comparison))
print("Overall comparison rows:", len(verified_overall))
print("Highest-value summary rows:", len(verified_best))
print("Year-wise charts displayed:", year_charts_created)
print("Total saved charts:", len(saved_charts))
print("Missing JSON files:", len(missing_files))
print("Read errors:", len(read_errors))

print(
    "Years loaded:",
    sorted(verified_master["Year"].dropna().unique().tolist())
)

print(
    "Quarters loaded:",
    sorted(verified_master["Quarter"].unique().tolist())
)

print("\nOutput folder:")
print(output_path)

print("\nVerified master preview:")
display(verified_master.head())

print("\nVerified highest observed value:")
display(verified_best)

print("\nUSER INDIA ANALYSIS COMPLETED!")
```

```python
import os
import json
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
from IPython.display import display

# ==========================================================
# STEP 1: PATHS
# ==========================================================

india_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\user\country\india"

state_path = os.path.join(india_path, "state")

# Output folder-ai state folder-ku veliya create pannuvom
output_path = os.path.join(
    india_path, "user_state_analysis_2018_2026"
)

state_csv_path = os.path.join(output_path, "state_masters")
comparison_path = os.path.join(output_path, "year_wise_comparisons")
chart_path = os.path.join(output_path, "charts")

for folder in [
    output_path, state_csv_path, comparison_path, chart_path
]:
    os.makedirs(folder, exist_ok=True)

years = list(range(2018, 2027))
quarters = ["Q1", "Q2", "Q3", "Q4"]

print("USER STATE ANALYSIS STARTED")
print("State folder exists:", os.path.exists(state_path))

if not os.path.exists(state_path):
    raise FileNotFoundError("User State folder path correct-ah irukka nu check pannu.")

# Only actual state folders
state_folders = sorted([
    folder
    for folder in os.listdir(state_path)
    if os.path.isdir(os.path.join(state_path, folder))
])

def state_name(folder):
    name = folder.replace("-", " ").title()

    if name == "Dadra & Nagar Haveli & Daman & Diu":
        name = "Dadra and Nagar Haveli and Daman and Diu"

    return name

print("Total state/UT folders:", len(state_folders))


# ==========================================================
# STEP 2: READ ALL STATE, YEAR, QUARTER JSON FILES
# ==========================================================

rows = []
missing_files = []
read_errors = []

for state_folder in state_folders:

    current_state = state_name(state_folder)

    for year in years:
        for q_num in range(1, 5):

            file_path = os.path.join(
                state_path,
                state_folder,
                str(year),
                f"{q_num}.json"
            )

            if not os.path.exists(file_path):
                missing_files.append({
                    "State": current_state,
                    "Year": year,
                    "Quarter": f"Q{q_num}"
                })
                continue

            try:
                with open(file_path, "r", encoding="utf-8") as file:
                    json_data = json.load(file)

                registered_count = (
                    json_data.get("data", {})
                    .get("aggregated", {})
                    .get("registeredCount")
                )

                if registered_count is None:
                    read_errors.append({
                        "State": current_state,
                        "Year": year,
                        "Quarter": f"Q{q_num}",
                        "Error": "registeredCount field missing"
                    })
                    continue

                rows.append({
                    "State": current_state,
                    "StateFolder": state_folder,
                    "Year": year,
                    "Quarter": f"Q{q_num}",
                    "QuarterNumber": q_num,
                    "RegisteredUsers": registered_count
                })

            except Exception as error:
                read_errors.append({
                    "State": current_state,
                    "Year": year,
                    "Quarter": f"Q{q_num}",
                    "Error": str(error)
                })

    print("Processed state:", current_state, flush=True)

if not rows:
    raise ValueError("Valid User State JSON data kidaikkala.")

# ==========================================================
# STEP 3: CREATE MASTER DATAFRAME
# ==========================================================

master_df = pd.DataFrame(rows)

master_df["Year"] = pd.to_numeric(master_df["Year"])
master_df["QuarterNumber"] = pd.to_numeric(
    master_df["QuarterNumber"]
)
master_df["RegisteredUsers"] = pd.to_numeric(
    master_df["RegisteredUsers"], errors="coerce"
)

master_df = master_df.sort_values(
    ["State", "Year", "QuarterNumber"]
).reset_index(drop=True)

print("\n========== USER STATE MASTER ==========")
print("Rows:", len(master_df))
print("States with data:", master_df["State"].nunique())
print("Years:", sorted(master_df["Year"].unique()))
print("Quarters:", sorted(master_df["Quarter"].unique()))

display(master_df.head(10))


# ==========================================================
# STEP 4: SAVE COMBINED AND INDIVIDUAL STATE MASTER CSVs
# ==========================================================

master_file = os.path.join(
    output_path, "user_state_master_2018_2026.csv"
)

master_df.to_csv(master_file, index=False)

for state_folder in state_folders:

    state_df = master_df[
        master_df["StateFolder"] == state_folder
    ][[
        "State", "Year", "Quarter",
        "QuarterNumber", "RegisteredUsers"
    ]]

    safe_name = re.sub(
        r"[^A-Za-z0-9_-]+", "_", state_folder
    )

    state_file = os.path.join(
        state_csv_path,
        f"{safe_name}_user_master_2018_2026.csv"
    )

    state_df.to_csv(state_file, index=False)

print("\nCombined master saved:", os.path.exists(master_file))
print("Individual state CSV files:", len(state_folders))


# ==========================================================
# STEP 5: QUARTER-WISE COMPARISON FOR EVERY STATE & YEAR
# Q1 vs Q2, Q2 vs Q3, Q3 vs Q4
# ==========================================================

def calculate_difference(current, previous):
    if pd.notna(current) and pd.notna(previous):
        return current - previous
    return np.nan

def calculate_percentage(current, previous):
    if (
        pd.notna(current)
        and pd.notna(previous)
        and previous != 0
    ):
        return (current - previous) / previous * 100
    return np.nan

comparison_rows = []

for state_folder in state_folders:

    current_state = state_name(state_folder)

    state_df = master_df[
        master_df["StateFolder"] == state_folder
    ]

    for year in years:

        year_df = state_df[
            state_df["Year"] == year
        ]

        if year_df.empty:
            continue

        values = {}

        for quarter in quarters:
            selected = year_df.loc[
                year_df["Quarter"] == quarter,
                "RegisteredUsers"
            ]

            values[quarter] = (
                selected.iloc[0]
                if not selected.empty
                else np.nan
            )

        q1, q2, q3, q4 = [
            values[q] for q in quarters
        ]

        comparison_rows.append({
            "State": current_state,
            "Year": year,
            "Q1": q1,
            "Q2": q2,
            "Q3": q3,
            "Q4": q4,

            "Q2_minus_Q1": calculate_difference(q2, q1),
            "Q2_change_percent": calculate_percentage(q2, q1),

            "Q3_minus_Q2": calculate_difference(q3, q2),
            "Q3_change_percent": calculate_percentage(q3, q2),

            "Q4_minus_Q3": calculate_difference(q4, q3),
            "Q4_change_percent": calculate_percentage(q4, q3),

            "Q4_minus_Q1": calculate_difference(q4, q1)
        })

comparison_df = pd.DataFrame(comparison_rows)

comparison_file = os.path.join(
    output_path,
    "user_state_quarter_comparison_2018_2026.csv"
)

comparison_df.to_csv(comparison_file, index=False)

print("\n========== USER STATE QUARTER COMPARISON ==========")
print("Comparison records:", len(comparison_df))
display(comparison_df.head(10))


# ==========================================================
# STEP 6: SAVE SEPARATE COMPARISON FOR EVERY YEAR
# ==========================================================

for year in years:

    year_df = comparison_df[
        comparison_df["Year"] == year
    ]

    year_file = os.path.join(
        comparison_path,
        f"user_state_comparison_{year}.csv"
    )

    year_df.to_csv(year_file, index=False)

print("Year-wise comparison CSVs saved.")


# ==========================================================
# STEP 7: FIND LATEST QUARTER AVAILABLE FOR ALL STATES
# ==========================================================

state_count = master_df["State"].nunique()

coverage = (
    master_df.groupby(
        ["Year", "Quarter", "QuarterNumber"]
    )["State"]
    .nunique()
    .reset_index(name="StatesAvailable")
)

complete_periods = coverage[
    coverage["StatesAvailable"] == state_count
]

if not complete_periods.empty:

    latest = complete_periods.sort_values(
        ["Year", "QuarterNumber"]
    ).iloc[-1]

else:

    latest = coverage.sort_values(
        ["StatesAvailable", "Year", "QuarterNumber"]
    ).iloc[-1]

ranking_year = int(latest["Year"])
ranking_quarter = latest["Quarter"]

ranking_df = master_df[
    (master_df["Year"] == ranking_year)
    & (master_df["Quarter"] == ranking_quarter)
].copy()

ranking_df = ranking_df.dropna(
    subset=["RegisteredUsers"]
).sort_values(
    "RegisteredUsers", ascending=False
).reset_index(drop=True)

ranking_df.insert(
    0, "Rank", range(1, len(ranking_df) + 1)
)

ranking_file = os.path.join(
    output_path, "user_state_ranking_latest_quarter.csv"
)

ranking_df.to_csv(ranking_file, index=False)

print("\n========== USER STATE RANKING ==========")
print("Ranking period:", ranking_year, ranking_quarter)
print("States ranked:", len(ranking_df))

display(
    ranking_df[[
        "Rank", "State", "Year",
        "Quarter", "RegisteredUsers"
    ]].head(10)
)


# ==========================================================
# STEP 8: BEST STATE SUMMARY
# ==========================================================

best_row = ranking_df.iloc[0]

best_summary = pd.DataFrame([{
    "ComparisonBasis": "Registered user count",
    "Year": ranking_year,
    "Quarter": ranking_quarter,
    "BestState": best_row["State"],
    "RegisteredUsers": int(best_row["RegisteredUsers"])
}])

best_file = os.path.join(
    output_path, "user_state_best_state_summary.csv"
)

best_summary.to_csv(best_file, index=False)

print("\n========== HIGHEST REGISTERED USER STATE ==========")
print("State:", best_row["State"])
print("Registered users:", f"{int(best_row['RegisteredUsers']):,}")
print("Period:", ranking_year, ranking_quarter)


# ==========================================================
# STEP 9: PROFESSIONAL DARK THEME
# ==========================================================

BG = "#0B1020"
PANEL = "#111A2E"
GRID = "#26344D"
TEXT = "#E6EDF7"
MUTED = "#9FB0C8"
CYAN = "#38BDF8"

year_colors = [
    "#38BDF8", "#A78BFA", "#34D399",
    "#FBBF24", "#FB7185", "#22D3EE",
    "#C084FC", "#4ADE80", "#F97316"
]

def style_chart(fig, ax, title, subtitle, ylabel):

    fig.patch.set_facecolor(BG)
    ax.set_facecolor(PANEL)

    ax.set_title(
        title,
        fontsize=15,
        fontweight="bold",
        color=TEXT,
        loc="left",
        pad=25
    )

    ax.text(
        0, 1.02, subtitle,
        transform=ax.transAxes,
        color=MUTED,
        fontsize=9,
        va="bottom"
    )

    ax.set_xlabel("Quarter", color=TEXT, labelpad=10)
    ax.set_ylabel(ylabel, color=TEXT, labelpad=10)
    ax.tick_params(colors=MUTED)

    ax.yaxis.set_major_formatter(
        FuncFormatter(
            lambda value, position:
            f"{value / 1_000_000:.1f}M"
            if abs(value) >= 1_000_000
            else f"{value / 1_000:.0f}K"
            if abs(value) >= 1_000
            else f"{value:.0f}"
        )
    )

    ax.grid(
        True,
        color=GRID,
        linestyle="--",
        linewidth=0.7,
        alpha=0.8
    )

    for spine in ax.spines.values():
        spine.set_color(GRID)

    ax.set_axisbelow(True)


# ==========================================================
# STEP 10: STATE-WISE QUARTERLY DARK CHARTS
# EACH LINE REPRESENTS ONE YEAR
# ==========================================================

charts_created = 0
available_years = sorted(master_df["Year"].unique())

for state_folder in state_folders:

    current_state = state_name(state_folder)

    state_df = master_df[
        master_df["StateFolder"] == state_folder
    ]

    fig, ax = plt.subplots(figsize=(10, 5.5))
    lines_created = 0

    for index, year in enumerate(available_years):

        year_df = state_df[
            state_df["Year"] == year
        ]

        values = []

        for quarter in quarters:

            selected = year_df.loc[
                year_df["Quarter"] == quarter,
                "RegisteredUsers"
            ]

            values.append(
                selected.iloc[0]
                if not selected.empty
                else np.nan
            )

        values = np.array(values, dtype=float)

        if np.isnan(values).all():
            continue

        ax.plot(
            range(4),
            values,
            marker="o",
            linewidth=2,
            markersize=4,
            color=year_colors[index % len(year_colors)],
            label=str(year)
        )

        lines_created += 1

    if lines_created == 0:
        plt.close(fig)
        continue

    ax.set_xticks(range(4))
    ax.set_xticklabels(quarters)

    style_chart(
        fig,
        ax,
        f"USER PERFORMANCE | {current_state.upper()}",
        "AGGREGATED USER DATA • QUARTERLY TREND BY YEAR",
        "Registered Users"
    )

    ax.legend(
        title="Year",
        ncol=3,
        facecolor=PANEL,
        edgecolor=GRID,
        labelcolor=TEXT,
        fontsize=8
    )

    ax.margins(y=0.15)
    fig.tight_layout(pad=2)

    safe_name = re.sub(
        r"[^A-Za-z0-9_-]+", "_", state_folder
    )

    chart_file = os.path.join(
        chart_path, f"{safe_name}_user_quarterly_dark.png"
    )

    fig.savefig(
        chart_file,
        dpi=180,
        facecolor=fig.get_facecolor(),
        bbox_inches="tight"
    )

    # Chart notebook-la display aagum
    plt.show()
    plt.close(fig)

    charts_created += 1

    print(
        f"Chart displayed and saved: {current_state}",
        flush=True
    )


# ==========================================================
# STEP 11: TOP 10 USER STATES DARK BAR CHART
# ==========================================================

top10 = ranking_df.head(10).sort_values(
    "RegisteredUsers", ascending=True
)

fig, ax = plt.subplots(figsize=(11, 6.5))

fig.patch.set_facecolor(BG)
ax.set_facecolor(PANEL)

bar_colors = [
    "#34D399" if state == best_row["State"] else CYAN
    for state in top10["State"]
]

bars = ax.barh(
    top10["State"],
    top10["RegisteredUsers"],
    color=bar_colors,
    height=0.65
)

for bar, value in zip(bars, top10["RegisteredUsers"]):
    ax.annotate(
        f"{value:,.0f}",
        (bar.get_width(), bar.get_y() + bar.get_height() / 2),
        xytext=(6, 0),
        textcoords="offset points",
        va="center",
        color=TEXT,
        fontsize=9
    )

ax.set_title(
    "TOP 10 STATES | REGISTERED USERS",
    fontsize=16,
    fontweight="bold",
    color=TEXT,
    loc="left",
    pad=25
)

ax.text(
    0, 1.02,
    f"INDIA • {ranking_year} {ranking_quarter}",
    transform=ax.transAxes,
    color=MUTED,
    fontsize=9
)

ax.set_xlabel("Registered Users", color=TEXT)
ax.tick_params(colors=MUTED)

ax.xaxis.set_major_formatter(
    FuncFormatter(
        lambda value, position:
        f"{value / 1_000_000:.1f}M"
        if value >= 1_000_000
        else f"{value / 1_000:.0f}K"
    )
)

ax.grid(axis="x", color=GRID, linestyle="--", linewidth=0.7)

for spine in ax.spines.values():
    spine.set_color(GRID)

ax.set_axisbelow(True)
ax.margins(x=0.20)
fig.tight_layout(pad=2)

ranking_chart_file = os.path.join(
    chart_path, "user_state_top10_dark.png"
)

fig.savefig(
    ranking_chart_file,
    dpi=220,
    facecolor=fig.get_facecolor(),
    bbox_inches="tight"
)

plt.show()
plt.close(fig)


# ==========================================================
# STEP 12: SAVE MISSING FILE / ERROR REPORTS
# ==========================================================

pd.DataFrame(
    missing_files,
    columns=["State", "Year", "Quarter"]
).to_csv(
    os.path.join(output_path, "user_state_missing_quarters.csv"),
    index=False
)

pd.DataFrame(
    read_errors
).to_csv(
    os.path.join(output_path, "user_state_read_errors.csv"),
    index=False
)


# ==========================================================
# STEP 13: FINAL VERIFICATION
# ==========================================================

verified_master = pd.read_csv(master_file)
verified_comparison = pd.read_csv(comparison_file)
verified_ranking = pd.read_csv(ranking_file)
verified_best = pd.read_csv(best_file)

saved_charts = [
    file for file in os.listdir(chart_path)
    if file.endswith(".png")
]

print("\n========== FINAL VERIFICATION ==========")
print("State folders:", len(state_folders))
print("Individual state CSVs:", len(os.listdir(state_csv_path)))
print("Master rows:", len(verified_master))
print("States in master:", verified_master["State"].nunique())
print("Comparison records:", len(verified_comparison))
print("Ranking rows:", len(verified_ranking))
print("Best-state summary saved:", os.path.exists(best_file))
print("State charts displayed:", charts_created)
print("Total PNG charts saved:", len(saved_charts))
print("Missing JSON files:", len(missing_files))
print("Read/schema errors:", len(read_errors))

print("\nOutput folder:")
print(output_path)

print("\nCharts folder:")
print(chart_path)

print("\nVerified best state:")
display(verified_best)

print("\nVerified ranking:")
display(verified_ranking.head(10))

print("\nUSER STATE ANALYSIS COMPLETED!")
```

```python
import os
import json
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
from IPython.display import display

# ==========================================================
# STEP 1: PATHS AND OUTPUT FOLDER
# ==========================================================

data_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data"
map_path = os.path.join(data_path, "map")

output_path = os.path.join(map_path, "map_analysis_2018_2026")
chart_path = os.path.join(output_path, "charts")
comparison_path = os.path.join(output_path, "quarter_comparisons")
ranking_path = os.path.join(output_path, "state_rankings")

for folder in [
    output_path, chart_path, comparison_path, ranking_path
]:
    os.makedirs(folder, exist_ok=True)

years = list(range(2018, 2027))
quarters = ["Q1", "Q2", "Q3", "Q4"]
categories = ["transaction", "merchant", "user"]

print("MAP ANALYSIS STARTED")
print("Map folder exists:", os.path.exists(map_path))


# ==========================================================
# STEP 2: HELPER FUNCTIONS
# ==========================================================

def pretty_state_name(name):
    name = str(name).replace("-", " ").title()

    if name == "Dadra & Nagar Haveli & Daman & Diu":
        return "Dadra and Nagar Haveli and Daman and Diu"

    return name


def normalize_name(name):
    return re.sub(r"[^a-z0-9]", "", str(name).lower())


def get_total_metric(metric_list):
    """Transaction JSON-la TOTAL metric edukkum."""
    if not isinstance(metric_list, list) or len(metric_list) == 0:
        return {}

    for metric in metric_list:
        if isinstance(metric, dict) and metric.get("type") == "TOTAL":
            return metric

    return metric_list[0] if isinstance(metric_list[0], dict) else {}


def safe_number(value):
    try:
        return float(value) if value is not None else np.nan
    except (TypeError, ValueError):
        return np.nan


# State names-to folder slugs map pannuvom
state_slugs_by_name = {}

for category in categories:
    country_dir = os.path.join(
        map_path, category, "hover", "country", "india"
    )
    state_dir = os.path.join(country_dir, "state")

    if os.path.exists(state_dir):
        for state_folder in os.listdir(state_dir):
            if os.path.isdir(os.path.join(state_dir, state_folder)):
                state_slugs_by_name[
                    normalize_name(state_folder)
                ] = state_folder


def find_state_folder(name):
    return state_slugs_by_name.get(normalize_name(name), "")


# ==========================================================
# STEP 3: READ COUNTRY-LEVEL STATE DATA
# TRANSACTION: count + amount
# MERCHANT / USER: registeredCount
# ==========================================================

state_rows = []
district_rows = []
missing_files = []
read_errors = []

for category in categories:

    print(f"\n========== MAP {category.upper()} ==========")

    country_path = os.path.join(
        map_path, category, "hover", "country", "india"
    )

    state_path = os.path.join(country_path, "state")

    print("Country path exists:", os.path.exists(country_path))
    print("State path exists:", os.path.exists(state_path))

    for year in years:
        for q_num in range(1, 5):

            file_path = os.path.join(
                country_path, str(year), f"{q_num}.json"
            )

            if not os.path.exists(file_path):
                missing_files.append({
                    "Category": category,
                    "Level": "Country-State",
                    "State": "",
                    "Year": year,
                    "Quarter": f"Q{q_num}",
                    "File": file_path
                })
                continue

            try:
                with open(file_path, "r", encoding="utf-8") as file:
                    json_data = json.load(file)

                payload = json_data.get("data", {})

                # -------------------------------
                # MAP TRANSACTION COUNTRY JSON
                # -------------------------------
                if category == "transaction":

                    items = payload.get("hoverDataList", [])

                    for item in items:
                        if not isinstance(item, dict):
                            continue

                        state_name = pretty_state_name(
                            item.get("name", "Unknown")
                        )

                        metric = get_total_metric(
                            item.get("metric", [])
                        )

                        state_rows.append({
                            "Category": "Transaction",
                            "State": state_name,
                            "StateFolder": find_state_folder(state_name),
                            "Year": year,
                            "Quarter": f"Q{q_num}",
                            "QuarterNumber": q_num,
                            "Transactions": safe_number(metric.get("count")),
                            "Amount": safe_number(metric.get("amount")),
                            "RegisteredCount": np.nan
                        })

                # -------------------------------
                # MAP MERCHANT / USER COUNTRY JSON
                # -------------------------------
                else:

                    hover_data = payload.get("hoverData", {})

                    if isinstance(hover_data, dict):
                        for state_raw, metrics in hover_data.items():

                            if not isinstance(metrics, dict):
                                continue

                            state_name = pretty_state_name(state_raw)

                            state_rows.append({
                                "Category": category.title(),
                                "State": state_name,
                                "StateFolder": find_state_folder(state_name),
                                "Year": year,
                                "Quarter": f"Q{q_num}",
                                "QuarterNumber": q_num,
                                "Transactions": np.nan,
                                "Amount": np.nan,
                                "RegisteredCount": safe_number(
                                    metrics.get("registeredCount")
                                )
                            })

                print(
                    f"{category.title()} India {year} Q{q_num}: read",
                    flush=True
                )

            except Exception as error:
                read_errors.append({
                    "Category": category,
                    "Level": "Country-State",
                    "Year": year,
                    "Quarter": f"Q{q_num}",
                    "Error": str(error)
                })


# ==========================================================
# STEP 4: READ STATE-LEVEL DISTRICT DATA
# ==========================================================

for category in categories:

    country_path = os.path.join(
        map_path, category, "hover", "country", "india"
    )

    state_path = os.path.join(country_path, "state")

    if not os.path.exists(state_path):
        continue

    state_folders = sorted([
        folder for folder in os.listdir(state_path)
        if os.path.isdir(os.path.join(state_path, folder))
    ])

    print(
        f"\nMap {category.title()} district/state-level folders:",
        len(state_folders)
    )

    for state_folder in state_folders:

        state_name = pretty_state_name(state_folder)

        for year in years:
            for q_num in range(1, 5):

                file_path = os.path.join(
                    state_path, state_folder,
                    str(year), f"{q_num}.json"
                )

                if not os.path.exists(file_path):
                    missing_files.append({
                        "Category": category,
                        "Level": "State-District",
                        "State": state_name,
                        "Year": year,
                        "Quarter": f"Q{q_num}",
                        "File": file_path
                    })
                    continue

                try:
                    with open(file_path, "r", encoding="utf-8") as file:
                        json_data = json.load(file)

                    payload = json_data.get("data", {})

                    # -------------------------------
                    # TRANSACTION DISTRICT JSON
                    # -------------------------------
                    if category == "transaction":

                        items = payload.get("hoverDataList", [])

                        for item in items:
                            if not isinstance(item, dict):
                                continue

                            metric = get_total_metric(
                                item.get("metric", [])
                            )

                            district_rows.append({
                                "Category": "Transaction",
                                "State": state_name,
                                "StateFolder": state_folder,
                                "District": item.get("name", "Unknown").title(),
                                "Year": year,
                                "Quarter": f"Q{q_num}",
                                "QuarterNumber": q_num,
                                "Transactions": safe_number(metric.get("count")),
                                "Amount": safe_number(metric.get("amount")),
                                "RegisteredCount": np.nan
                            })

                    # -------------------------------
                    # MERCHANT / USER DISTRICT JSON
                    # -------------------------------
                    else:

                        hover_data = payload.get("hoverData", {})

                        if isinstance(hover_data, dict):
                            for district_raw, metrics in hover_data.items():

                                if not isinstance(metrics, dict):
                                    continue

                                district_rows.append({
                                    "Category": category.title(),
                                    "State": state_name,
                                    "StateFolder": state_folder,
                                    "District": str(district_raw).title(),
                                    "Year": year,
                                    "Quarter": f"Q{q_num}",
                                    "QuarterNumber": q_num,
                                    "Transactions": np.nan,
                                    "Amount": np.nan,
                                    "RegisteredCount": safe_number(
                                        metrics.get("registeredCount")
                                    )
                                })

                except Exception as error:
                    read_errors.append({
                        "Category": category,
                        "Level": "State-District",
                        "State": state_name,
                        "Year": year,
                        "Quarter": f"Q{q_num}",
                        "Error": str(error)
                    })

        print(f"District data read: {category.title()} - {state_name}")


# ==========================================================
# STEP 5: CREATE MASTER DATAFRAMES
# ==========================================================

state_columns = [
    "Category", "State", "StateFolder",
    "Year", "Quarter", "QuarterNumber",
    "Transactions", "Amount", "RegisteredCount"
]

district_columns = [
    "Category", "State", "StateFolder", "District",
    "Year", "Quarter", "QuarterNumber",
    "Transactions", "Amount", "RegisteredCount"
]

state_master = pd.DataFrame(
    state_rows, columns=state_columns
)

district_master = pd.DataFrame(
    district_rows, columns=district_columns
)

if state_master.empty:
    raise ValueError("Map country-level state data kidaikkala.")

for col in ["Transactions", "Amount", "RegisteredCount"]:
    state_master[col] = pd.to_numeric(
        state_master[col], errors="coerce"
    )

    district_master[col] = pd.to_numeric(
        district_master[col], errors="coerce"
    )

state_master["Year"] = pd.to_numeric(
    state_master["Year"], errors="coerce"
)

district_master["Year"] = pd.to_numeric(
    district_master["Year"], errors="coerce"
)

print("\n========== MAP STATE MASTER ==========")
print("Rows:", len(state_master))
print("Categories:", state_master["Category"].unique().tolist())
print("States:", state_master["State"].nunique())
display(state_master.head(10))

print("\n========== MAP DISTRICT MASTER ==========")
print("Rows:", len(district_master))
print("District names:", district_master["District"].nunique())
display(district_master.head(10))


# ==========================================================
# STEP 6: SAVE COMBINED + SEPARATE CATEGORY MASTER CSVs
# ==========================================================

state_master_file = os.path.join(
    output_path, "map_state_master_all_categories_2018_2026.csv"
)

district_master_file = os.path.join(
    output_path, "map_district_master_all_categories_2018_2026.csv"
)

state_master.to_csv(state_master_file, index=False)
district_master.to_csv(district_master_file, index=False)

for category_name in ["Transaction", "Merchant", "User"]:

    cat_folder = os.path.join(output_path, category_name.lower())
    os.makedirs(cat_folder, exist_ok=True)

    state_part = state_master[
        state_master["Category"] == category_name
    ]

    district_part = district_master[
        district_master["Category"] == category_name
    ]

    state_part.to_csv(
        os.path.join(
            cat_folder,
            f"map_{category_name.lower()}_state_master_2018_2026.csv"
        ),
        index=False
    )

    district_part.to_csv(
        os.path.join(
            cat_folder,
            f"map_{category_name.lower()}_district_master_2018_2026.csv"
        ),
        index=False
    )

print("\nCombined state and district master files saved.")


# ==========================================================
# STEP 7: STATE QUARTER-TO-QUARTER COMPARISON
# ==========================================================

metric_by_category = {
    "Transaction": ["Transactions", "Amount"],
    "Merchant": ["RegisteredCount"],
    "User": ["RegisteredCount"]
}

def difference(current, previous):
    if pd.notna(current) and pd.notna(previous):
        return current - previous
    return np.nan

def percentage_change(current, previous):
    if (
        pd.notna(current)
        and pd.notna(previous)
        and previous != 0
    ):
        return (current - previous) / previous * 100
    return np.nan

comparison_rows = []

for category_name, metrics in metric_by_category.items():

    cat_df = state_master[
        state_master["Category"] == category_name
    ]

    for metric in metrics:

        valid = cat_df.dropna(subset=[metric])

        for (state_name, year), group in valid.groupby(["State", "Year"]):

            values = {}

            for quarter in quarters:
                selected = group.loc[
                    group["Quarter"] == quarter, metric
                ]

                values[quarter] = (
                    selected.iloc[0] if not selected.empty else np.nan
                )

            q1, q2, q3, q4 = [values[q] for q in quarters]

            comparison_rows.append({
                "Category": category_name,
                "Metric": metric,
                "State": state_name,
                "Year": int(year),
                "Q1": q1, "Q2": q2, "Q3": q3, "Q4": q4,
                "Q2_minus_Q1": difference(q2, q1),
                "Q2_change_percent": percentage_change(q2, q1),
                "Q3_minus_Q2": difference(q3, q2),
                "Q3_change_percent": percentage_change(q3, q2),
                "Q4_minus_Q3": difference(q4, q3),
                "Q4_change_percent": percentage_change(q4, q3)
            })

comparison_df = pd.DataFrame(comparison_rows)

comparison_file = os.path.join(
    output_path, "map_state_quarter_comparisons_2018_2026.csv"
)

comparison_df.to_csv(comparison_file, index=False)

for category_name in ["Transaction", "Merchant", "User"]:

    cat_compare = comparison_df[
        comparison_df["Category"] == category_name
    ]

    cat_compare.to_csv(
        os.path.join(
            comparison_path,
            f"map_{category_name.lower()}_state_quarter_comparison.csv"
        ),
        index=False
    )

print("\n========== STATE QUARTER COMPARISON ==========")
print("Comparison records:", len(comparison_df))
display(comparison_df.head(12))


# ==========================================================
# STEP 8: FIND BEST / TOP STATES FOR EACH CATEGORY
# USE LATEST QUARTER WITH MAXIMUM STATE COVERAGE
# ==========================================================

ranking_rows = []

for category_name, metrics in metric_by_category.items():

    cat_df = state_master[
        state_master["Category"] == category_name
    ]

    for metric in metrics:

        valid = cat_df.dropna(subset=[metric]).copy()

        if valid.empty:
            continue

        coverage = (
            valid.groupby(["Year", "Quarter", "QuarterNumber"])["State"]
            .nunique()
            .reset_index(name="StatesAvailable")
        )

        max_coverage = coverage["StatesAvailable"].max()

        best_periods = coverage[
            coverage["StatesAvailable"] == max_coverage
        ].sort_values(["Year", "QuarterNumber"])

        selected_period = best_periods.iloc[-1]

        selected_year = int(selected_period["Year"])
        selected_quarter = selected_period["Quarter"]

        period_df = valid[
            (valid["Year"] == selected_year)
            & (valid["Quarter"] == selected_quarter)
        ].sort_values(metric, ascending=False).reset_index(drop=True)

        period_df = period_df.drop_duplicates(
            subset=["State"], keep="first"
        ).reset_index(drop=True)

        period_df.insert(0, "Rank", range(1, len(period_df) + 1))

        period_df["Metric"] = metric
        period_df["RankingBasis"] = category_name
        period_df["RankingYear"] = selected_year
        period_df["RankingQuarter"] = selected_quarter

        ranking_rows.append(period_df)

        safe_metric = re.sub(
            r"[^A-Za-z0-9_-]+", "_", metric
        )

        rank_file = os.path.join(
            ranking_path,
            f"map_{category_name.lower()}_{safe_metric}_ranking.csv"
        )

        period_df.to_csv(rank_file, index=False)

        print(
            f"\n{category_name} ranking: {metric} | "
            f"{selected_year} {selected_quarter}"
        )

        display(
            period_df[["Rank", "State", metric]].head(10)
        )

ranking_df = (
    pd.concat(ranking_rows, ignore_index=True)
    if ranking_rows else pd.DataFrame()
)

ranking_master_file = os.path.join(
    output_path, "map_state_rankings_all_categories.csv"
)

ranking_df.to_csv(ranking_master_file, index=False)


# ==========================================================
# STEP 9: PROFESSIONAL DARK THEME CHARTS
# ==========================================================

BG = "#0B1020"
PANEL = "#111A2E"
GRID = "#26344D"
TEXT = "#E6EDF7"
MUTED = "#9FB0C8"
CYAN = "#38BDF8"

YEAR_COLORS = [
    "#38BDF8", "#A78BFA", "#34D399",
    "#FBBF24", "#FB7185", "#22D3EE",
    "#C084FC", "#4ADE80", "#F97316"
]

def style_axis(fig, ax, title, subtitle, ylabel):

    fig.patch.set_facecolor(BG)
    ax.set_facecolor(PANEL)

    ax.set_title(
        title,
        fontsize=15,
        fontweight="bold",
        color=TEXT,
        loc="left",
        pad=25
    )

    ax.text(
        0, 1.02, subtitle,
        transform=ax.transAxes,
        color=MUTED,
        fontsize=9,
        va="bottom"
    )

    ax.set_xlabel("Quarter", color=TEXT, labelpad=10)
    ax.set_ylabel(ylabel, color=TEXT, labelpad=10)
    ax.tick_params(colors=MUTED)

    ax.grid(
        True, color=GRID,
        linestyle="--", linewidth=0.7, alpha=0.8
    )

    for spine in ax.spines.values():
        spine.set_color(GRID)

    ax.set_axisbelow(True)
    ax.yaxis.set_major_formatter(
        FuncFormatter(
            lambda x, pos:
            f"{x / 1_000_000:.1f}M"
            if abs(x) >= 1_000_000
            else f"{x / 1_000:.0f}K"
            if abs(x) >= 1_000
            else f"{x:.0f}"
        )
    )


def create_year_trend_chart(category_name, metric, ylabel):

    cat_df = state_master[
        state_master["Category"] == category_name
    ]

    valid = cat_df.dropna(subset=[metric])

    if valid.empty:
        print("No chart data:", category_name, metric)
        return

    # States-ai quarter-wise sum panni one India-level trend create pannuvom
    totals = (
        valid.groupby(["Year", "Quarter", "QuarterNumber"])[metric]
        .sum()
        .reset_index()
    )

    fig, ax = plt.subplots(figsize=(12, 6.5))

    available_years = sorted(
        totals["Year"].dropna().astype(int).unique()
    )

    for index, year in enumerate(available_years):

        year_df = totals[totals["Year"] == year]
        values = []

        for quarter in quarters:
            selected = year_df.loc[
                year_df["Quarter"] == quarter, metric
            ]

            values.append(
                selected.iloc[0] if not selected.empty else np.nan
            )

        values = np.array(values, dtype=float)

        if np.isnan(values).all():
            continue

        ax.plot(
            range(4), values,
            color=YEAR_COLORS[index % len(YEAR_COLORS)],
            marker="o", linewidth=2, markersize=5,
            label=str(year)
        )

    ax.set_xticks(range(4))
    ax.set_xticklabels(quarters)

    style_axis(
        fig, ax,
        f"MAP {category_name.upper()} | {metric.upper()}",
        "INDIA  •  SUM OF AVAILABLE STATE-LEVEL VALUES BY QUARTER",
        ylabel
    )

    ax.legend(
        title="Year", ncol=3,
        facecolor=PANEL, edgecolor=GRID,
        labelcolor=TEXT, fontsize=8
    )

    fig.tight_layout(pad=2)

    file_name = f"map_{category_name.lower()}_{metric.lower()}_trend_dark.png"
    save_file = os.path.join(chart_path, file_name)

    fig.savefig(
        save_file, dpi=200,
        facecolor=fig.get_facecolor(),
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)

    print("Chart saved:", save_file)


# Transaction count and amount trends
create_year_trend_chart(
    "Transaction", "Transactions", "Transactions"
)

create_year_trend_chart(
    "Transaction", "Amount", "Transaction Amount"
)

# Merchant and User registered count trends
create_year_trend_chart(
    "Merchant", "RegisteredCount", "Registered Merchants"
)

create_year_trend_chart(
    "User", "RegisteredCount", "Registered Users"
)


# ==========================================================
# STEP 10: TOP 10 STATE RANKING CHARTS
# ==========================================================

def create_ranking_chart(category_name, metric, title, ylabel):

    ranked = ranking_df[
        (ranking_df["RankingBasis"] == category_name)
        & (ranking_df["Metric"] == metric)
    ].copy()

    if ranked.empty:
        return

    top10 = ranked.head(10).sort_values(
        metric, ascending=True
    )

    fig, ax = plt.subplots(figsize=(11, 6.5))
    fig.patch.set_facecolor(BG)
    ax.set_facecolor(PANEL)

    bars = ax.barh(
        top10["State"],
        top10[metric],
        color=CYAN,
        height=0.65
    )

    for bar, value in zip(bars, top10[metric]):
        ax.annotate(
            f"{value:,.0f}",
            (bar.get_width(), bar.get_y() + bar.get_height() / 2),
            xytext=(6, 0),
            textcoords="offset points",
            va="center",
            color=TEXT,
            fontsize=8
        )

    first = ranked.iloc[0]

    ax.set_title(
        title,
        fontsize=15,
        fontweight="bold",
        color=TEXT,
        loc="left",
        pad=25
    )

    ax.text(
        0, 1.02,
        f"{first['RankingYear']} {first['RankingQuarter']}  •  LATEST MAX-COVERAGE QUARTER",
        transform=ax.transAxes,
        color=MUTED,
        fontsize=9
    )

    ax.set_xlabel(ylabel, color=TEXT)
    ax.tick_params(colors=MUTED)

    ax.grid(
        axis="x", color=GRID,
        linestyle="--", linewidth=0.7
    )

    for spine in ax.spines.values():
        spine.set_color(GRID)

    ax.set_axisbelow(True)
    ax.margins(x=0.2)
    fig.tight_layout(pad=2)

    safe_metric = re.sub(
        r"[^A-Za-z0-9_-]+", "_", metric
    )

    save_file = os.path.join(
        chart_path,
        f"map_{category_name.lower()}_{safe_metric}_top10_dark.png"
    )

    fig.savefig(
        save_file, dpi=220,
        facecolor=fig.get_facecolor(),
        bbox_inches="tight"
    )

    plt.show()
    plt.close(fig)

    print("Ranking chart saved:", save_file)


create_ranking_chart(
    "Transaction", "Transactions",
    "TOP 10 STATES | TRANSACTION COUNT",
    "Transactions"
)

create_ranking_chart(
    "Merchant", "RegisteredCount",
    "TOP 10 STATES | REGISTERED MERCHANTS",
    "Registered Merchants"
)

create_ranking_chart(
    "User", "RegisteredCount",
    "TOP 10 STATES | REGISTERED USERS",
    "Registered Users"
)


# ==========================================================
# STEP 11: SAVE MISSING FILE & ERROR REPORTS
# ==========================================================

missing_file = os.path.join(
    output_path, "map_missing_files_2018_2026.csv"
)

error_file = os.path.join(
    output_path, "map_read_errors_2018_2026.csv"
)

pd.DataFrame(missing_files).to_csv(
    missing_file, index=False
)

pd.DataFrame(read_errors).to_csv(
    error_file, index=False
)


# ==========================================================
# STEP 12: VERIFY SAVED CSV FILES
# ==========================================================

verified_states = pd.read_csv(state_master_file)
verified_districts = pd.read_csv(district_master_file)
verified_comparisons = pd.read_csv(comparison_file)
verified_rankings = pd.read_csv(ranking_master_file)

saved_charts = [
    file for file in os.listdir(chart_path)
    if file.endswith(".png")
]

print("\n========== FINAL VERIFICATION ==========")

print("State master rows:", len(verified_states))
print("District master rows:", len(verified_districts))
print("Comparison rows:", len(verified_comparisons))
print("Ranking rows:", len(verified_rankings))

print(
    "Categories:",
    sorted(verified_states["Category"].dropna().unique().tolist())
)

print("Charts saved:", len(saved_charts))
print("Missing files:", len(missing_files))
print("Read errors:", len(read_errors))

print("\nOutput folder:")
print(output_path)

print("\nState master preview:")
display(verified_states.head(10))

print("\nDistrict master preview:")
display(verified_districts.head(10))

print("\nMap analysis completed!")
```

```python
import os
import re
import pandas as pd

output_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\map\map_analysis_2018_2026"

def normalize_state_name(value):
    value = str(value).lower().replace("&", " and ")
    return re.sub(r"[^a-z0-9]", "", value)

target_name = normalize_state_name(
    "Dadra and Nagar Haveli and Daman and Diu"
)

fixed_files = 0
fixed_rows = 0

for root, dirs, files in os.walk(output_path):
    for filename in files:

        if not filename.lower().endswith(".csv"):
            continue

        file_path = os.path.join(root, filename)

        try:
            df = pd.read_csv(file_path)

            if "State" not in df.columns or "StateFolder" not in df.columns:
                continue

            matches = df["State"].apply(normalize_state_name) == target_name

            missing_slug = (
                df["StateFolder"].isna()
                | df["StateFolder"].astype(str).str.strip().eq("")
            )

            fix_mask = matches & missing_slug

            if fix_mask.any():
                fixed_rows += int(fix_mask.sum())
                df.loc[fix_mask, "StateFolder"] = (
                    "dadra-&-nagar-haveli-&-daman-&-diu"
                )

                df.to_csv(file_path, index=False)
                fixed_files += 1

        except Exception as error:
            print("Could not check:", filename, error)

print("CSV files corrected:", fixed_files)
print("Rows corrected:", fixed_rows)
print("State-folder mapping fix completed.")
```

```python
import os
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
from IPython.display import display

# ==========================================================
# STEP 1: LOAD EXISTING MAP MASTER CSV
# ==========================================================

output_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\map\map_analysis_2018_2026"

master_file = os.path.join(
    output_path,
    "map_state_master_all_categories_2018_2026.csv"
)

if not os.path.exists(master_file):
    raise FileNotFoundError(
        "Map state master CSV kidaikkala. Path-ai check pannu."
    )

df = pd.read_csv(master_file)

df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df["QuarterNumber"] = pd.to_numeric(
    df["QuarterNumber"], errors="coerce"
)

for col in ["Transactions", "Amount", "RegisteredCount"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df["Category"] = df["Category"].astype(str).str.title()

quarters = ["Q1", "Q2", "Q3", "Q4"]
full_years = list(range(2018, 2026))  # 2018-2025 complete-year comparison

# Output folders
charts_path = os.path.join(output_path, "all_state_charts")
summary_path = os.path.join(output_path, "best_year_best_state")

os.makedirs(charts_path, exist_ok=True)
os.makedirs(summary_path, exist_ok=True)

print("MAP MASTER LOADED")
print("Rows:", len(df))
print("States:", df["State"].nunique())
print("Categories:", sorted(df["Category"].dropna().unique()))


# ==========================================================
# STEP 2: PROFESSIONAL DARK THEME
# ==========================================================

BG = "#0B1020"
PANEL = "#111A2E"
GRID = "#26344D"
TEXT = "#E6EDF7"
MUTED = "#9FB0C8"
CYAN = "#38BDF8"
GREEN = "#34D399"

YEAR_COLORS = [
    "#38BDF8", "#A78BFA", "#34D399",
    "#FBBF24", "#FB7185", "#22D3EE",
    "#C084FC", "#4ADE80", "#F97316"
]

def compact_number(value):
    if not np.isfinite(value):
        return ""
    if abs(value) >= 1_000_000_000:
        return f"{value / 1_000_000_000:.1f}B"
    if abs(value) >= 1_000_000:
        return f"{value / 1_000_000:.1f}M"
    if abs(value) >= 1_000:
        return f"{value / 1_000:.0f}K"
    return f"{value:.0f}"


def style_axis(ax, title, ylabel):
    ax.set_facecolor(PANEL)
    ax.set_title(title, color=TEXT, fontsize=11, fontweight="bold")
    ax.set_xlabel("Quarter", color=TEXT)
    ax.set_ylabel(ylabel, color=TEXT, fontsize=9)
    ax.tick_params(colors=MUTED, labelsize=8)
    ax.grid(True, color=GRID, linestyle="--", linewidth=0.6)

    for spine in ax.spines.values():
        spine.set_color(GRID)

    ax.yaxis.set_major_formatter(
        FuncFormatter(lambda x, pos: compact_number(x))
    )
    ax.set_axisbelow(True)


# ==========================================================
# STEP 3: CREATE ONE 4-PANEL CHART FOR EVERY STATE
# Panels:
#   1. Transaction count
#   2. Transaction amount
#   3. Registered merchants
#   4. Registered users
# Every panel contains quarterly lines for 2018-2026.
# ==========================================================

metrics = [
    ("Transaction", "Transactions", "Transaction Count"),
    ("Transaction", "Amount", "Transaction Amount"),
    ("Merchant", "RegisteredCount", "Registered Merchants"),
    ("User", "RegisteredCount", "Registered Users")
]

states = sorted(df["State"].dropna().unique())
state_charts_created = 0

for state in states:

    state_df = df[df["State"] == state].copy()

    fig, axes = plt.subplots(2, 2, figsize=(16, 10))
    fig.patch.set_facecolor(BG)

    for ax, (category, metric, label) in zip(
        axes.flat, metrics
    ):

        category_df = state_df[
            state_df["Category"] == category
        ].copy()

        if category_df.empty:
            ax.set_facecolor(PANEL)
            ax.set_title(
                f"{label} — No data",
                color=TEXT
            )
            ax.set_xticks([])
            ax.set_yticks([])
            continue

        for i, year in enumerate(range(2018, 2027)):

            year_df = category_df[
                category_df["Year"] == year
            ]

            values = []

            for q in quarters:
                selected = year_df.loc[
                    year_df["Quarter"] == q, metric
                ]

                values.append(
                    selected.iloc[0]
                    if not selected.empty
                    else np.nan
                )

            values = np.array(values, dtype=float)

            if np.isnan(values).all():
                continue

            ax.plot(
                quarters,
                values,
                marker="o",
                markersize=3.5,
                linewidth=1.7,
                color=YEAR_COLORS[i],
                label=str(year)
            )

        style_axis(ax, label, label)

        # Legend in compact format
        ax.legend(
            title="Year",
            ncol=3,
            fontsize=7,
            title_fontsize=7,
            facecolor=PANEL,
            edgecolor=GRID,
            labelcolor=TEXT
        )

    fig.suptitle(
        f"MAP ANALYTICS | {state.upper()}",
        fontsize=18,
        fontweight="bold",
        color=TEXT,
        y=0.99
    )

    fig.text(
        0.5, 0.955,
        "PHONEPE PULSE • QUARTERLY TRENDS • 2018–2026",
        ha="center",
        color=MUTED,
        fontsize=10
    )

    fig.tight_layout(rect=[0, 0, 1, 0.94])

    safe_name = re.sub(
        r"[^A-Za-z0-9_-]+", "_", state
    )

    chart_file = os.path.join(
        charts_path,
        f"{safe_name}_map_analytics_dark.png"
    )

    fig.savefig(
        chart_file,
        dpi=180,
        facecolor=fig.get_facecolor(),
        bbox_inches="tight"
    )

    # Display every state chart in Jupyter
    plt.show()
    plt.close(fig)

    state_charts_created += 1

    print(
        f"Chart displayed and saved: {state}",
        flush=True
    )


# ==========================================================
# STEP 4: BEST YEAR ANALYSIS
# Transaction: full-year total (Q1-Q4), only 2018-2025.
# Merchant/User: Q4 year-end registered-count snapshot.
# 2026 is partial; do not rank it against complete years.
# ==========================================================

annual_rows = []

for year in full_years:

    # Transaction count and amount: sum quarter activity
    for metric in ["Transactions", "Amount"]:

        part = df[
            (df["Category"] == "Transaction")
            & (df["Year"] == year)
        ].copy()

        total = part[metric].sum(min_count=1)

        state_coverage = part.dropna(
            subset=[metric]
        )["State"].nunique()

        quarter_coverage = part.dropna(
            subset=[metric]
        )["Quarter"].nunique()

        annual_rows.append({
            "Category": "Transaction",
            "Metric": metric,
            "Year": year,
            "ComparisonBasis": "Full-year Q1-Q4 sum",
            "ObservedTotal": total,
            "StatesWithData": state_coverage,
            "QuartersWithData": quarter_coverage
        })

    # Merchant and User: Q4 cumulative snapshot
    for category in ["Merchant", "User"]:

        part = df[
            (df["Category"] == category)
            & (df["Year"] == year)
            & (df["Quarter"] == "Q4")
        ].copy()

        total = part["RegisteredCount"].sum(min_count=1)

        state_coverage = part.dropna(
            subset=["RegisteredCount"]
        )["State"].nunique()

        annual_rows.append({
            "Category": category,
            "Metric": "RegisteredCount",
            "Year": year,
            "ComparisonBasis": "Q4 year-end snapshot",
            "ObservedTotal": total,
            "StatesWithData": state_coverage,
            "QuartersWithData": 1 if not part.empty else 0
        })

annual_df = pd.DataFrame(annual_rows)

# Select the highest observed year for each valid metric
best_year_rows = []

for (category, metric), group in annual_df.groupby(
    ["Category", "Metric"]
):

    valid = group.dropna(subset=["ObservedTotal"])

    if valid.empty:
        continue

    best = valid.loc[valid["ObservedTotal"].idxmax()]

    best_year_rows.append({
        "Category": category,
        "Metric": metric,
        "BestYear": int(best["Year"]),
        "ObservedTotal": best["ObservedTotal"],
        "ComparisonBasis": best["ComparisonBasis"],
        "StatesWithData": int(best["StatesWithData"]),
        "QuartersWithData": int(best["QuartersWithData"])
    })

best_year_df = pd.DataFrame(best_year_rows)

print("\n========== ANNUAL TOTALS ==========")
display(annual_df)

print("\n========== BEST YEAR BY METRIC ==========")
display(best_year_df)

annual_file = os.path.join(
    summary_path, "map_annual_totals_2018_2025.csv"
)

best_year_file = os.path.join(
    summary_path, "map_best_year_by_metric.csv"
)

annual_df.to_csv(annual_file, index=False)
best_year_df.to_csv(best_year_file, index=False)


# ==========================================================
# STEP 5: BEST STATE FOR EACH YEAR AND METRIC
# Transaction: state yearly Q1-Q4 total.
# Merchant/User: state Q4 count.
# ==========================================================

best_state_rows = []

for year in full_years:

    # Transaction: annual total per state
    transaction_part = df[
        (df["Category"] == "Transaction")
        & (df["Year"] == year)
    ]

    for metric in ["Transactions", "Amount"]:

        by_state = (
            transaction_part.groupby("State")[metric]
            .sum(min_count=1)
            .dropna()
        )

        if not by_state.empty:

            best_state = by_state.idxmax()

            best_state_rows.append({
                "Category": "Transaction",
                "Metric": metric,
                "Year": year,
                "ComparisonBasis": "Full-year Q1-Q4 sum",
                "BestState": best_state,
                "Value": by_state.max(),
                "StatesCompared": len(by_state)
            })

    # Merchant/User: year-end Q4 snapshot
    for category in ["Merchant", "User"]:

        part = df[
            (df["Category"] == category)
            & (df["Year"] == year)
            & (df["Quarter"] == "Q4")
        ]

        by_state = (
            part.groupby("State")["RegisteredCount"]
            .max()
            .dropna()
        )

        if not by_state.empty:

            best_state = by_state.idxmax()

            best_state_rows.append({
                "Category": category,
                "Metric": "RegisteredCount",
                "Year": year,
                "ComparisonBasis": "Q4 year-end snapshot",
                "BestState": best_state,
                "Value": by_state.max(),
                "StatesCompared": len(by_state)
            })

best_state_df = pd.DataFrame(best_state_rows)

print("\n========== BEST STATE FOR EACH YEAR ==========")
display(best_state_df)

best_state_file = os.path.join(
    summary_path, "map_best_state_by_year_2018_2025.csv"
)

best_state_df.to_csv(best_state_file, index=False)


# ==========================================================
# STEP 6: LATEST AVAILABLE QUARTER — STATE RANKINGS
# Show best state for each metric for the latest period
# with maximum state coverage.
# ==========================================================

latest_rankings = []

for category, metric, label in metrics:

    part = df[
        (df["Category"] == category)
        & df[metric].notna()
    ].copy()

    if part.empty:
        continue

    coverage = (
        part.groupby(["Year", "Quarter", "QuarterNumber"])["State"]
        .nunique()
        .reset_index(name="StateCoverage")
    )

    max_coverage = coverage["StateCoverage"].max()

    periods = coverage[
        coverage["StateCoverage"] == max_coverage
    ].sort_values(["Year", "QuarterNumber"])

    period = periods.iloc[-1]

    selected = part[
        (part["Year"] == period["Year"])
        & (part["Quarter"] == period["Quarter"])
    ].copy()

    selected = (
        selected.groupby("State", as_index=False)[metric]
        .max()
        .dropna()
        .sort_values(metric, ascending=False)
        .reset_index(drop=True)
    )

    selected.insert(0, "Rank", range(1, len(selected) + 1))
    selected["Category"] = category
    selected["Metric"] = metric
    selected["Year"] = int(period["Year"])
    selected["Quarter"] = period["Quarter"]
    selected["StatesCompared"] = len(selected)

    latest_rankings.append(selected)

latest_rank_df = pd.concat(
    latest_rankings, ignore_index=True
) if latest_rankings else pd.DataFrame()

latest_rank_file = os.path.join(
    summary_path, "map_latest_state_rankings.csv"
)

latest_rank_df.to_csv(latest_rank_file, index=False)

print("\n========== LATEST STATE RANKINGS ==========")

for category in ["Transaction", "Merchant", "User"]:

    for metric in latest_rank_df.loc[
        latest_rank_df["Category"] == category, "Metric"
    ].unique():

        print(f"\n{category} — {metric}")

        display(
            latest_rank_df[
                (latest_rank_df["Category"] == category)
                & (latest_rank_df["Metric"] == metric)
            ][["Rank", "State", "Year", "Quarter", metric]].head(10)
        )


# ==========================================================
# STEP 7: PROFESSIONAL CHART — BEST YEAR COMPARISON
# ==========================================================

fig, axes = plt.subplots(2, 2, figsize=(16, 10))
fig.patch.set_facecolor(BG)

annual_metrics = [
    ("Transaction", "Transactions", "Annual Transactions"),
    ("Transaction", "Amount", "Annual Transaction Amount"),
    ("Merchant", "RegisteredCount", "Year-end Registered Merchants"),
    ("User", "RegisteredCount", "Year-end Registered Users")
]

for ax, (category, metric, label) in zip(
    axes.flat, annual_metrics
):

    part = annual_df[
        (annual_df["Category"] == category)
        & (annual_df["Metric"] == metric)
    ].sort_values("Year")

    if part.empty:
        ax.set_facecolor(PANEL)
        ax.set_title(label, color=TEXT)
        continue

    best_year = int(
        part.loc[part["ObservedTotal"].idxmax(), "Year"]
    )

    bar_colors = [
        GREEN if int(year) == best_year else CYAN
        for year in part["Year"]
    ]

    ax.bar(
        part["Year"].astype(str),
        part["ObservedTotal"],
        color=bar_colors
    )

    style_axis(
        ax,
        f"{label}\nBest observed year: {best_year}",
        "Observed total"
    )

    ax.set_xlabel("Year", color=TEXT)
    ax.tick_params(axis="x", rotation=45)

fig.suptitle(
    "MAP ANALYTICS | BEST YEAR COMPARISON",
    fontsize=18,
    fontweight="bold",
    color=TEXT
)

fig.tight_layout(rect=[0, 0, 1, 0.96])

best_year_chart = os.path.join(
    charts_path, "map_best_year_comparison_dark.png"
)

fig.savefig(
    best_year_chart,
    dpi=220,
    facecolor=fig.get_facecolor(),
    bbox_inches="tight"
)

plt.show()
plt.close(fig)


# ==========================================================
# STEP 8: SAVE AND VERIFY
# ==========================================================

best_file_list = [
    annual_file,
    best_year_file,
    best_state_file,
    latest_rank_file,
    best_year_chart
]

saved_state_charts = [
    file for file in os.listdir(charts_path)
    if file.endswith("_map_analytics_dark.png")
]

verified_master = pd.read_csv(master_file)
verified_best_year = pd.read_csv(best_year_file)
verified_best_state = pd.read_csv(best_state_file)
verified_rankings = pd.read_csv(latest_rank_file)

print("\n========== FINAL VERIFICATION ==========")
print("States in master:", verified_master["State"].nunique())
print("Individual state charts saved:", len(saved_state_charts))
print("State charts displayed:", len(saved_state_charts))
print("Best-year records:", len(verified_best_year))
print("Best-state-per-year records:", len(verified_best_state))
print("Latest ranking rows:", len(verified_rankings))

print("\nOutput folder:", output_path)
print("Every state chart folder:", charts_path)
print("Best-year and state summaries:", summary_path)

print("\nBest year summary:")
display(verified_best_year)

print("\nBest state for each year:")
display(verified_best_state)

print("\nSaved-file verification:")
for file_path in best_file_list:
    print(os.path.basename(file_path), "exists:", os.path.exists(file_path))

print("\nMAP STATE CHARTS AND BEST-YEAR ANALYSIS COMPLETED!")
```

```python
import os
import json
import pandas as pd

# 1. Source and output folders
base_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\state"

output_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\transaction_analysis_2018_2026"

os.makedirs(output_path, exist_ok=True)

state_output_path = os.path.join(output_path, "state_masters")
os.makedirs(state_output_path, exist_ok=True)

print("Source folder exists:", os.path.exists(base_path))

# 2. State folder names-ai readable format-ku convert pannuvom
state_name_mapping = {
    "andaman-&-nicobar-islands": "Andaman & Nicobar Islands",
    "andhra-pradesh": "Andhra Pradesh",
    "arunachal-pradesh": "Arunachal Pradesh",
    "chhattisgarh": "Chhattisgarh",
    "dadra-&-nagar-haveli-&-daman-&-diu":
        "Dadra and Nagar Haveli and Daman and Diu",
    "himachal-pradesh": "Himachal Pradesh",
    "jammu-&-kashmir": "Jammu & Kashmir",
    "madhya-pradesh": "Madhya Pradesh",
    "tamil-nadu": "Tamil Nadu",
    "uttar-pradesh": "Uttar Pradesh",
    "west-bengal": "West Bengal"
}

def readable_state_name(folder_name):
    if folder_name in state_name_mapping:
        return state_name_mapping[folder_name]

    return folder_name.replace("-", " ").replace("&", "&").title()


# 3. JSON files read pannuvom
records = []
missing_files = []
read_errors = []

if not os.path.exists(base_path):
    raise FileNotFoundError(
        "Transaction state folder kidaikkala. Path-ai check pannu."
    )

state_folders = [
    folder for folder in os.listdir(base_path)
    if os.path.isdir(os.path.join(base_path, folder))
]

print("State folders found:", len(state_folders))

for state_folder in sorted(state_folders):

    state_path = os.path.join(base_path, state_folder)
    state_name = readable_state_name(state_folder)

    # Year folders: 2018 to 2026
    for year in range(2018, 2027):

        year_path = os.path.join(state_path, str(year))

        # 2026-la Q1, Q2 mattum expected
        quarters = [1, 2] if year == 2026 else [1, 2, 3, 4]

        # Andha year folder illainaa skip pannuvom
        if not os.path.isdir(year_path):
            continue

        for quarter in quarters:

            json_path = os.path.join(
                year_path, f"{quarter}.json"
            )

            if not os.path.isfile(json_path):
                missing_files.append({
                    "State": state_name,
                    "StateFolder": state_folder,
                    "Year": year,
                    "Quarter": f"Q{quarter}",
                    "ExpectedFile": json_path
                })
                continue

            try:
                with open(json_path, "r", encoding="utf-8") as file:
                    content = json.load(file)

                data = content.get("data") or {}
                transaction_data = data.get("transactionData") or []

                for item in transaction_data:

                    category = item.get("name", "Unknown")

                    instruments = item.get("paymentInstruments") or []

                    # TOTAL metric irundhaa adhai mattum use pannuvom
                    total_instruments = [
                        instrument for instrument in instruments
                        if str(instrument.get("type", "")).upper() == "TOTAL"
                    ]

                    selected_instruments = (
                        total_instruments if total_instruments
                        else instruments
                    )

                    transaction_count = sum(
                        float(instrument.get("count") or 0)
                        for instrument in selected_instruments
                    )

                    transaction_amount = sum(
                        float(instrument.get("amount") or 0)
                        for instrument in selected_instruments
                    )

                    records.append({
                        "State": state_name,
                        "StateFolder": state_folder,
                        "Year": year,
                        "Quarter": f"Q{quarter}",
                        "QuarterNumber": quarter,
                        "Category": category,
                        "Transactions": transaction_count,
                        "Amount": transaction_amount
                    })

            except Exception as error:
                read_errors.append({
                    "State": state_name,
                    "Year": year,
                    "Quarter": f"Q{quarter}",
                    "File": json_path,
                    "Error": str(error)
                })


# 4. DataFrame create pannuvom
columns = [
    "State", "StateFolder", "Year", "Quarter",
    "QuarterNumber", "Category", "Transactions", "Amount"
]

master_df = pd.DataFrame(records, columns=columns)

if master_df.empty:
    raise ValueError(
        "Data read aagala. JSON path and JSON structure-ai verify pannanum."
    )

master_df = master_df.sort_values(
    ["Year", "QuarterNumber", "State", "Category"]
).reset_index(drop=True)

# 5. Main master CSV
master_file = os.path.join(
    output_path, "phonepe_india_master_2018_2026.csv"
)
master_df.to_csv(master_file, index=False)

# 6. State-wise master CSV files
for state_folder, state_data in master_df.groupby("StateFolder"):

    state_file = os.path.join(
        state_output_path,
        f"{state_folder}_transactions.csv"
    )

    state_data.to_csv(state_file, index=False)

# 7. State total summary
state_totals = (
    master_df.groupby(["State", "StateFolder"], as_index=False)
    .agg(
        Transactions=("Transactions", "sum"),
        Amount=("Amount", "sum")
    )
    .sort_values("Transactions", ascending=False)
)

state_totals.to_csv(
    os.path.join(
        output_path,
        "phonepe_india_state_totals_2018_2026.csv"
    ),
    index=False
)

# 8. Year-wise totals
yearly_totals = (
    master_df.groupby("Year", as_index=False)
    .agg(
        Transactions=("Transactions", "sum"),
        Amount=("Amount", "sum")
    )
)

yearly_totals.to_csv(
    os.path.join(
        output_path,
        "phonepe_india_yearly_totals_2018_2026.csv"
    ),
    index=False
)

# 9. Category-wise totals
category_totals = (
    master_df.groupby("Category", as_index=False)
    .agg(
        Transactions=("Transactions", "sum"),
        Amount=("Amount", "sum")
    )
    .sort_values("Transactions", ascending=False)
)

category_totals.to_csv(
    os.path.join(
        output_path,
        "phonepe_india_category_totals_2018_2026.csv"
    ),
    index=False
)

# 10. Quarter comparison: State + Year + Category
quarter_comparison = master_df.pivot_table(
    index=["State", "StateFolder", "Year", "Category"],
    columns="Quarter",
    values="Transactions",
    aggfunc="sum",
    fill_value=0
).reset_index()

for quarter in ["Q1", "Q2", "Q3", "Q4"]:
    if quarter not in quarter_comparison.columns:
        quarter_comparison[quarter] = 0

quarter_comparison["Q2_minus_Q1"] = (
    quarter_comparison["Q2"] - quarter_comparison["Q1"]
)

quarter_comparison["Q3_minus_Q2"] = (
    quarter_comparison["Q3"] - quarter_comparison["Q2"]
)

quarter_comparison["Q4_minus_Q3"] = (
    quarter_comparison["Q4"] - quarter_comparison["Q3"]
)

quarter_comparison.to_csv(
    os.path.join(
        output_path,
        "phonepe_india_quarter_comparison_2018_2026.csv"
    ),
    index=False
)

# 11. Missing-file and read-error reports
pd.DataFrame(missing_files).to_csv(
    os.path.join(output_path, "missing_transaction_json_files.csv"),
    index=False
)

pd.DataFrame(read_errors).to_csv(
    os.path.join(output_path, "transaction_json_read_errors.csv"),
    index=False
)

# 12. Final verification
print("\n========== TRANSACTION MASTER COMPLETE ==========")
print("Master CSV:", master_file)
print("Total master rows:", len(master_df))
print("Unique states:", master_df["StateFolder"].nunique())
print("Years found:", sorted(master_df["Year"].unique().tolist()))
print("Categories found:", master_df["Category"].nunique())
print("Category names:", sorted(master_df["Category"].unique().tolist()))
print("Missing JSON files:", len(missing_files))
print("Read errors:", len(read_errors))

print("\nSample master data:")
display(master_df.head(10))

print("\nYearly transaction totals:")
display(yearly_totals)

print("\nCategory-wise totals:")
display(category_totals)

print("\nOutput folder:")
print(output_path)
```

```python
import os
import pandas as pd
import numpy as np

# File paths
agg_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\aggregated\transaction\country\india\transaction_analysis_2018_2026\phonepe_india_master_2018_2026.csv"

map_path = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\map\map_analysis_2018_2026\transaction\map_transaction_state_master_2018_2026.csv"

output_folder = os.path.dirname(agg_path)

# 1. Load both master CSVs
agg = pd.read_csv(agg_path)
map_df = pd.read_csv(map_path)

print("Aggregated rows:", len(agg))
print("Map rows:", len(map_df))
print("Map categories:", map_df["Category"].dropna().unique())

# 2. Aggregated JSON-la amount field illai.
# Incorrect 0.0 values-ai blank-aa correct pannuvom.
agg["Amount"] = np.nan
agg.to_csv(agg_path, index=False)

print("\nAggregated Amount column corrected: unavailable values blank")

# 3. Compare state-quarter transaction counts
keys = ["StateFolder", "Year", "Quarter"]

agg["StateFolder"] = agg["StateFolder"].astype(str).str.strip().str.lower()
map_df["StateFolder"] = map_df["StateFolder"].astype(str).str.strip().str.lower()

agg["Year"] = pd.to_numeric(agg["Year"], errors="coerce")
map_df["Year"] = pd.to_numeric(map_df["Year"], errors="coerce")

agg["Quarter"] = agg["Quarter"].astype(str).str.upper().str.strip()
map_df["Quarter"] = map_df["Quarter"].astype(str).str.upper().str.strip()

agg["Transactions"] = pd.to_numeric(
    agg["Transactions"], errors="coerce"
)
map_df["Transactions"] = pd.to_numeric(
    map_df["Transactions"], errors="coerce"
)

# Aggregated categories (P2P, Retail, Utility) total
agg_total = (
    agg.groupby(keys, as_index=False)["Transactions"]
       .sum()
       .rename(columns={"Transactions": "Aggregated_Transactions"})
)

# Map state master-la duplicate category rows irundhaal aggregate pannum
map_total = (
    map_df.groupby(keys, as_index=False)["Transactions"]
          .sum()
          .rename(columns={"Transactions": "Map_Transactions"})
)

comparison = agg_total.merge(
    map_total,
    on=keys,
    how="outer",
    indicator=True
)

# Difference only if both datasets have matching records
comparison["Difference"] = (
    comparison["Aggregated_Transactions"] - comparison["Map_Transactions"]
)

comparison["Status"] = np.where(
    comparison["_merge"] != "both",
    "Missing from one dataset",
    np.where(
        np.isclose(
            comparison["Aggregated_Transactions"],
            comparison["Map_Transactions"],
            rtol=1e-9,
            atol=1e-6,
            equal_nan=False
        ),
        "Match",
        "Mismatch"
    )
)

comparison = comparison.drop(columns=["_merge"])

comparison_path = os.path.join(
    output_folder,
    "aggregated_vs_map_transaction_validation.csv"
)
comparison.to_csv(comparison_path, index=False)

# 4. Validation summary
print("\n========== VALIDATION SUMMARY ==========")
print("Total comparison records:", len(comparison))
print("Exact matches:", (comparison["Status"] == "Match").sum())
print("Mismatches:", (comparison["Status"] == "Mismatch").sum())
print(
    "Missing in one dataset:",
    (comparison["Status"] == "Missing from one dataset").sum()
)

print("\nStatus counts:")
display(comparison["Status"].value_counts().rename_axis("Status").to_frame("Records"))

print("\nFirst comparison records:")
display(comparison.head(10))

print("\nMismatches (first 20):")
display(
    comparison[comparison["Status"] != "Match"].head(20)
)

print("\nValidation report saved at:")
print(comparison_path)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# 1. Map master CSV path
map_folder = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\map\map_analysis_2018_2026"

master_file = os.path.join(
    map_folder,
    "map_state_master_all_categories_2018_2026.csv"
)

# 2. Chart output folder
chart_folder = os.path.join(
    map_folder,
    "state_4_panel_charts"
)

os.makedirs(chart_folder, exist_ok=True)

# 3. Load master data
if not os.path.exists(master_file):
    raise FileNotFoundError(
        "Map master CSV kidaikkala. File path check pannu."
    )

df = pd.read_csv(master_file)

required_columns = [
    "State", "StateFolder", "Year", "Quarter",
    "QuarterNumber", "Category", "Transactions",
    "Amount", "RegisteredCount"
]

missing_columns = [
    col for col in required_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(f"Required columns missing: {missing_columns}")

print("Map master loaded:", df.shape)
print("Categories:", sorted(df["Category"].dropna().unique()))

# 4. Prepare timeline
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df["QuarterNumber"] = pd.to_numeric(
    df["QuarterNumber"], errors="coerce"
)

df = df.dropna(subset=["Year", "QuarterNumber"])
df["Year"] = df["Year"].astype(int)
df["QuarterNumber"] = df["QuarterNumber"].astype(int)

periods = (
    df[["Year", "QuarterNumber"]]
    .drop_duplicates()
    .sort_values(["Year", "QuarterNumber"])
    .to_records(index=False)
)

period_keys = [int(year) * 10 + int(q) for year, q in periods]
period_labels = [
    f"{year} Q{q}" for year, q in periods
]

# 5. Helper: particular state + category metric values
def get_metric(state_df, category, metric):

    temp = state_df[
        state_df["Category"].astype(str).str.strip().str.lower()
        == category.lower()
    ].copy()

    if temp.empty:
        return pd.Series(index=period_keys, dtype=float)

    temp["PeriodKey"] = (
        temp["Year"] * 10 + temp["QuarterNumber"]
    ).astype(int)

    temp[metric] = pd.to_numeric(temp[metric], errors="coerce")

    # Merchant/User registeredCount is a snapshot:
    # same quarter duplicate irundhaal last value use pannuvom.
    if metric == "RegisteredCount":
        temp = temp.sort_values(["Year", "QuarterNumber"])
        values = temp.groupby("PeriodKey")[metric].last()
    else:
        values = temp.groupby("PeriodKey")[metric].sum(min_count=1)

    return values.reindex(period_keys)


# 6. Create 4-panel charts for every state
state_groups = df.groupby("StateFolder", dropna=True)

chart_count = 0
summary_rows = []

for state_folder, state_df in state_groups:

    state_name = state_df["State"].dropna().iloc[0]
    state_title = str(state_name)

    transaction_count = get_metric(
        state_df, "Transaction", "Transactions"
    )

    transaction_amount = get_metric(
        state_df, "Transaction", "Amount"
    )

    registered_merchants = get_metric(
        state_df, "Merchant", "RegisteredCount"
    )

    registered_users = get_metric(
        state_df, "User", "RegisteredCount"
    )

    # Scale values for readable charts
    chart_metrics = [
        (
            "Transaction Count",
            transaction_count / 1_000_000,
            "Transactions (Millions)",
            "#38BDF8"
        ),
        (
            "Transaction Amount",
            transaction_amount / 1_000_000_000_000,
            "Amount (₹ Trillion)",
            "#F59E0B"
        ),
        (
            "Registered Merchants",
            registered_merchants / 1_000_000,
            "Merchants (Millions)",
            "#34D399"
        ),
        (
            "Registered Users",
            registered_users / 1_000_000,
            "Users (Millions)",
            "#C084FC"
        )
    ]

    fig, axes = plt.subplots(2, 2, figsize=(16, 10))
    fig.patch.set_facecolor("#0B1120")

    for ax, (title, values, ylabel, line_color) in zip(
        axes.flatten(), chart_metrics
    ):

        ax.set_facecolor("#111827")

        y = values.to_numpy(dtype=float)
        x = np.arange(len(period_labels))

        ax.plot(
            x, y,
            color=line_color,
            linewidth=2.2,
            marker="o",
            markersize=3.5
        )

        ax.set_title(
            title, color="white",
            fontsize=12, fontweight="bold", pad=10
        )

        ax.set_ylabel(ylabel, color="#D1D5DB", fontsize=9)
        ax.tick_params(axis="both", colors="#D1D5DB", labelsize=7)

        ax.grid(
            True, linestyle="--",
            alpha=0.22, color="#9CA3AF"
        )

        for spine in ax.spines.values():
            spine.set_color("#374151")

        # Ella quarter label-um clutter aagama display pannuvom
        tick_positions = np.arange(len(period_labels))
        ax.set_xticks(tick_positions[::2])
        ax.set_xticklabels(
            [period_labels[i] for i in tick_positions[::2]],
            rotation=55,
            ha="right",
            fontsize=7,
            color="#D1D5DB"
        )

    fig.suptitle(
        f"{state_title} | PhonePe Map Analysis",
        color="white",
        fontsize=17,
        fontweight="bold",
        y=0.99
    )

    fig.text(
        0.5, 0.015,
        "Source: PhonePe Pulse | Quarterly trends | 2018–2026 available data",
        ha="center",
        color="#9CA3AF",
        fontsize=9
    )

    plt.tight_layout(rect=[0, 0.045, 1, 0.95])

    # Safe filename
    safe_name = "".join(
        ch if ch.isalnum() or ch in "-_ " else "_"
        for ch in str(state_folder)
    ).strip().replace(" ", "_")

    save_file = os.path.join(
        chart_folder,
        f"{safe_name}_4_panel_analysis.png"
    )

    plt.savefig(
        save_file,
        dpi=180,
        facecolor=fig.get_facecolor(),
        bbox_inches="tight"
    )

    # Notebook-la first chart mattum preview
    if chart_count == 0:
        plt.show()

    plt.close(fig)
    chart_count += 1

    # Summary report-ku latest available value
    def latest_value(series):
        valid = series.dropna()
        return valid.iloc[-1] if len(valid) else np.nan

    summary_rows.append({
        "State": state_title,
        "StateFolder": state_folder,
        "QuartersAvailable": int(
            state_df[["Year", "QuarterNumber"]]
            .drop_duplicates().shape[0]
        ),
        "LatestTransactionCount": latest_value(transaction_count),
        "LatestTransactionAmount": latest_value(transaction_amount),
        "LatestRegisteredMerchants": latest_value(registered_merchants),
        "LatestRegisteredUsers": latest_value(registered_users)
    })


# 7. Save chart summary
summary_df = pd.DataFrame(summary_rows)

summary_file = os.path.join(
    chart_folder,
    "map_state_4_panel_chart_summary.csv"
)

summary_df.to_csv(summary_file, index=False)

print("\n========== MAP STATE CHARTS COMPLETE ==========")
print("States processed:", chart_count)
print("Charts saved:", chart_count)
print("Output folder:", chart_folder)
print("Summary CSV:", summary_file)

display(summary_df.head(10))
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Folder paths
map_folder = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\map\map_analysis_2018_2026"

master_file = os.path.join(
    map_folder,
    "map_state_master_all_categories_2018_2026.csv"
)

output_folder = os.path.join(
    map_folder,
    "best_year_best_state"
)

chart_folder = os.path.join(output_folder, "charts")

os.makedirs(output_folder, exist_ok=True)
os.makedirs(chart_folder, exist_ok=True)

# 2. Load data
df = pd.read_csv(master_file)

df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df["QuarterNumber"] = pd.to_numeric(
    df["QuarterNumber"], errors="coerce"
)

for col in ["Transactions", "Amount", "RegisteredCount"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df["Category"] = (
    df["Category"].astype(str).str.strip().str.lower()
)

print("Master loaded:", df.shape)
print("States:", df["StateFolder"].nunique())


# =====================================================
# PART A — TRANSACTION ANNUAL ANALYSIS
# =====================================================

transaction = df[df["Category"] == "transaction"].copy()

# State + year basis-la quarter count and annual totals
annual_transaction = (
    transaction.groupby(
        ["State", "StateFolder", "Year"], as_index=False
    )
    .agg(
        Transactions=("Transactions", "sum"),
        Amount=("Amount", "sum"),
        QuartersAvailable=("QuarterNumber", "nunique")
    )
)

# Full year = all 4 quarters available
annual_transaction["YearStatus"] = np.where(
    annual_transaction["QuartersAvailable"] == 4,
    "Full year",
    "Partial year"
)

annual_transaction.to_csv(
    os.path.join(output_folder, "map_transaction_annual_state_totals.csv"),
    index=False
)

# Full-year state records only, 2018-2025
full_year = annual_transaction[
    (annual_transaction["Year"] <= 2025)
    & (annual_transaction["QuartersAvailable"] == 4)
].copy()

expected_states = transaction["StateFolder"].nunique()

# How many states have all 4 quarters for each year?
coverage = (
    full_year.groupby("Year")["StateFolder"]
    .nunique()
    .rename("StatesWithFullYear")
    .reset_index()
)

# State rank within each year
full_year["TransactionCountRank"] = (
    full_year.groupby("Year")["Transactions"]
    .rank(method="min", ascending=False)
)

full_year["TransactionAmountRank"] = (
    full_year.groupby("Year")["Amount"]
    .rank(method="min", ascending=False)
)

full_year.to_csv(
    os.path.join(output_folder, "map_transaction_state_annual_rankings.csv"),
    index=False
)

# Best state per year: transaction count
best_count = (
    full_year.sort_values(
        ["Year", "Transactions"], ascending=[True, False]
    )
    .groupby("Year", as_index=False)
    .head(1)
    .sort_values("Year")
)

best_count.to_csv(
    os.path.join(
        output_folder,
        "map_best_transaction_state_by_year.csv"
    ),
    index=False
)

# Best state per year: transaction amount
best_amount = (
    full_year.sort_values(
        ["Year", "Amount"], ascending=[True, False]
    )
    .groupby("Year", as_index=False)
    .head(1)
    .sort_values("Year")
)

best_amount.to_csv(
    os.path.join(
        output_folder,
        "map_best_transaction_amount_state_by_year.csv"
    ),
    index=False
)

# India-wide trend: use only years where all states
# have complete Q1-Q4 data
india_transaction = (
    full_year.groupby("Year", as_index=False)
    .agg(
        Transactions=("Transactions", "sum"),
        Amount=("Amount", "sum"),
        StatesWithFullYear=("StateFolder", "nunique")
    )
)

india_transaction["CoverageComplete"] = (
    india_transaction["StatesWithFullYear"] == expected_states
)

india_transaction.to_csv(
    os.path.join(output_folder, "map_india_annual_transaction_trend.csv"),
    index=False
)


# =====================================================
# PART B — MERCHANT AND USER YEAR-END SNAPSHOTS
# =====================================================

registered = df[
    df["Category"].isin(["merchant", "user"])
].copy()

# Registered counts are snapshots, so DO NOT sum Q1+Q2+Q3+Q4.
# Compare Q4 snapshots for full years 2018-2025.
year_end = registered[
    (registered["Year"] >= 2018)
    & (registered["Year"] <= 2025)
    & (registered["QuarterNumber"] == 4)
].copy()

year_end.to_csv(
    os.path.join(
        output_folder,
        "map_registered_merchant_user_year_end_snapshots.csv"
    ),
    index=False
)

# Rank states per year using the Q4 snapshot
year_end["Rank"] = (
    year_end.groupby(["Category", "Year"])["RegisteredCount"]
    .rank(method="min", ascending=False)
)

year_end.to_csv(
    os.path.join(
        output_folder,
        "map_registered_merchant_user_annual_state_rankings.csv"
    ),
    index=False
)

best_registration_state = (
    year_end.sort_values(
        ["Category", "Year", "RegisteredCount"],
        ascending=[True, True, False]
    )
    .groupby(["Category", "Year"], as_index=False)
    .head(1)
    .sort_values(["Category", "Year"])
)

best_registration_state.to_csv(
    os.path.join(
        output_folder,
        "map_best_merchant_user_state_by_year.csv"
    ),
    index=False
)

# 2026 Q2 is the latest available snapshot in this dataset.
latest_snapshot = registered[
    (registered["Year"] == 2026)
    & (registered["QuarterNumber"] == 2)
].copy()

latest_snapshot["Rank"] = (
    latest_snapshot.groupby("Category")["RegisteredCount"]
    .rank(method="min", ascending=False)
)

latest_snapshot = latest_snapshot.sort_values(
    ["Category", "Rank"]
)

latest_snapshot.to_csv(
    os.path.join(
        output_folder,
        "map_latest_2026_q2_merchant_user_state_rankings.csv"
    ),
    index=False
)


# =====================================================
# PART C — DARK-THEME CHARTS
# =====================================================

def save_bar_chart(data, x_col, y_col, title, ylabel, filename):
    if data.empty:
        print("Skipped empty chart:", title)
        return

    fig, ax = plt.subplots(figsize=(13, 6))
    fig.patch.set_facecolor("#0B1120")
    ax.set_facecolor("#111827")

    ax.bar(
        data[x_col].astype(str),
        data[y_col],
        color="#38BDF8"
    )

    ax.set_title(title, color="white", fontsize=15, pad=14)
    ax.set_ylabel(ylabel, color="#D1D5DB")
    ax.tick_params(axis="x", colors="#D1D5DB", rotation=35)
    ax.tick_params(axis="y", colors="#D1D5DB")
    ax.grid(axis="y", linestyle="--", alpha=0.2)

    for spine in ax.spines.values():
        spine.set_color("#374151")

    plt.tight_layout()
    plt.savefig(
        os.path.join(chart_folder, filename),
        dpi=180,
        bbox_inches="tight",
        facecolor=fig.get_facecolor()
    )
    plt.close(fig)


# Chart 1: India's annual transaction count
complete_india = india_transaction[
    india_transaction["CoverageComplete"]
].copy()

save_bar_chart(
    complete_india,
    "Year",
    "Transactions",
    "PhonePe India — Annual Transaction Count",
    "Transactions",
    "india_annual_transaction_count.png"
)

# Chart 2: Best state by transaction count, each full year
save_bar_chart(
    best_count,
    "Year",
    "Transactions",
    "Top State by Annual Transaction Count",
    "Transactions",
    "best_state_transaction_count_by_year.png"
)

# Chart 3: Best state by transaction amount, each full year
save_bar_chart(
    best_amount,
    "Year",
    "Amount",
    "Top State by Annual Transaction Amount",
    "Amount (as provided in source)",
    "best_state_transaction_amount_by_year.png"
)

# Chart 4 and 5: Registration snapshots across years
for category in ["merchant", "user"]:

    category_data = year_end[
        year_end["Category"] == category
    ]

    if category_data.empty:
        print("No year-end snapshot data for:", category)
        continue

    state_counts = (
        category_data.groupby("Year", as_index=False)
        ["RegisteredCount"].sum()
    )

    save_bar_chart(
        state_counts,
        "Year",
        "RegisteredCount",
        f"State-wise Registered {category.title()} Counts — Q4 Snapshot",
        "Sum of state registration snapshots",
        f"{category}_registration_q4_trend.png"
    )


# =====================================================
# PART D — FINAL REPORT
# =====================================================

print("\n========== MAP FINAL ANALYSIS: STEP 2 ==========")

print("\nAnnual transaction records:", len(annual_transaction))
print("Full-year state records (2018-2025):", len(full_year))
print("Expected states:", expected_states)

print("\nIndia yearly data coverage:")
display(india_transaction)

print("\nBest state by transaction count each year:")
display(
    best_count[
        ["Year", "State", "Transactions", "QuartersAvailable"]
    ]
)

print("\nBest state by transaction amount each year:")
display(
    best_amount[
        ["Year", "State", "Amount", "QuartersAvailable"]
    ]
)

print("\nBest merchant/user state by year-end snapshot:")
display(
    best_registration_state[
        ["Category", "Year", "State", "RegisteredCount"]
    ]
)

print("\nTop states in latest available 2026 Q2 snapshot:")
display(
    latest_snapshot[
        ["Category", "Rank", "State", "RegisteredCount"]
    ].head(20)
)

print("\nCharts folder:", chart_folder)
print("Analysis output folder:", output_folder)
print("Map best-year analysis files saved successfully.")
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. File paths
map_folder = r"C:\Users\HP\OneDrive\Documents\fita project\pulse-main\pulse-main\data\map\map_analysis_2018_2026"

district_file = os.path.join(
    map_folder,
    "map_district_master_all_categories_2018_2026.csv"
)

output_folder = os.path.join(
    map_folder,
    "district_analysis_2018_2026"
)

chart_folder = os.path.join(output_folder, "charts")

os.makedirs(output_folder, exist_ok=True)
os.makedirs(chart_folder, exist_ok=True)

# 2. Load district master
if not os.path.exists(district_file):
    raise FileNotFoundError(
        "District master CSV kidaikkala. File path check pannu."
    )

df = pd.read_csv(district_file)

print("District master shape:", df.shape)
print("Categories:", df["Category"].dropna().unique())
print("Unique districts:", df["District"].nunique())

# 3. Clean columns
for col in ["Year", "QuarterNumber", "Transactions",
            "Amount", "RegisteredCount"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df["Category"] = (
    df["Category"].astype(str).str.strip().str.lower()
)

df["Quarter"] = (
    df["Quarter"].astype(str).str.upper().str.strip()
)

# District name unique-a identify panna State-um include pannuvom
district_keys = ["State", "StateFolder", "District"]


# ==================================================
# PART A — TRANSACTION DISTRICT RANKINGS
# ==================================================

transaction = df[df["Category"] == "transaction"].copy()

# Latest available quarter snapshot: 2026 Q2
latest_transaction = transaction[
    (transaction["Year"] == 2026)
    & (transaction["QuarterNumber"] == 2)
].copy()

latest_transaction_ranking = (
    latest_transaction.groupby(
        district_keys, as_index=False
    )
    .agg(
        Transactions=("Transactions", "sum"),
        Amount=("Amount", "sum")
    )
    .sort_values("Transactions", ascending=False)
)

latest_transaction_ranking.insert(
    0, "Rank",
    range(1, len(latest_transaction_ranking) + 1)
)

latest_transaction_ranking.to_csv(
    os.path.join(
        output_folder,
        "district_transaction_ranking_2026_Q2.csv"
    ),
    index=False
)

# Annual transaction rankings: full years 2018–2025
annual_transaction = (
    transaction[
        (transaction["Year"] >= 2018)
        & (transaction["Year"] <= 2025)
    ]
    .groupby(
        district_keys + ["Year"],
        as_index=False
    )
    .agg(
        Transactions=("Transactions", "sum"),
        Amount=("Amount", "sum"),
        QuartersAvailable=("QuarterNumber", "nunique")
    )
)

# 4-quarter records only for annual comparison
annual_complete = annual_transaction[
    annual_transaction["QuartersAvailable"] == 4
].copy()

annual_complete["Rank"] = (
    annual_complete.groupby("Year")["Transactions"]
    .rank(method="min", ascending=False)
)

annual_complete = annual_complete.sort_values(
    ["Year", "Rank"]
)

annual_complete.to_csv(
    os.path.join(
        output_folder,
        "district_transaction_annual_rankings_2018_2025.csv"
    ),
    index=False
)

# Best transaction district for each complete year
best_district_by_year = (
    annual_complete.sort_values(
        ["Year", "Transactions"],
        ascending=[True, False]
    )
    .groupby("Year", as_index=False)
    .head(1)
    .sort_values("Year")
)

best_district_by_year.to_csv(
    os.path.join(
        output_folder,
        "best_transaction_district_by_year.csv"
    ),
    index=False
)


# ==================================================
# PART B — MERCHANT AND USER DISTRICT RANKINGS
# ==================================================

registration = df[
    df["Category"].isin(["merchant", "user"])
].copy()

# Latest available snapshot
latest_registration = registration[
    (registration["Year"] == 2026)
    & (registration["QuarterNumber"] == 2)
].copy()

registration_ranking = (
    latest_registration.groupby(
        ["Category"] + district_keys,
        as_index=False
    )
    .agg(
        RegisteredCount=("RegisteredCount", "max")
    )
    .sort_values(
        ["Category", "RegisteredCount"],
        ascending=[True, False]
    )
)

registration_ranking["Rank"] = (
    registration_ranking.groupby("Category")["RegisteredCount"]
    .rank(method="min", ascending=False)
)

registration_ranking.to_csv(
    os.path.join(
        output_folder,
        "district_merchant_user_ranking_2026_Q2.csv"
    ),
    index=False
)


# ==================================================
# PART C — TOP 15 DISTRICT CHARTS
# ==================================================

def make_top_chart(data, label_col, value_col, title,
                   xlabel, filename, color="#38BDF8"):

    if data.empty:
        print("Chart skipped - no data:", title)
        return

    chart_data = data.head(15).copy()
    chart_data = chart_data.sort_values(value_col, ascending=True)

    labels = (
        chart_data["District"].astype(str)
        + " — "
        + chart_data["State"].astype(str)
    )

    fig, ax = plt.subplots(figsize=(12, 7))
    fig.patch.set_facecolor("#0B1120")
    ax.set_facecolor("#111827")

    ax.barh(
        labels,
        chart_data[value_col],
        color=color
    )

    ax.set_title(title, color="white", fontsize=14, pad=14)
    ax.set_xlabel(xlabel, color="#D1D5DB")
    ax.tick_params(axis="both", colors="#D1D5DB", labelsize=8)
    ax.grid(axis="x", linestyle="--", alpha=0.2)

    for spine in ax.spines.values():
        spine.set_color("#374151")

    plt.tight_layout()
    plt.savefig(
        os.path.join(chart_folder, filename),
        dpi=180,
        bbox_inches="tight",
        facecolor=fig.get_facecolor()
    )
    plt.close(fig)


# Chart 1: Top transaction count districts
make_top_chart(
    latest_transaction_ranking,
    "District",
    "Transactions",
    "Top 15 Districts by Transaction Count — 2026 Q2",
    "Transactions",
    "top15_district_transaction_count_2026_Q2.png",
    "#38BDF8"
)

# Chart 2: Top transaction amount districts
make_top_chart(
    latest_transaction_ranking.sort_values(
        "Amount", ascending=False
    ),
    "District",
    "Amount",
    "Top 15 Districts by Transaction Amount — 2026 Q2",
    "Amount from Map source",
    "top15_district_transaction_amount_2026_Q2.png",
    "#F59E0B"
)

# Chart 3: Top merchant districts
merchant_rank = registration_ranking[
    registration_ranking["Category"] == "merchant"
].sort_values("RegisteredCount", ascending=False)

make_top_chart(
    merchant_rank,
    "District",
    "RegisteredCount",
    "Top 15 Districts by Registered Merchants — 2026 Q2",
    "Registered merchants",
    "top15_district_merchants_2026_Q2.png",
    "#34D399"
)

# Chart 4: Top user districts
user_rank = registration_ranking[
    registration_ranking["Category"] == "user"
].sort_values("RegisteredCount", ascending=False)

make_top_chart(
    user_rank,
    "District",
    "RegisteredCount",
    "Top 15 Districts by Registered Users — 2026 Q2",
    "Registered users",
    "top15_district_users_2026_Q2.png",
    "#C084FC"
)


# ==================================================
# PART D — FINAL SUMMARY
# ==================================================

print("\n========== MAP DISTRICT ANALYSIS COMPLETE ==========")
print("District master rows:", len(df))
print("Unique district names:", df["District"].nunique())
print("States / UTs:", df["StateFolder"].nunique())
print("2026 Q2 transaction district rows:", len(latest_transaction_ranking))
print("Annual complete district records:", len(annual_complete))
print("Charts saved:", len([
    f for f in os.listdir(chart_folder)
    if f.lower().endswith(".png")
]))

print("\nTop 10 districts by transaction count — 2026 Q2:")
display(
    latest_transaction_ranking[
        ["Rank", "District", "State", "Transactions"]
    ].head(10)
)

print("\nTop 10 districts by registered merchants — 2026 Q2:")
display(
    merchant_rank[
        ["Rank", "District", "State", "RegisteredCount"]
    ].head(10)
)

print("\nTop 10 districts by registered users — 2026 Q2:")
display(
    user_rank[
        ["Rank", "District", "State", "RegisteredCount"]
    ].head(10)
)

print("\nBest transaction district by complete year:")
display(
    best_district_by_year[
        ["Year", "District", "State", "Transactions"]
    ]
)

print("\nOutput folder:")
print(output_folder)
```

```python
import os
import json
import re
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ======================================================
# 1. PATHS AND SETTINGS
# ======================================================

top_path = (
    r"C:\Users\HP\OneDrive\Documents\fita project"
    r"\pulse-main\pulse-main\data\top"
)

output_folder = os.path.join(
    top_path, "top_analysis_2018_2026"
)

chart_folder = os.path.join(output_folder, "charts")
comparison_folder = os.path.join(
    output_folder, "quarter_comparisons"
)

os.makedirs(chart_folder, exist_ok=True)
os.makedirs(comparison_folder, exist_ok=True)

categories = ["transaction", "merchant", "user"]

quarters_by_year = {
    year: ([1, 2] if year == 2026 else [1, 2, 3, 4])
    for year in range(2018, 2027)
}

state_name_mapping = {
    "andaman-&-nicobar-islands": "Andaman & Nicobar Islands",
    "andhra-pradesh": "Andhra Pradesh",
    "arunachal-pradesh": "Arunachal Pradesh",
    "dadra-&-nagar-haveli-&-daman-&-diu":
        "Dadra and Nagar Haveli and Daman and Diu",
    "jammu-&-kashmir": "Jammu & Kashmir",
    "madhya-pradesh": "Madhya Pradesh",
    "tamil-nadu": "Tamil Nadu",
    "uttar-pradesh": "Uttar Pradesh",
    "west-bengal": "West Bengal"
}


def readable_state_name(folder_name):
    if folder_name in state_name_mapping:
        return state_name_mapping[folder_name]

    return folder_name.replace("-", " ").title()


def safe_filename(text):
    return re.sub(r"[^A-Za-z0-9_-]+", "_", str(text)).strip("_")


print("Top folder exists:", os.path.exists(top_path))
print("Output folder:", output_folder)


# ======================================================
# 2. JSON READING FUNCTION
# ======================================================

records = []
missing_files = []
read_errors = []


def add_items(
    items,
    category,
    level,
    scope,
    state_name,
    state_folder,
    year,
    quarter,
    source_file
):
    """JSON list-la irukkura entity records-ai collect pannum."""

    if not isinstance(items, list):
        return

    for item in items:

        if not isinstance(item, dict):
            continue

        if category == "transaction":

            entity_name = item.get("entityName")

            metric = item.get("metric") or {}

            # TOTAL metric-ai mattum use pannuvom
            if (
                metric.get("type") is not None
                and str(metric.get("type")).upper() != "TOTAL"
            ):
                continue

            metric_value = metric.get("count")
            metric_name = "Transactions"

        else:
            entity_name = item.get("name")
            metric_value = item.get("registeredCount")
            metric_name = "RegisteredCount"

        if entity_name is None or metric_value is None:
            continue

        try:
            metric_value = float(metric_value)
        except (TypeError, ValueError):
            continue

        records.append({
            "Category": category.title(),
            "Scope": scope,
            "Level": level,
            "State": state_name,
            "StateFolder": state_folder,
            "EntityName": str(entity_name).strip(),
            "Year": year,
            "Quarter": f"Q{quarter}",
            "QuarterNumber": quarter,
            "MetricName": metric_name,
            "MetricValue": metric_value,
            "SourceFile": source_file
        })


# ======================================================
# 3. READ COUNTRY-LEVEL AND STATE-LEVEL JSON FILES
# ======================================================

for category in categories:

    print(f"\nReading {category.title()} data...")

    country_path = os.path.join(
        top_path, category, "country", "india"
    )

    state_root = os.path.join(country_path, "state")

    # Country-level JSON files
    for year, quarter_list in quarters_by_year.items():

        for quarter in quarter_list:

            json_path = os.path.join(
                country_path, str(year), f"{quarter}.json"
            )

            if not os.path.isfile(json_path):
                missing_files.append({
                    "Category": category.title(),
                    "Scope": "Country",
                    "StateFolder": "india",
                    "Year": year,
                    "Quarter": f"Q{quarter}",
                    "File": json_path
                })
                continue

            try:
                with open(json_path, "r", encoding="utf-8") as file:
                    content = json.load(file)

                data = content.get("data") or {}

                add_items(
                    data.get("states"),
                    category,
                    "State",
                    "Country",
                    "India",
                    "india",
                    year,
                    quarter,
                    json_path
                )

                add_items(
                    data.get("districts"),
                    category,
                    "District",
                    "Country",
                    "India",
                    "india",
                    year,
                    quarter,
                    json_path
                )

            except Exception as error:
                read_errors.append({
                    "Category": category.title(),
                    "Scope": "Country",
                    "Year": year,
                    "Quarter": f"Q{quarter}",
                    "File": json_path,
                    "Error": str(error)
                })

    # State-level JSON files
    if os.path.isdir(state_root):

        state_folders = [
            folder for folder in os.listdir(state_root)
            if os.path.isdir(os.path.join(state_root, folder))
        ]

        for state_folder in sorted(state_folders):

            state_name = readable_state_name(state_folder)
            state_path = os.path.join(state_root, state_folder)

            for year, quarter_list in quarters_by_year.items():

                for quarter in quarter_list:

                    json_path = os.path.join(
                        state_path, str(year), f"{quarter}.json"
                    )

                    if not os.path.isfile(json_path):
                        missing_files.append({
                            "Category": category.title(),
                            "Scope": "State",
                            "StateFolder": state_folder,
                            "Year": year,
                            "Quarter": f"Q{quarter}",
                            "File": json_path
                        })
                        continue

                    try:
                        with open(
                            json_path, "r", encoding="utf-8"
                        ) as file:
                            content = json.load(file)

                        data = content.get("data") or {}

                        add_items(
                            data.get("districts"),
                            category,
                            "District",
                            "State",
                            state_name,
                            state_folder,
                            year,
                            quarter,
                            json_path
                        )

                    except Exception as error:
                        read_errors.append({
                            "Category": category.title(),
                            "Scope": "State",
                            "StateFolder": state_folder,
                            "Year": year,
                            "Quarter": f"Q{quarter}",
                            "File": json_path,
                            "Error": str(error)
                        })


# ======================================================
# 4. CREATE MASTER CSV
# ======================================================

columns = [
    "Category", "Scope", "Level", "State", "StateFolder",
    "EntityName", "Year", "Quarter", "QuarterNumber",
    "MetricName", "MetricValue", "SourceFile"
]

master = pd.DataFrame(records, columns=columns)

if master.empty:
    raise ValueError(
        "Top data read aagala. JSON structure-ai check pannanum."
    )

master = master.sort_values(
    ["Category", "Scope", "Level", "Year",
     "QuarterNumber", "MetricValue"],
    ascending=[True, True, True, True, True, False]
).reset_index(drop=True)

master_file = os.path.join(
    output_folder, "top_master_all_2018_2026.csv"
)

master.to_csv(master_file, index=False)


# ======================================================
# 5. GENERATE RANKINGS
# ======================================================

ranked = master.copy()

rank_keys = ["Category", "Scope", "Level", "Year", "Quarter"]

ranked["Rank"] = (
    ranked.groupby(rank_keys)["MetricValue"]
    .rank(method="min", ascending=False)
    .astype("Int64")
)

ranked = ranked.sort_values(
    rank_keys + ["Rank", "EntityName"]
)

rank_file = os.path.join(
    output_folder, "top_rankings_all_2018_2026.csv"
)

ranked.to_csv(rank_file, index=False)

# Separate useful ranking CSVs
ranking_files = {
    "top_state_rankings_2018_2026.csv":
        ranked[
            (ranked["Scope"] == "Country")
            & (ranked["Level"] == "State")
        ],

    "top_india_district_rankings_2018_2026.csv":
        ranked[
            (ranked["Scope"] == "Country")
            & (ranked["Level"] == "District")
        ],

    "top_state_district_rankings_2018_2026.csv":
        ranked[
            (ranked["Scope"] == "State")
            & (ranked["Level"] == "District")
        ]
}

for filename, data in ranking_files.items():
    data.to_csv(
        os.path.join(output_folder, filename),
        index=False
    )


# ======================================================
# 6. QUARTER COMPARISONS
# ======================================================

comparison_groups = [
    (
        "country_state",
        ranked[
            (ranked["Scope"] == "Country")
            & (ranked["Level"] == "State")
        ],
        ["Category", "EntityName", "Year"]
    ),
    (
        "country_district",
        ranked[
            (ranked["Scope"] == "Country")
            & (ranked["Level"] == "District")
        ],
        ["Category", "EntityName", "Year"]
    ),
    (
        "state_district",
        ranked[
            (ranked["Scope"] == "State")
            & (ranked["Level"] == "District")
        ],
        ["Category", "State", "StateFolder", "EntityName", "Year"]
    )
]

for file_prefix, data, index_cols in comparison_groups:

    if data.empty:
        print("No data for comparison:", file_prefix)
        continue

    pivot = data.pivot_table(
        index=index_cols,
        columns="Quarter",
        values="MetricValue",
        aggfunc="first"
    ).reset_index()

    for quarter in ["Q1", "Q2", "Q3", "Q4"]:
        if quarter not in pivot.columns:
            pivot[quarter] = np.nan

    pivot["Q2_minus_Q1"] = pivot["Q2"] - pivot["Q1"]
    pivot["Q3_minus_Q2"] = pivot["Q3"] - pivot["Q2"]
    pivot["Q4_minus_Q3"] = pivot["Q4"] - pivot["Q3"]

    pivot.to_csv(
        os.path.join(
            comparison_folder,
            f"{file_prefix}_quarter_comparison.csv"
        ),
        index=False
    )


# ======================================================
# 7. PROFESSIONAL DARK-THEME CHARTS
# Latest available quarter: 2026 Q2
# ======================================================

def make_top_chart(data, category, level, metric_name):

    if data.empty:
        print("No chart data:", category, level)
        return

    chart_data = data.nlargest(10, "MetricValue").copy()
    chart_data = chart_data.sort_values("MetricValue")

    fig, ax = plt.subplots(figsize=(12, 7))

    fig.patch.set_facecolor("#0B1120")
    ax.set_facecolor("#111827")

    ax.barh(
        chart_data["EntityName"].astype(str),
        chart_data["MetricValue"],
        color="#38BDF8"
    )

    ax.set_title(
        f"Top 10 {level}s — {category} — 2026 Q2",
        color="white",
        fontsize=15,
        fontweight="bold",
        pad=15
    )

    ax.set_xlabel(metric_name, color="#D1D5DB")
    ax.tick_params(axis="both", colors="#D1D5DB", labelsize=9)
    ax.grid(axis="x", linestyle="--", alpha=0.22)

    for spine in ax.spines.values():
        spine.set_color("#374151")

    plt.tight_layout()

    filename = safe_filename(
        f"top10_{category}_{level}_2026_Q2.png"
    )

    plt.savefig(
        os.path.join(chart_folder, filename),
        dpi=180,
        facecolor=fig.get_facecolor(),
        bbox_inches="tight"
    )

    plt.close(fig)


latest = ranked[
    (ranked["Year"] == 2026)
    & (ranked["QuarterNumber"] == 2)
    & (ranked["Scope"] == "Country")
].copy()

for category in ["Transaction", "Merchant", "User"]:

    category_data = latest[
        latest["Category"] == category
    ]

    state_data = category_data[
        category_data["Level"] == "State"
    ]

    district_data = category_data[
        category_data["Level"] == "District"
    ]

    if not state_data.empty:
        make_top_chart(
            state_data,
            category,
            "State",
            state_data["MetricName"].iloc[0]
        )

    if not district_data.empty:
        make_top_chart(
            district_data,
            category,
            "District",
            district_data["MetricName"].iloc[0]
        )


# ======================================================
# 8. SAVE MISSING-FILE AND ERROR REPORTS
# ======================================================

pd.DataFrame(missing_files).to_csv(
    os.path.join(output_folder, "top_missing_json_files.csv"),
    index=False
)

pd.DataFrame(read_errors).to_csv(
    os.path.join(output_folder, "top_read_errors.csv"),
    index=False
)


# ======================================================
# 9. FINAL SUMMARY
# ======================================================

print("\n" + "=" * 52)
print("       TOP DATASET PROCESSING COMPLETE")
print("=" * 52)

print("Master rows:", len(master))
print("Categories:", sorted(master["Category"].unique()))
print("Unique entity names:", master["EntityName"].nunique())
print("Years:", sorted(master["Year"].unique()))
print("Missing JSON files:", len(missing_files))
print("Read errors:", len(read_errors))
print("Ranking records:", len(ranked))
print("Charts created:", len([
    f for f in os.listdir(chart_folder)
    if f.lower().endswith(".png")
]))

print("\nRecords by Category / Scope / Level:")

summary = (
    master.groupby(
        ["Category", "Scope", "Level"], as_index=False
    )
    .agg(
        Rows=("EntityName", "size"),
        UniqueEntities=("EntityName", "nunique")
    )
)

display(summary)

print("\nLatest country-level state rankings — 2026 Q2:")

latest_state_ranks = latest[latest["Level"] == "State"]

display(
    latest_state_ranks[
        ["Category", "Rank", "EntityName", "MetricValue"]
    ]
    .sort_values(["Category", "Rank"])
    .groupby("Category")
    .head(10)
)

print("\nOutput folder:", output_folder)
print("Master CSV:", master_file)
print("Ranking CSV:", rank_file)
print("Charts folder:", chart_folder)
```

```python
import os
import pandas as pd

data_root = (
    r"C:\Users\HP\OneDrive\Documents\fita project"
    r"\pulse-main\pulse-main\data"
)

files_to_check = {
    "Aggregated Transaction Master": os.path.join(
        data_root, "aggregated", "transaction", "country", "india",
        "transaction_analysis_2018_2026",
        "phonepe_india_master_2018_2026.csv"
    ),

    "Aggregated Merchant India Master": os.path.join(
        data_root, "aggregated", "merchant", "country", "india",
        "merchant_analysis_2018_2026",
        "merchant_india_master_2018_2026.csv"
    ),

    "Aggregated Merchant State Master": os.path.join(
        data_root, "aggregated", "merchant", "country", "india",
        "merchant_state_analysis_2018_2026",
        "merchant_state_master_2018_2026.csv"
    ),

    "Aggregated User State Master": os.path.join(
        data_root, "aggregated", "user", "country", "india",
        "user_state_analysis_2018_2026",
        "user_state_master_2018_2026.csv"
    ),

    "Map State Master": os.path.join(
        data_root, "map", "map_analysis_2018_2026",
        "map_state_master_all_categories_2018_2026.csv"
    ),

    "Map District Master": os.path.join(
        data_root, "map", "map_analysis_2018_2026",
        "map_district_master_all_categories_2018_2026.csv"
    ),

    "Map Transaction Validation": os.path.join(
        data_root, "aggregated", "transaction", "country", "india",
        "transaction_analysis_2018_2026",
        "aggregated_vs_map_transaction_validation.csv"
    ),

    "Top Master": os.path.join(
        data_root, "top", "top_analysis_2018_2026",
        "top_master_all_2018_2026.csv"
    ),

    "Top Rankings": os.path.join(
        data_root, "top", "top_analysis_2018_2026",
        "top_rankings_all_2018_2026.csv"
    ),

    "Top Missing Files Report": os.path.join(
        data_root, "top", "top_analysis_2018_2026",
        "top_missing_json_files.csv"
    ),

    "Top Read Errors Report": os.path.join(
        data_root, "top", "top_analysis_2018_2026",
        "top_read_errors.csv"
    )
}

results = []

for name, path in files_to_check.items():

    if not os.path.isfile(path):
        results.append({
            "Dataset File": name,
            "Status": "NOT FOUND",
            "Rows": None,
            "Path": path
        })
        continue

    try:
        temp = pd.read_csv(path)

        results.append({
            "Dataset File": name,
            "Status": "FOUND",
            "Rows": len(temp),
            "Path": path
        })

    except pd.errors.EmptyDataError:
        # Empty read-error report can mean there were no errors.
        if name == "Top Read Errors Report":
            results.append({
                "Dataset File": name,
                "Status": "EMPTY - check reported 0 read errors",
                "Rows": 0,
                "Path": path
            })
        else:
            results.append({
                "Dataset File": name,
                "Status": "EMPTY CSV",
                "Rows": 0,
                "Path": path
            })

    except Exception as error:
        results.append({
            "Dataset File": name,
            "Status": f"READ ERROR: {error}",
            "Rows": None,
            "Path": path
        })

audit_df = pd.DataFrame(results)

print("========== PHONEPE PROJECT FINAL AUDIT ==========")
display(audit_df[["Dataset File", "Status", "Rows"]])

print("\nFiles found:", (audit_df["Status"] == "FOUND").sum())
print("Files not found:", (audit_df["Status"] == "NOT FOUND").sum())

audit_path = os.path.join(
    data_root, "phonepe_project_final_audit.csv"
)

audit_df.to_csv(audit_path, index=False)

print("\nFinal audit report saved:")
print(audit_path)
```

```python
import os
import re
import pandas as pd
import matplotlib.pyplot as plt

# 1. Top dataset paths
output_folder = (
    r"C:\Users\HP\OneDrive\Documents\fita project"
    r"\pulse-main\pulse-main\data\top\top_analysis_2018_2026"
)

master_file = os.path.join(
    output_folder, "top_master_all_2018_2026.csv"
)

year_chart_folder = os.path.join(
    output_folder, "year_wise_charts"
)

state_chart_folder = os.path.join(
    output_folder, "state_wise_district_charts_2026_Q2"
)

os.makedirs(year_chart_folder, exist_ok=True)
os.makedirs(state_chart_folder, exist_ok=True)

df = pd.read_csv(master_file)

# 2. Prepare columns
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df["QuarterNumber"] = pd.to_numeric(
    df["QuarterNumber"], errors="coerce"
)
df["MetricValue"] = pd.to_numeric(
    df["MetricValue"], errors="coerce"
)

df["Category"] = df["Category"].astype(str).str.title()
df["Scope"] = df["Scope"].astype(str).str.title()
df["Level"] = df["Level"].astype(str).str.title()

# Year-end snapshot: 2018-2025 Q4, 2026 Q2
snapshots = {
    year: 4 for year in range(2018, 2026)
}
snapshots[2026] = 2

categories = ["Transaction", "Merchant", "User"]


def safe_name(value):
    return re.sub(r"[^A-Za-z0-9_-]+", "_", str(value)).strip("_")


def save_top_chart(data, title, filename, value_label):
    """Top 10 records-ku dark-theme horizontal bar chart save pannum."""

    if data.empty:
        return False

    data = (
        data.sort_values("MetricValue", ascending=False)
        .drop_duplicates(subset=["EntityName"])
        .head(10)
        .sort_values("MetricValue", ascending=True)
    )

    labels = data["EntityName"].astype(str)

    fig, ax = plt.subplots(figsize=(12, 6))
    fig.patch.set_facecolor("#0B1120")
    ax.set_facecolor("#111827")

    ax.barh(labels, data["MetricValue"], color="#38BDF8")

    ax.set_title(
        title, color="white", fontsize=14,
        fontweight="bold", pad=14
    )
    ax.set_xlabel(value_label, color="#D1D5DB")
    ax.tick_params(axis="both", colors="#D1D5DB", labelsize=9)
    ax.grid(axis="x", linestyle="--", alpha=0.2)

    for spine in ax.spines.values():
        spine.set_color("#374151")

    plt.tight_layout()
    plt.savefig(
        filename,
        dpi=160,
        bbox_inches="tight",
        facecolor=fig.get_facecolor()
    )
    plt.close(fig)

    return True


# ==================================================
# PART A — YEAR-WISE TOP STATE AND INDIA DISTRICT CHARTS
# ==================================================

year_chart_count = 0

for year, quarter in snapshots.items():

    for category in categories:

        snapshot = df[
            (df["Year"] == year)
            & (df["QuarterNumber"] == quarter)
            & (df["Category"] == category)
            & (df["Scope"] == "Country")
        ]

        # Top states for this year's selected quarter
        states = snapshot[snapshot["Level"] == "State"]

        filename = os.path.join(
            year_chart_folder,
            f"{year}_Q{quarter}_{category}_top_states.png"
        )

        if save_top_chart(
            states,
            f"Top States — {category} — {year} Q{quarter}",
            filename,
            states["MetricName"].iloc[0]
            if not states.empty else "Metric value"
        ):
            year_chart_count += 1

        # Top districts across India
        districts = snapshot[snapshot["Level"] == "District"]

        filename = os.path.join(
            year_chart_folder,
            f"{year}_Q{quarter}_{category}_top_india_districts.png"
        )

        if save_top_chart(
            districts,
            f"Top Districts in India — {category} — {year} Q{quarter}",
            filename,
            districts["MetricName"].iloc[0]
            if not districts.empty else "Metric value"
        ):
            year_chart_count += 1


# ==================================================
# PART B — STATE-WISE DISTRICT CHARTS
# Latest available snapshot: 2026 Q2
# ==================================================

latest_state_data = df[
    (df["Year"] == 2026)
    & (df["QuarterNumber"] == 2)
    & (df["Scope"] == "State")
    & (df["Level"] == "District")
].copy()

state_chart_count = 0

for (state_folder, category), group in latest_state_data.groupby(
    ["StateFolder", "Category"], dropna=True
):

    state_name = group["State"].dropna()
    state_title = (
        state_name.iloc[0]
        if not state_name.empty
        else str(state_folder)
    )

    filename = os.path.join(
        state_chart_folder,
        f"{safe_name(state_folder)}_{category}_top_districts.png"
    )

    if save_top_chart(
        group,
        f"{state_title} — Top Districts by {category} — 2026 Q2",
        filename,
        group["MetricName"].iloc[0]
    ):
        state_chart_count += 1


# ==================================================
# PART C — REPORT
# ==================================================

print("\n========== TOP YEAR & STATE CHARTS COMPLETE ==========")
print("Year-wise state/district charts:", year_chart_count)
print("State-wise district charts:", state_chart_count)

print("\nYear-wise chart folder:")
print(year_chart_folder)

print("\nState-wise chart folder:")
print(state_chart_folder)

print("\nAvailable state-level districts in 2026 Q2:")
print(latest_state_data["StateFolder"].nunique())

print("\nCharts by category:")
display(
    latest_state_data.groupby(
        ["Category"], as_index=False
    ).agg(
        Records=("EntityName", "size"),
        States=("StateFolder", "nunique")
    )
)
```

```python
import os
import pandas as pd

# Main PhonePe project data folder
data_root = (
    r"C:\Users\HP\OneDrive\Documents\fita project"
    r"\pulse-main\pulse-main\data"
)

# 1. Main folders
folders = {
    "Aggregated": os.path.join(data_root, "aggregated"),
    "Map": os.path.join(data_root, "map"),
    "Top": os.path.join(data_root, "top")
}

print("=" * 55)
print("       PHONEPE COMPLETE DATA CHECK")
print("=" * 55)

# 2. Each folder-la evlo JSON, CSV, PNG files irukku?
for name, folder in folders.items():

    print(f"\n===== {name.upper()} DATA =====")

    if not os.path.isdir(folder):
        print("Folder: NOT FOUND")
        continue

    json_count = 0
    csv_count = 0
    png_count = 0

    for root, dirs, files in os.walk(folder):
        for filename in files:
            ext = os.path.splitext(filename)[1].lower()

            if ext == ".json":
                json_count += 1
            elif ext == ".csv":
                csv_count += 1
            elif ext == ".png":
                png_count += 1

    print("Folder exists: YES")
    print("JSON files:", json_count)
    print("CSV files:", csv_count)
    print("Chart images:", png_count)


# 3. Important master CSV files check
important_files = {
    "Aggregated Transaction Master": os.path.join(
        data_root, "aggregated", "transaction", "country", "india",
        "transaction_analysis_2018_2026",
        "phonepe_india_master_2018_2026.csv"
    ),

    "Aggregated Merchant India Master": os.path.join(
        data_root, "aggregated", "merchant", "country", "india",
        "merchant_analysis_2018_2026",
        "merchant_india_master_2018_2026.csv"
    ),

    "Aggregated Merchant State Master": os.path.join(
        data_root, "aggregated", "merchant", "country", "india",
        "merchant_state_analysis_2018_2026",
        "merchant_state_master_2018_2026.csv"
    ),

    "Aggregated User State Master": os.path.join(
        data_root, "aggregated", "user", "country", "india",
        "user_state_analysis_2018_2026",
        "user_state_master_2018_2026.csv"
    ),

    "Map State Master": os.path.join(
        data_root, "map", "map_analysis_2018_2026",
        "map_state_master_all_categories_2018_2026.csv"
    ),

    "Map District Master": os.path.join(
        data_root, "map", "map_analysis_2018_2026",
        "map_district_master_all_categories_2018_2026.csv"
    ),

    "Map Transaction Validation": os.path.join(
        data_root, "aggregated", "transaction", "country", "india",
        "transaction_analysis_2018_2026",
        "aggregated_vs_map_transaction_validation.csv"
    ),

    "Top Master": os.path.join(
        data_root, "top", "top_analysis_2018_2026",
        "top_master_all_2018_2026.csv"
    ),

    "Top Rankings": os.path.join(
        data_root, "top", "top_analysis_2018_2026",
        "top_rankings_all_2018_2026.csv"
    ),

    "Top Missing Files Report": os.path.join(
        data_root, "top", "top_analysis_2018_2026",
        "top_missing_json_files.csv"
    )
}

results = []

print("\n" + "=" * 55)
print("       IMPORTANT MASTER FILE CHECK")
print("=" * 55)

for name, path in important_files.items():

    if not os.path.isfile(path):
        results.append({
            "File": name,
            "Status": "NOT FOUND",
            "Rows": None
        })
        print(f"{name}: NOT FOUND")
        continue

    try:
        temp = pd.read_csv(path)

        results.append({
            "File": name,
            "Status": "FOUND",
            "Rows": len(temp)
        })

        print(f"{name}: FOUND | Rows: {len(temp):,}")

    except pd.errors.EmptyDataError:
        results.append({
            "File": name,
            "Status": "EMPTY CSV",
            "Rows": 0
        })
        print(f"{name}: EMPTY CSV")

    except Exception as error:
        results.append({
            "File": name,
            "Status": "READ ERROR",
            "Rows": None
        })
        print(f"{name}: READ ERROR - {error}")


# 4. Aggregated User India master kandupidikka extra search
user_root = os.path.join(
    data_root, "aggregated", "user", "country", "india"
)

print("\n===== AGGREGATED USER INDIA FILE SEARCH =====")

found_user_files = []

if os.path.isdir(user_root):
    for root, dirs, files in os.walk(user_root):
        # Generated analysis output folders mattum search result-la kaattuvom
        for filename in files:
            lower_name = filename.lower()

            if (
                lower_name.endswith(".csv")
                and "master" in lower_name
                and "user" in lower_name
                and "state" not in lower_name
            ):
                found_user_files.append(os.path.join(root, filename))

if found_user_files:
    for path in found_user_files:
        print("FOUND:", path)
else:
    print("User India master CSV not located by filename; inspect folder if needed.")


# 5. Chart folders check
chart_folders = {
    "Map State Charts": os.path.join(
        data_root, "map", "map_analysis_2018_2026",
        "state_4_panel_charts"
    ),

    "Map Best-Year Charts": os.path.join(
        data_root, "map", "map_analysis_2018_2026",
        "best_year_best_state", "charts"
    ),

    "Map District Charts": os.path.join(
        data_root, "map", "map_analysis_2018_2026",
        "district_analysis_2018_2026", "charts"
    ),

    "Top Original Charts": os.path.join(
        data_root, "top", "top_analysis_2018_2026", "charts"
    ),

    "Top Year-Wise Charts": os.path.join(
        data_root, "top", "top_analysis_2018_2026",
        "year_wise_charts"
    ),

    "Top State-Wise District Charts": os.path.join(
        data_root, "top", "top_analysis_2018_2026",
        "state_wise_district_charts_2026_Q2"
    )
}

print("\n" + "=" * 55)
print("       CHART FOLDER CHECK")
print("=" * 55)

for name, folder in chart_folders.items():

    if os.path.isdir(folder):
        count = sum(
            1 for filename in os.listdir(folder)
            if filename.lower().endswith(".png")
        )
        print(f"{name}: {count} PNG files")
    else:
        print(f"{name}: Folder not found")


# 6. Save audit report
audit_df = pd.DataFrame(results)

audit_file = os.path.join(
    data_root, "phonepe_complete_data_check.csv"
)

audit_df.to_csv(audit_file, index=False)

print("\n" + "=" * 55)
print("DATA CHECK FINISHED")
print("Audit report saved:", audit_file)
print("=" * 55)
```

```python
%%writefile app.py
from pathlib import Path

import pandas as pd
import plotly.express as px
import streamlit as st

st.set_page_config(
    page_title="PhonePe Pulse Analytics",
    page_icon="📱",
    layout="wide"
)

# Project and data folders
ROOT = Path.cwd().parent
DATA = ROOT / "data"

FILES = {
    "Aggregated Transaction": DATA / "aggregated/transaction/country/india/transaction_analysis_2018_2026/phonepe_india_master_2018_2026.csv",
    "Aggregated Merchant": DATA / "aggregated/merchant/country/india/merchant_analysis_2018_2026/merchant_india_master_2018_2026.csv",
    "Aggregated Merchant State": DATA / "aggregated/merchant/country/india/merchant_state_analysis_2018_2026/merchant_state_master_2018_2026.csv",
    "Aggregated User India": DATA / "aggregated/user/country/india/user_india_analysis_2018_2026/user_india_master_2018_2026.csv",
    "Aggregated User State": DATA / "aggregated/user/country/india/user_state_analysis_2018_2026/user_state_master_2018_2026.csv",
    "Map State": DATA / "map/map_analysis_2018_2026/map_state_master_all_categories_2018_2026.csv",
    "Map District": DATA / "map/map_analysis_2018_2026/map_district_master_all_categories_2018_2026.csv",
    "Top Master": DATA / "top/top_analysis_2018_2026/top_master_all_2018_2026.csv",
    "Validation": DATA / "aggregated/transaction/country/india/transaction_analysis_2018_2026/aggregated_vs_map_transaction_validation.csv",
    "Top Missing Files": DATA / "top/top_analysis_2018_2026/top_missing_json_files.csv"
}

@st.cache_data
def load_csv(path):
    try:
        return pd.read_csv(path, low_memory=False)
    except Exception:
        return pd.DataFrame()

def prepare(df):
    df = df.copy()

    if "Category" in df.columns:
        df["Category"] = df["Category"].astype(str).str.title()

    for col in [
        "Year", "QuarterNumber", "Transactions",
        "Amount", "RegisteredCount", "MetricValue"
    ]:
        if col in df.columns:
            df[col] = pd.to_numeric(df[col], errors="coerce")

    return df

def chart_style(fig):
    fig.update_layout(
        template="plotly_dark",
        paper_bgcolor="#0B1120",
        plot_bgcolor="#111827",
        font_color="#F1F5F9"
    )
    return fig

def show_trend(df, metric, title, split_category=False):
    df = df.copy()

    if df.empty or metric not in df.columns or "Year" not in df.columns:
        st.info("Trend data available illa.")
        return

    group_cols = ["Year"]

    if "QuarterNumber" in df.columns:
        group_cols.append("QuarterNumber")

    if split_category and "Category" in df.columns:
        group_cols.append("Category")

    trend = (
        df.groupby(group_cols, as_index=False)
        .agg(Value=(metric, "sum"))
        .sort_values(group_cols)
    )

    if "QuarterNumber" in trend.columns:
        trend["Period"] = (
            trend["Year"].astype(int).astype(str)
            + " Q"
            + trend["QuarterNumber"].astype(int).astype(str)
        )
    else:
        trend["Period"] = trend["Year"].astype(int).astype(str)

    fig = px.line(
        trend,
        x="Period",
        y="Value",
        color="Category" if split_category and "Category" in trend.columns else None,
        markers=True,
        title=title
    )

    st.plotly_chart(chart_style(fig), use_container_width=True)

# Load prepared datasets
data = {name: prepare(load_csv(path)) for name, path in FILES.items()}

# Professional dark theme
st.markdown("""
<style>
[data-testid="stAppViewContainer"] {background:#0B1120;}
[data-testid="stSidebar"] {background:#111827;}
[data-testid="stMetric"] {
    background:#111827;
    border:1px solid #334155;
    padding:14px;
    border-radius:12px;
}
</style>
""", unsafe_allow_html=True)

st.sidebar.title("📱 PhonePe Pulse")
st.sidebar.caption("Interactive Analytics Dashboard")

page = st.sidebar.radio(
    "Navigation",
    ["Overview", "Aggregated", "Map Analysis", "Top Rankings", "Data Quality"]
)

st.sidebar.divider()
st.sidebar.caption("Data range: 2018–2026")
st.sidebar.caption("2026 has Q1–Q2 data only.")

# --------------------------------------------------
# OVERVIEW
# --------------------------------------------------
if page == "Overview":
    st.title("PhonePe Pulse Analytics")
    st.write("Explore transaction trends, registration snapshots and rankings.")

    df = data["Map State"]

    if df.empty:
        st.error("Map State master CSV load aagala.")
    else:
        tx = df[df["Category"] == "Transaction"].copy()

        if tx.empty:
            st.warning("Transaction data kidaikkala.")
        else:
            latest_year = int(tx["Year"].max())
            latest_q = int(
                tx.loc[tx["Year"] == latest_year, "QuarterNumber"].max()
            )

            latest = df[
                (df["Year"] == latest_year)
                & (df["QuarterNumber"] == latest_q)
            ]

            latest_tx = latest[latest["Category"] == "Transaction"]
            latest_merchants = latest[latest["Category"] == "Merchant"]
            latest_users = latest[latest["Category"] == "User"]

            c1, c2, c3, c4 = st.columns(4)

            c1.metric(
                f"Transactions · {latest_year} Q{latest_q}",
                f"{latest_tx['Transactions'].sum():,.0f}"
            )

            amount = latest_tx["Amount"]
            c2.metric(
                "Transaction Amount",
                f"{amount.sum():,.0f}" if amount.notna().any() else "Unavailable"
            )

            c3.metric(
                "Registered Merchants",
                f"{latest_merchants['RegisteredCount'].sum():,.0f}"
            )

            c4.metric(
                "Registered Users",
                f"{latest_users['RegisteredCount'].sum():,.0f}"
            )

            st.caption(
                "Merchant/User figures are registration snapshots. "
                "2026 data covers only the available quarters."
            )

            show_trend(tx, "Transactions", "Transaction Trend by Year")

            top_states = (
                latest_tx.groupby("State", as_index=False)
                .agg(Transactions=("Transactions", "sum"))
                .nlargest(10, "Transactions")
                .sort_values("Transactions")
            )

            fig = px.bar(
                top_states,
                x="Transactions",
                y="State",
                orientation="h",
                title=f"Top 10 States · {latest_year} Q{latest_q}"
            )

            st.plotly_chart(chart_style(fig), use_container_width=True)

# --------------------------------------------------
# AGGREGATED
# --------------------------------------------------
elif page == "Aggregated":
    st.title("Aggregated Analysis")

    choice = st.selectbox(
        "Dataset",
        [
            "Aggregated Transaction",
            "Aggregated Merchant",
            "Aggregated Merchant State",
            "Aggregated User India",
            "Aggregated User State"
        ]
    )

    df = data[choice]

    if df.empty:
        st.error(f"{choice} CSV not found or unreadable.")
    else:
        st.success(f"Loaded {len(df):,} rows.")

        if "State" in df.columns:
            states = ["All States"] + sorted(
                df["State"].dropna().astype(str).unique().tolist()
            )
            selected = st.selectbox("State", states)

            if selected != "All States":
                df = df[df["State"].astype(str) == selected]

        if "Category" in df.columns:
            categories = ["All Categories"] + sorted(
                df["Category"].dropna().astype(str).unique().tolist()
            )
            selected_category = st.selectbox("Category", categories)

            if selected_category != "All Categories":
                df = df[df["Category"].astype(str) == selected_category]

        metrics = [
            m for m in ["Transactions", "Amount", "RegisteredCount", "registeredCount"]
            if m in df.columns
        ]

        if metrics:
            metric = st.selectbox("Metric", metrics)
            df[metric] = pd.to_numeric(df[metric], errors="coerce")

            if df[metric].notna().any():
                show_trend(
                    df, metric, f"{metric} Trend",
                    split_category="Category" in df.columns
                )
            else:
                st.info(
                    f"{metric} source-la available illa. "
                    "Blank values-ai zero-nu consider panna vendam."
                )

        st.subheader("Data Preview")
        st.dataframe(df.head(500), use_container_width=True)

# --------------------------------------------------
# MAP ANALYSIS
# --------------------------------------------------
elif page == "Map Analysis":
    st.title("Map State & District Analysis")

    states = data["Map State"]
    districts = data["Map District"]

    if states.empty:
        st.error("Map State CSV load aagala.")
    else:
        category = st.selectbox(
            "Category", ["Transaction", "Merchant", "User"]
        )

        state_options = ["All States"] + sorted(
            states["State"].dropna().astype(str).unique().tolist()
        )
        chosen_state = st.selectbox("State", state_options)

        metric = (
            st.selectbox("Metric", ["Transactions", "Amount"])
            if category == "Transaction"
            else "RegisteredCount"
        )

        trend = states[states["Category"] == category].copy()

        if chosen_state != "All States":
            trend = trend[trend["State"] == chosen_state]

        years = sorted(trend["Year"].dropna().astype(int).unique().tolist())

        if years:
            chosen_year = st.selectbox(
                "Trend year", years, index=len(years) - 1
            )
            trend = trend[trend["Year"] == chosen_year]

        show_trend(trend, metric, f"{category} · {metric} Trend")

        st.subheader("District Rankings")

        if not districts.empty:
            d = districts[districts["Category"] == category].copy()

            if chosen_state != "All States":
                d = d[d["State"] == chosen_state]

            years_d = sorted(d["Year"].dropna().astype(int).unique().tolist())

            if years_d:
                year_d = st.selectbox(
                    "District year", years_d, index=len(years_d) - 1
                )
                d = d[d["Year"] == year_d]

                quarters = sorted(
                    d["QuarterNumber"].dropna().astype(int).unique().tolist()
                )

                if quarters:
                    quarter_d = st.selectbox(
                        "District quarter", quarters, index=len(quarters) - 1
                    )
                    d = d[d["QuarterNumber"] == quarter_d]

                    if metric in d.columns:
                        ranking = (
                            d.groupby(["State", "District"], as_index=False)
                            .agg(Value=(metric, "sum"))
                            .nlargest(15, "Value")
                            .sort_values("Value")
                        )

                        ranking["Label"] = (
                            ranking["District"].astype(str)
                            + " — " + ranking["State"].astype(str)
                        )

                        fig = px.bar(
                            ranking, x="Value", y="Label",
                            orientation="h", title="Top 15 Districts"
                        )
                        st.plotly_chart(
                            chart_style(fig), use_container_width=True
                        )
                        st.dataframe(ranking, use_container_width=True)

# --------------------------------------------------
# TOP RANKINGS
# --------------------------------------------------
elif page == "Top Rankings":
    st.title("Top States & Districts")

    df = data["Top Master"]

    if df.empty:
        st.error("Top master CSV load aagala.")
    else:
        category = st.selectbox(
            "Category", ["Transaction", "Merchant", "User"]
        )

        view = st.radio(
            "Ranking level",
            ["Top States", "India Districts", "Districts Within a State"],
            horizontal=True
        )

        if view == "Top States":
            scope, level = "Country", "State"
        elif view == "India Districts":
            scope, level = "Country", "District"
        else:
            scope, level = "State", "District"

        result = df[
            (df["Category"] == category)
            & (df["Scope"] == scope)
            & (df["Level"] == level)
        ].copy()

        if result.empty:
            st.warning("No records for this selection.")
        else:
            if view == "Districts Within a State":
                state_names = sorted(
                    result["State"].dropna().astype(str).unique().tolist()
                )
                if state_names:
                    selected_state = st.selectbox("Choose State", state_names)
                    result = result[result["State"] == selected_state]

            years = sorted(result["Year"].dropna().astype(int).unique().tolist())

            if years:
                year = st.selectbox("Year", years, index=len(years) - 1)
                result = result[result["Year"] == year]

                quarters = sorted(
                    result["QuarterNumber"].dropna().astype(int).unique().tolist()
                )

                if quarters:
                    quarter = st.selectbox(
                        "Quarter", quarters, index=len(quarters) - 1
                    )
                    result = result[result["QuarterNumber"] == quarter]

            ranking = (
                result.sort_values("MetricValue", ascending=False)
                .drop_duplicates(subset=["EntityName", "StateFolder"])
                .head(15)
                .copy()
            )

            if not ranking.empty:
                ranking["Rank"] = range(1, len(ranking) + 1)

                fig = px.bar(
                    ranking.sort_values("MetricValue"),
                    x="MetricValue",
                    y="EntityName",
                    orientation="h",
                    title=f"Top {category} Rankings"
                )
                st.plotly_chart(chart_style(fig), use_container_width=True)
                st.dataframe(ranking, use_container_width=True)

        st.caption(
            "Top dataset lists selected entities from the source, "
            "not a complete district census."
        )

# --------------------------------------------------
# DATA QUALITY
# --------------------------------------------------
elif page == "Data Quality":
    st.title("Data Quality & Validation")

    rows = []
    for name, path in FILES.items():
        rows.append({
            "Dataset": name,
            "File Exists": path.is_file(),
            "Rows": len(data[name])
        })

    st.dataframe(pd.DataFrame(rows), use_container_width=True)

    validation = data["Validation"]
    st.subheader("Aggregated vs Map Transaction Validation")

    if not validation.empty and "Status" in validation.columns:
        st.dataframe(
            validation["Status"].value_counts().rename("Records"),
            use_container_width=True
        )
        st.metric(
            "Exact matches",
            int((validation["Status"] == "Match").sum())
        )
        st.metric(
            "Mismatches",
            int((validation["Status"] == "Mismatch").sum())
        )
    else:
        st.warning("Validation report unavailable.")

    missing = data["Top Missing Files"]
    st.subheader("Top Dataset Missing Files")

    if not missing.empty:
        st.metric("Missing JSON files", len(missing))
        st.dataframe(missing, use_container_width=True)
    else:
        st.success("No missing-file records in the report.")

    st.caption(
        "Aggregated Transaction source does not include Amount. "
        "Unavailable values are not zero."
    )

st.sidebar.divider()
st.sidebar.caption("PhonePe Pulse · Analytics Project")
```

```python
from pathlib import Path
import numpy as np
import shutil

app_file = Path(
    r"C:\Users\HP\OneDrive\Documents\fita project"
    r"\pulse-main\pulse-main\phonepe_dashboard\app.py"
)

if not app_file.is_file():
    print("app.py file kidaikkala baby. File path check pannu.")
else:
    source = app_file.read_text(encoding="utf-8")

    backup_file = app_file.with_name("app_before_state_percentages.py")
    if not backup_file.exists():
        shutil.copy2(app_file, backup_file)

    start_marker = "# ---------- STATE ANALYSIS ----------"
    start = source.find(start_marker)

    # 3D section irundhaal adhai preserve pannuvom
    possible_end_markers = [
        "# ---------- 3D INSIGHTS ----------",
        "# ---------- DISTRICT ANALYSIS ----------"
    ]

    end_positions = [
        source.find(marker, start + len(start_marker))
        for marker in possible_end_markers
        if source.find(marker, start + len(start_marker)) != -1
    ]

    if start == -1 or not end_positions:
        print("State Analysis section markers kidaikkala. App file change pannala.")
    else:
        end = min(end_positions)

        new_state_section = r'''
# ---------- STATE ANALYSIS ----------
elif page == "State Analysis":
    st.title("State Analysis")
    st.write(
        "State contribution percentage, quarterly growth and transaction trends."
    )

    df = data["Map State"].copy()

    if df.empty:
        st.error("Map State master CSV unavailable.")
    else:
        df["Category"] = df["Category"].astype(str).str.title()

        for col in ["Year", "QuarterNumber", "Transactions",
                    "Amount", "RegisteredCount"]:
            if col in df.columns:
                df[col] = pd.to_numeric(df[col], errors="coerce")

        category = st.selectbox(
            "Category",
            ["Transaction", "Merchant", "User"],
            key="percentage_category"
        )

        category_df = df[df["Category"] == category].copy()

        if category == "Transaction":
            metric_options = ["Transactions", "Amount"]
        else:
            metric_options = ["RegisteredCount"]

        metric = st.selectbox(
            "Metric",
            metric_options,
            key="percentage_metric"
        )

        state_names = sorted(
            category_df["State"].dropna().astype(str).unique().tolist()
        )

        state_choice = st.selectbox(
            "Choose state for trend",
            ["All States"] + state_names,
            key="percentage_state"
        )

        years = sorted(
            category_df["Year"].dropna().astype(int).unique().tolist()
        )

        if not years:
            st.warning("Year data unavailable.")
        else:
            year_choice = st.selectbox(
                "Year",
                years,
                index=len(years) - 1,
                key="percentage_year"
            )

            available_quarters = sorted(
                category_df.loc[
                    category_df["Year"] == year_choice,
                    "QuarterNumber"
                ].dropna().astype(int).unique().tolist()
            )

            if not available_quarters:
                st.warning("Quarter data unavailable.")
            else:
                quarter_choice = st.selectbox(
                    "Quarter",
                    available_quarters,
                    index=len(available_quarters) - 1,
                    key="percentage_quarter"
                )

                # Each state's value for every available quarter
                state_periods = (
                    category_df.groupby(
                        ["State", "StateFolder", "Year", "QuarterNumber"],
                        as_index=False
                    )
                    .agg(Value=(metric, "sum"))
                    .sort_values(["StateFolder", "Year", "QuarterNumber"])
                )

                # Calculate each state's share of the state-level total
                period_totals = (
                    state_periods.groupby(
                        ["Year", "QuarterNumber"], as_index=False
                    )
                    .agg(PeriodTotal=("Value", "sum"))
                )

                state_periods = state_periods.merge(
                    period_totals,
                    on=["Year", "QuarterNumber"],
                    how="left"
                )

                state_periods["SharePct"] = np.where(
                    state_periods["PeriodTotal"] != 0,
                    state_periods["Value"]
                    / state_periods["PeriodTotal"] * 100,
                    np.nan
                )

                # Quarter-over-quarter growth by state
                state_periods["GrowthPct"] = (
                    state_periods.groupby("StateFolder")["Value"]
                    .pct_change(fill_method=None) * 100
                )

                # Current period state ranking
                current_states = state_periods[
                    (state_periods["Year"] == year_choice)
                    & (state_periods["QuarterNumber"] == quarter_choice)
                ].copy()

                # Find the previous calendar quarter
                if quarter_choice == 1:
                    previous_year = year_choice - 1
                    previous_quarter = 4
                else:
                    previous_year = year_choice
                    previous_quarter = quarter_choice - 1

                previous_values = state_periods[
                    (state_periods["Year"] == previous_year)
                    & (state_periods["QuarterNumber"] == previous_quarter)
                ][["StateFolder", "Value"]].rename(
                    columns={"Value": "PreviousValue"}
                )

                current_states = current_states.merge(
                    previous_values,
                    on="StateFolder",
                    how="left"
                )

                current_states["GrowthPct"] = np.where(
                    current_states["PreviousValue"].notna()
                    & (current_states["PreviousValue"] != 0),
                    (
                        current_states["Value"]
                        / current_states["PreviousValue"] - 1
                    ) * 100,
                    np.nan
                )

                current_states = current_states.sort_values(
                    "Value", ascending=False
                )

                # KPI cards
                current_total = current_states["Value"].sum()

                paired_states = current_states.dropna(
                    subset=["PreviousValue"]
                )

                previous_total = paired_states["PreviousValue"].sum()

                if state_choice == "All States":
                    selected_row = None
                    selected_value = current_total
                    selected_share = 100.0 if current_total != 0 else np.nan

                    overall_growth = (
                        (paired_states["Value"].sum() / previous_total - 1)
                        * 100
                        if previous_total != 0 and len(paired_states) > 0
                        else np.nan
                    )
                    selected_growth = overall_growth

                else:
                    matched_row = current_states[
                        current_states["State"] == state_choice
                    ]

                    selected_row = (
                        matched_row.iloc[0] if not matched_row.empty else None
                    )

                    selected_value = (
                        selected_row["Value"]
                        if selected_row is not None else np.nan
                    )

                    selected_share = (
                        selected_row["SharePct"]
                        if selected_row is not None else np.nan
                    )

                    selected_growth = (
                        selected_row["GrowthPct"]
                        if selected_row is not None else np.nan
                    )

                def readable_value(value):
                    if pd.isna(value):
                        return "N/A"
                    return f"{value:,.0f}"

                def readable_pct(value):
                    if pd.isna(value):
                        return "N/A"
                    return f"{value:+.2f}%"

                c1, c2, c3 = st.columns(3)

                c1.metric(
                    f"{metric} · {year_choice} Q{quarter_choice}",
                    readable_value(selected_value)
                )

                c2.metric(
                    "Share of State Total",
                    f"{selected_share:.2f}%"
                    if pd.notna(selected_share) else "N/A"
                )

                c3.metric(
                    "Growth vs Previous Quarter",
                    readable_pct(selected_growth)
                )

                st.caption(
                    "Share % = selected state value divided by the sum of "
                    "available state values for the selected period. "
                    "Growth % compares with the previous calendar quarter. "
                    "N/A means the previous-quarter baseline is unavailable."
                )

                # Trend chart with percentage information on hover
                st.subheader("Quarterly Trend with Percentages")

                if state_choice == "All States":
                    trend_data = (
                        state_periods.groupby(
                            ["Year", "QuarterNumber"], as_index=False
                        )
                        .agg(Value=("Value", "sum"))
                        .sort_values(["Year", "QuarterNumber"])
                    )

                    trend_data["GrowthPct"] = (
                        trend_data["Value"].pct_change(fill_method=None) * 100
                    )

                    trend_data["SharePct"] = 100.0

                    trend_data["Period"] = (
                        trend_data["Year"].astype(int).astype(str)
                        + " Q"
                        + trend_data["QuarterNumber"].astype(int).astype(str)
                    )

                    fig = px.line(
                        trend_data,
                        x="Period",
                        y="Value",
                        markers=True,
                        hover_data={
                            "GrowthPct": ":.2f",
                            "SharePct": ":.2f",
                            "Value": ":,.0f"
                        },
                        title=f"All-State {metric} and Quarterly Growth"
                    )

                else:
                    trend_data = state_periods[
                        state_periods["State"] == state_choice
                    ].copy()

                    trend_data["Period"] = (
                        trend_data["Year"].astype(int).astype(str)
                        + " Q"
                        + trend_data["QuarterNumber"].astype(int).astype(str)
                    )

                    fig = px.line(
                        trend_data,
                        x="Period",
                        y="Value",
                        markers=True,
                        hover_data={
                            "GrowthPct": ":.2f",
                            "SharePct": ":.2f",
                            "Value": ":,.0f"
                        },
                        title=f"{state_choice} · {metric} with Growth % and Share %"
                    )

                fig.update_traces(
                    line=dict(color="#C084FC", width=3),
                    marker=dict(size=7, color="#E879F9")
                )
                st.plotly_chart(chart_style(fig), use_container_width=True)

                # Top 10 states chart with percentage labels
                st.subheader(
                    f"State Contribution · {year_choice} Q{quarter_choice}"
                )

                top_states = current_states.head(10).copy()
                top_states["ShareLabel"] = top_states["SharePct"].map(
                    lambda x: f"{x:.1f}%" if pd.notna(x) else "N/A"
                )
                top_states = top_states.sort_values("Value")

                fig = px.bar(
                    top_states,
                    x="Value",
                    y="State",
                    orientation="h",
                    text="ShareLabel",
                    hover_data={
                        "Value": ":,.0f",
                        "SharePct": ":.2f",
                        "GrowthPct": ":.2f",
                        "PreviousValue": ":,.0f"
                    },
                    title="Top 10 States · Value with Share %"
                )

                fig.update_traces(
                    marker_color="#8B5CF6",
                    textposition="outside",
                    cliponaxis=False
                )
                fig.update_layout(margin=dict(r=70))
                st.plotly_chart(fig.update_layout(template="plotly_dark"), use_container_width=True)

                # All states percentage table
                st.subheader("Every State · Percentage Breakdown")

                table = current_states[
                    ["State", "Value", "SharePct", "PreviousValue", "GrowthPct"]
                ].copy()

                table = table.rename(columns={
                    "Value": metric,
                    "SharePct": "Share %",
                    "PreviousValue": "Previous Quarter Value",
                    "GrowthPct": "Growth %"
                })

                st.dataframe(
                    table,
                    column_config={
                        "Share %": st.column_config.NumberColumn(
                            "Share %", format="%.2f%%"
                        ),
                        "Growth %": st.column_config.NumberColumn(
                            "Growth %", format="%.2f%%"
                        )
                    },
                    hide_index=True,
                    use_container_width=True
                )
'''

        updated = source[:start] + new_state_section + "\n" + source[end:]

        try:
            compile(updated, str(app_file), "exec")
            app_file.write_text(updated, encoding="utf-8")

            print("State percentage analysis added successfully! ✅")
            print("Backup:", backup_file)
            print("Added: state share %, quarter growth %, chart labels and table.")
            

        except SyntaxError as error:
            print("Syntax check failed; original app file is unchanged.")
            print(error)
```

```python
from pathlib import Path
import re
import shutil

app_file = Path(
    r"C:\Users\HP\OneDrive\Documents\fita project"
    r"\pulse-main\pulse-main\phonepe_dashboard\app.py"
)

if not app_file.exists():
    print("app.py file not found. Check the dashboard folder.")
else:
    source = app_file.read_text(encoding="utf-8")

    # Backup existing app
    backup = app_file.with_name("app_before_overview_design.py")
    if not backup.exists():
        shutil.copy2(app_file, backup)

    updated = source

    # 1. Remove the 3D Insights menu item, if present
    updated = re.sub(
        r'["\']3D Insights["\']\s*,\s*',
        '',
        updated
    )
    updated = re.sub(
        r',\s*["\']3D Insights["\']',
        '',
        updated
    )

    # 2. Remove the separate 3D Insights page branch, if present
    branch = re.search(
        r'(?m)^elif page == ["\']3D Insights["\']:\s*$',
        updated
    )

    if branch:
        start = branch.start()

        # Include its heading comment when one exists
        comment = updated.rfind(
            "# ---------- 3D INSIGHTS",
            0,
            start
        )
        if comment >= 0:
            start = comment

        next_page = re.search(
            r'(?m)^elif page == ',
            updated[branch.end():]
        )

        if next_page:
            end = branch.end() + next_page.start()
            updated = updated[:start] + updated[end:]
        else:
            updated = updated[:start]

    # 3. Add a premium overview theme
    theme_marker = "OVERVIEW_ONLY_PREMIUM_THEME"

    if theme_marker not in updated:
        css = '''# OVERVIEW_ONLY_PREMIUM_THEME
st.markdown("""
<style>
[data-testid="stAppViewContainer"] {
    background:
        radial-gradient(ellipse at 10% 0%, rgba(124,58,237,.22), transparent 34%),
        radial-gradient(ellipse at 95% 10%, rgba(219,39,119,.15), transparent 30%),
        linear-gradient(145deg, #070B1B, #0B1028 60%, #090D20);
}
[data-testid="stSidebar"] {
    background: linear-gradient(180deg, #0A1230, #161035);
    border-right: 1px solid rgba(167,139,250,.35);
}
[data-testid="stMetric"] {
    background: linear-gradient(135deg, #21164D, #101B3B);
    border: 1px solid rgba(168,85,247,.7);
    border-radius: 18px;
    padding: 18px;
    box-shadow: 0 8px 26px rgba(0,0,0,.25);
}
[data-testid="stMetric"]:hover {
    border-color: #E879F9;
}
[data-testid="stMetricValue"] {
    color: #FFFFFF;
    font-weight: 800;
    overflow-wrap: anywhere;
}
[data-testid="stPlotlyChart"] {
    background: rgba(17,25,54,.75);
    border: 1px solid rgba(129,140,248,.4);
    border-radius: 20px;
    padding: 8px;
}
h1 {
    color: #FFFFFF !important;
    letter-spacing: -.04em;
}
.phonepe-hero {
    position: relative;
    overflow: hidden;
    min-height: 230px;
    padding: 28px;
    margin-bottom: 24px;
    border-radius: 24px;
    border: 1px solid rgba(192,132,252,.7);
    background:
        radial-gradient(ellipse at 90% 20%, rgba(236,72,153,.40), transparent 30%),
        radial-gradient(ellipse at 70% 115%, rgba(124,58,237,.60), transparent 42%),
        linear-gradient(110deg, #100936, #1D1057 55%, #0B1640);
    box-shadow: 0 16px 45px rgba(55,20,130,.25);
}
.phonepe-hero::after {
    content: "";
    position: absolute;
    width: 430px;
    height: 120px;
    right: 8%;
    bottom: -64px;
    border: 2px solid rgba(217,70,239,.65);
    border-radius: 50%;
    transform: rotate(-12deg);
    box-shadow: 0 0 28px rgba(168,85,247,.30);
    pointer-events: none;
}
.phonepe-hero .hero-kicker {
    color: #D8B4FE;
    font-weight: 800;
    letter-spacing: .12em;
}
.phonepe-hero .hero-title {
    color: #FFFFFF;
    font-size: clamp(2rem,4vw,3.3rem);
    font-weight: 900;
    letter-spacing: -.05em;
}
.phonepe-hero .hero-title span {
    background: linear-gradient(90deg,#C084FC,#E879F9,#F9A8D4);
    -webkit-background-clip: text;
    -webkit-text-fill-color: transparent;
}
.phonepe-hero .hero-subtitle {
    color: #E4DEFF;
    max-width: 680px;
    line-height: 1.6;
}
.phonepe-hero .hero-tags span {
    background: rgba(124,58,237,.28);
    border: 1px solid rgba(192,132,252,.6);
    border-radius: 999px;
    color: #F3E8FF;
}
.phonepe-hero .hero-art {
    border-color: rgba(217,70,239,.6);
    box-shadow: 0 0 42px rgba(139,92,246,.35);
}
.feature-card {
    background: linear-gradient(135deg,#171541,#101936);
    border: 1px solid rgba(139,92,246,.45);
    border-radius: 17px;
    padding: 18px;
}
@media(max-width:700px) {
    .phonepe-hero { padding: 20px; }
}
</style>
""", unsafe_allow_html=True)

'''

        # Insert theme before sidebar/navigation rendering
        insert_at = updated.find("st.sidebar.title")
        if insert_at == -1:
            insert_at = updated.find("page = st.sidebar.radio")
        if insert_at == -1:
            insert_at = updated.find('if page == "Overview":')

        if insert_at == -1:
            print("Navigation marker not found. App not updated.")
            updated = None
        else:
            updated = updated[:insert_at] + css + updated[insert_at:]

    # 4. Make sure Overview has a hero banner
    if updated is not None:
        overview_start = updated.find('if page == "Overview":')

        if overview_start == -1:
            print("Overview page marker not found. App not updated.")
            updated = None
        else:
            next_branch = re.search(
                r'(?m)^elif page == ',
                updated[overview_start + 1:]
            )

            overview_end = (
                overview_start + 1 + next_branch.start()
                if next_branch else len(updated)
            )

            overview = updated[overview_start:overview_end]

            if "phonepe-hero" not in overview:
                hero = '''if page == "Overview":
    st.markdown("""
    <div class="phonepe-hero">
      <div class="hero-kicker">✦ &nbsp; PULSE ANALYTICS STUDIO</div>
      <div class="hero-title">PhonePe <span>Pulse</span></div>
      <div class="hero-subtitle">
        A clearer view of transactions, merchants and users across India.
      </div>
      <div class="hero-tags">
        <span>AGGREGATED DATA</span>
        <span>MAP INTELLIGENCE</span>
        <span>TOP RANKINGS</span>
      </div>
      <div class="hero-art">
        <div class="hero-ring"></div>
        <div class="hero-ring2"></div>
        <div class="hero-symbol">₹</div>
        <div class="hero-art-label">INDIA · DIGITAL PAYMENTS</div>
      </div>
    </div>
    """, unsafe_allow_html=True)
'''

                # Remove old introductory heading to avoid duplicate title
                overview = re.sub(
                    r'(?m)^    st\.title\(["\']PhonePe Pulse Analytics["\']\)\s*\n',
                    '',
                    overview,
                    count=1
                )

                # Remove old one-line intro, if present
                overview = re.sub(
                    r'(?m)^    st\.write\("Explore transaction activity, registration trends and rankings\."\)\s*\n',
                    '',
                    overview,
                    count=1
                )

                # Replace the page opening with banner
                overview = overview.replace(
                    'if page == "Overview":',
                    hero.rstrip(),
                    1
                )

                updated = (
                    updated[:overview_start]
                    + overview
                    + updated[overview_end:]
                )

    # 5. Validate syntax, then save safely
    if updated is not None:
        try:
            compile(updated, str(app_file), "exec")
            app_file.write_text(updated, encoding="utf-8")

            print("Overview design updated successfully! ✅")
            print("3D Insights menu removed if present.")
            print("Overview hero and decorative background ready.")
            print("Backup:", backup)
            print("Next: browser tab-la Overview select panni Ctrl + R press pannu.")

        except SyntaxError as error:
            print("Syntax check failed; original app file remains unchanged.")
            print(error)
```

```python
from pathlib import Path
import pandas as pd

# Project and report locations
project = Path(
    r"C:\Users\HP\OneDrive\Documents\fita project"
    r"\pulse-main\pulse-main"
)

report_path = (
    project / "data" / "top" / "top_analysis_2018_2026"
    / "top_missing_json_files.csv"
)

if not report_path.exists():
    print("Missing-file report kandupidikka mudiyala:")
    print(report_path)

else:
    report = pd.read_csv(report_path)

    print("Report loaded:", len(report), "rows")
    print("Columns:", list(report.columns))

    # Focus only on Merchant state-level entries
    report = report[
        report["Category"].astype(str).str.lower().eq("merchant")
        & report["Scope"].astype(str).str.lower().eq("state")
    ].copy()

    if "File" not in report.columns:
        print("File path column illai. Columns-ai check pannanum.")

    else:
        def file_exists(value):
            if pd.isna(value):
                return False

            raw = str(value).strip().strip('"')
            path = Path(raw)

            # Check the reported path first
            if path.is_file():
                return True

            # If the report contains a path from another root,
            # try rebuilding it from this project's data folder.
            normalised = raw.replace("/", "\\")
            marker = "\\data\\"

            if marker in normalised.lower():
                parts = normalised.lower().split(marker, 1)
                suffix = normalised[len(parts[0]) + len(marker):]
                alternative = project / "data" / suffix

                if alternative.is_file():
                    return True

            return False

        report["Actually Exists"] = report["File"].apply(file_exists)

        found = int(report["Actually Exists"].sum())
        missing = len(report) - found

        print("\n--- LOCAL FILE CHECK ---")
        print("Merchant state entries:", len(report))
        print("Files found locally:", found)
        print("Files not found at reported paths:", missing)

        if missing == 0:
            print("\nAll reported files are present locally.")
            print("Missing report may be outdated.")
        elif found == 0:
            print("\nNone of the reported paths were found.")
            print("Next, verify the actual folder and path.")
        else:
            print("\nSome files exist and some are unavailable.")

        print("\n--- SAMPLE RESULTS ---")
        print(
            report[
                ["StateFolder", "Year", "Quarter",
                 "Actually Exists"]
            ].head(15).to_string(index=False)
        )
```

```python
from pathlib import Path
from urllib.request import Request, urlopen
from urllib.error import HTTPError, URLError
from urllib.parse import quote
import pandas as pd

project = Path(
    r"C:\Users\HP\OneDrive\Documents\fita project"
    r"\pulse-main\pulse-main"
)

report_path = (
    project / "data" / "top" / "top_analysis_2018_2026"
    / "top_missing_json_files.csv"
)

report = pd.read_csv(report_path)
report = report[
    report["Category"].astype(str).str.lower().eq("merchant")
    & report["Scope"].astype(str).str.lower().eq("state")
].copy()

results = []

for _, row in report.iterrows():
    state = str(row["StateFolder"]).strip()
    year = int(row["Year"])
    quarter = str(row["Quarter"]).strip().upper().replace("Q", "")

    relative_path = (
        f"data/top/merchant/country/india/state/"
        f"{state}/{year}/{quarter}.json"
    )

    url = (
        "https://raw.githubusercontent.com/PhonePe/pulse/main/"
        + quote(relative_path, safe="/.-_~")
    )

    request = Request(
        url,
        headers={"User-Agent": "PhonePe-Pulse-Source-Check"}
    )

    try:
        with urlopen(request, timeout=15) as response:
            status = response.status
    except HTTPError as e:
        status = e.code
    except (URLError, TimeoutError, OSError):
        status = "NETWORK_ERROR"

    results.append({
        "State": state,
        "Year": year,
        "Quarter": quarter,
        "HTTP_Status": status
    })

check = pd.DataFrame(results)

print("Official GitHub verification")
print("----------------------------")
print("Files checked:", len(check))
print("Available online:", (check["HTTP_Status"] == 200).sum())
print("Not found online:", (check["HTTP_Status"] == 404).sum())
print(
    "Network errors:",
    (check["HTTP_Status"] == "NETWORK_ERROR").sum()
)

print("\nDetailed results:")
print(check.to_string(index=False))
```

```python
from pathlib import Path
import pandas as pd
from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL
from getpass import getpass

# Project data folder
project = Path(
    r"C:\Users\HP\OneDrive\Documents\fita project"
    r"\pulse-main\pulse-main"
)
data = project / "data"

# CSV file path and MySQL table name
datasets = {
    "aggregated_transaction": (
        data / "aggregated/transaction/country/india/"
        "transaction_analysis_2018_2026/"
        "phonepe_india_master_2018_2026.csv"
    ),
    "aggregated_merchant_india": (
        data / "aggregated/merchant/country/india/"
        "merchant_analysis_2018_2026/"
        "merchant_india_master_2018_2026.csv"
    ),
    "aggregated_merchant_state": (
        data / "aggregated/merchant/country/india/"
        "merchant_state_analysis_2018_2026/"
        "merchant_state_master_2018_2026.csv"
    ),
    "aggregated_user_india": (
        data / "aggregated/user/country/india/"
        "user_india_analysis_2018_2026/"
        "user_india_master_2018_2026.csv"
    ),
    "aggregated_user_state": (
        data / "aggregated/user/country/india/"
        "user_state_analysis_2018_2026/"
        "user_state_master_2018_2026.csv"
    ),
    "map_state": (
        data / "map/map_analysis_2018_2026/"
        "map_state_master_all_categories_2018_2026.csv"
    ),
    "map_district": (
        data / "map/map_analysis_2018_2026/"
        "map_district_master_all_categories_2018_2026.csv"
    ),
    "top_master": (
        data / "top/top_analysis_2018_2026/"
        "top_master_all_2018_2026.csv"
    ),
    "transaction_validation": (
        data / "aggregated/transaction/country/india/"
        "transaction_analysis_2018_2026/"
        "aggregated_vs_map_transaction_validation.csv"
    ),
    "top_missing_files": (
        data / "top/top_analysis_2018_2026/"
        "top_missing_json_files.csv"
    ),
}

# Check all files before loading anything
missing_files = [
    str(path)
    for path in datasets.values()
    if not path.is_file()
]

if missing_files:
    print("Some CSV files were not found:")
    for path in missing_files:
        print(path)

else:
    # Reuse the existing connection if it is available
    if "engine" not in globals():
        username = input("MySQL username [root]: ").strip() or "root"
        password = getpass("MySQL password: ")

        url = URL.create(
            "mysql+pymysql",
            username=username,
            password=password,
            host="localhost",
            port=3306,
            database="phonepe_pulse",
        )
        engine = create_engine(url)

    try:
        with engine.connect() as connection:
            db_name = connection.execute(
                text("SELECT DATABASE()")
            ).scalar()

        if db_name != "phonepe_pulse":
            raise RuntimeError(
                f"Wrong database connected: {db_name}"
            )

        print("Connected to:", db_name)

        for table_name, csv_path in datasets.items():
            df = pd.read_csv(csv_path, low_memory=False)

            # Create the table and load its CSV records
            df.to_sql(
                name=table_name,
                con=engine,
                if_exists="replace",
                index=False,
                chunksize=1000,
            )

            print(
                f"Loaded {table_name}: "
                f"{len(df):,} rows"
            )

        # Verify that tables exist and contain the expected rows
        print("\n--- DATABASE VERIFICATION ---")

        with engine.connect() as connection:
            for table_name, csv_path in datasets.items():
                csv_rows = len(
                    pd.read_csv(csv_path, low_memory=False)
                )

                sql_rows = connection.execute(
                    text(
                        f"SELECT COUNT(*) FROM `{table_name}`"
                    )
                ).scalar_one()

                status = (
                    "PASS" if csv_rows == sql_rows
                    else "CHECK"
                )

                print(
                    f"{table_name}: "
                    f"CSV={csv_rows:,}, "
                    f"MySQL={sql_rows:,} [{status}]"
                )

    except Exception as error:
        print("Loading or verification failed.")
        print(type(error).__name__, error)
```
